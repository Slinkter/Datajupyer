## 1. Introducción

La visualización de datos es el arte de convertir números en imágenes. En la ciencia de datos, esto es crítico porque el cerebro humano identifica patrones, tendencias y anomalías mucho más rápido en un gráfico que en una tabla de miles de filas.

Pandas integra la librería `Matplotlib`, lo que nos permite crear visualizaciones directamente desde nuestros DataFrames sin necesidad de escribir código complejo.

````python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Carga de datos y preparación de la tabla pivote (repaso del notebook anterior)
DATA_PATH = Path('population_total.csv')
df_raw = pd.read_csv(DATA_PATH).dropna()
df_pivot = df_raw.pivot(index='year', columns='country', values='population')

# Filtramos los países principales para evitar saturar los gráficos
top_countries = ['United States', 'India', 'China', 'Indonesia', 'Brazil']
df_pivot_filtered = df_pivot[top_countries]
````

## 2. El Puente Pedagógico: ¿Qué gráfico elegir?

No todos los datos se cuentan igual. Elegir el gráfico incorrecto puede llevar a conclusiones erróneas.

#### Guía de Decisión Visual
```
   [ ¿Qué quieres analizar? ]
             |
             v
    +--------+--------+--------+--------+--------+
    |        |        |        |        |        |
 [Tiempo] [Comp.] [Prop.] [Distrib.] [Relación] [Frecuencia]
    |        |        |        |        |        |
    v        v        v        v        v        v
 LinePlot BarPlot PieChart HistPlot ScatterPlot BoxPlot
```

1. **LinePlot**: Ideal para series temporales (evolución en el tiempo).
2. **BarPlot**: Ideal para comparar magnitudes entre categorías.
3. **PieChart**: Ideal para ver la composición de un todo (partes de un 100%).
4. **HistPlot**: Ideal para ver cómo se distribuyen los valores de una variable.
5. **ScatterPlot**: Ideal para ver si existe una correlación entre dos variables numéricas.

### A. Gráfico de Líneas (`kind='line'`)
Utilizamos líneas cuando queremos ver la **tendencia**.

````python
df_pivot_filtered.plot(kind='line', 
                      figsize=(10, 6), 
                      title='Evolución de la Población (1955-2020)', 
                      xlabel='Año', 
                      ylabel='Población')
plt.grid(True)
plt.show()
````

### B. Gráfico de Barras (`kind='bar'`)
Ideal para comparar un momento específico en el tiempo.

````python
# Tomamos solo el año 2020 para la comparación
df_2020 = df_pivot_filtered.loc[[2020]].T

df_2020.plot(kind='bar', 
              color='skyblue', 
              figsize=(8, 5), 
              title='Población por País en 2020', 
              xlabel='País', 
              ylabel='Población')
plt.xticks(rotation=45)
plt.show()
````

### C. Gráfico de Pastel (`kind='pie'`)
Muestra la cuota de mercado o porcentaje del total.

````python
# Usamos los datos de 2020
df_2020[2020].plot(kind='pie', 
                   autopct='%1.1f%%', 
                   startangle=140, 
                   figsize=(7, 7), 
                   title='Distribución Poblacional en 2020')
plt.ylabel('') # Eliminar la etiqueta del eje Y para limpiar el gráfico
plt.show()
````

### D. Histogramas (`kind='hist'`)
Nos permite ver la densidad de los datos.

````python
df_pivot_filtered[['United States', 'India']].plot(kind='hist', 
                                                   bins=20, 
                                                   alpha=0.5, 
                                                   figsize=(8, 5), 
                                                   title='Distribución de Población: USA vs India')
plt.xlabel('Población')
plt.ylabel('Frecuencia')
plt.legend(['EE.UU.', 'India'])
plt.show()
````

### E. Gráfico de Dispersión (`kind='scatter'`)
Para analizar la correlación entre dos variables.

````python
# Regresamos al formato largo para el scatter plot
df_sample = df_raw[df_raw['country'].isin(top_countries)]

df_sample.plot(kind='scatter', 
               x='year', 
               y='population', 
               alpha=0.6, 
               figsize=(10, 6), 
               title='Relación Año vs Población', 
               color='forestgreen')
plt.show()
````

## 📋 Resumen

| Tipo de Gráfico | Argumento `kind` | Caso de Uso Principal |
|---|---|---|
| **Líneas** | `'line'` | Tendencias temporales, series de tiempo |
| **Barras** | `'bar'` | Comparación de categorías discretas |
| **Pastel** | `'pie'` | Proporciones y porcentajes del total |
| **Histograma** | `'hist'` | Distribución de frecuencias de una variable |
| **Dispersión** | `'scatter'` | Correlación entre dos variables numéricas |
