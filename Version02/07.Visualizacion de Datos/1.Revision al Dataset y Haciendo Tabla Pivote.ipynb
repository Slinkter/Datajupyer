# Revisión de Datos y Tablas Pivote con Pandas

## 🎯 Objetivos
- Cargar y explorar un conjunto de datos utilizando `pandas`.
- Limpiar datos mediante la eliminación de valores nulos.
- Transformar datos de formato largo (*long format*) a formato ancho (*wide format*) utilizando la función `.pivot()`.
- Preparar los datos para una visualización eficiente.

## 1. Introducción

En el análisis de datos, a menudo nos encontramos con datos en formato **long** (donde cada fila es una observación única). Aunque este formato es ideal para el almacenamiento y la limpieza, no siempre es el más intuitivo para el análisis humano o para crear gráficos de líneas donde queremos comparar categorías a lo largo del tiempo.

Aquí es donde entran las **Tablas Pivote**, que nos permiten reorganizar los datos para obtener una vista resumida y estructurada.

````python
import pandas as pd
from pathlib import Path

# Definir la ruta al archivo de datos
DATA_PATH = Path('population_total.csv')
````

### Carga y Exploración Inicial

Primero, cargamos el dataset y observamos su estructura.

````python
df_population_raw = pd.read_csv(DATA_PATH)
df_population_raw.head()
````

## 2. El Puente Pedagógico: ¿Por qué pivotar?

### Intuición
Imagina que tienes una lista de compras donde cada fila es un artículo y su precio. Si quieres comparar el precio del mismo artículo en diferentes tiendas, es más fácil si las tiendas son las columnas y los artículos son las filas.

### El Concepto de Formato
- **Long Format**: Cada fila es una instancia (País, Año, Población).
- **Wide Format**: Las categorías se convierten en columnas, facilitando la comparación directa.

#### Diagrama de Transformación
```
      [ Formato Largo ]                   [ Formato Ancho ]
    +-----------+------+-----------+     +------+-----------+-----------+
    | country   | year | population|     | year | Country A | Country B |
    +-----------+------+-----------+     +------+-----------+-----------+
    | China     | 2020 | 1.43e9    | --> | 2020 | 1.43e9    | 3.31e8    |
    | USA       | 2020 | 3.31e8    |     | 2019 | ...       | ...       |
    +-----------+------+-----------+     +------+-----------+-----------+
```

### Implementación de la Tabla Pivote

Para usar `.pivot()`, debemos asegurarnos de que no haya duplicados para la combinación de `index` y `columns` y que no existan valores nulos que puedan romper la estructura.

````python
# 1. Limpieza: Eliminar valores nulos para evitar errores en el pivoteo
df_cleaned = df_population_raw.dropna().copy()

# 2. Pivoteo:
# index: lo que queremos en las filas (tiempo)
# columns: lo que queremos como categorías (países)
# values: el dato numérico a mostrar (población)
df_pivot = df_cleaned.pivot(index='year', columns='country', values='population')

# 3. Filtrado: Seleccionamos solo los países de mayor interés para mayor claridad
top_countries = ['United States', 'India', 'China', 'Indonesia', 'Brazil']
df_pivot_filtered = df_pivot[top_countries]

df_pivot_filtered.head()
````

## 📋 Resumen

| Función | Propósito | Uso Clave |
|---|---|---|
| `pd.read_csv()` | Cargar datos desde archivo | `pd.read_csv('path.csv')` |
| `.dropna()` | Limpiar valores faltantes | `df.dropna(inplace=True)` |
| `.pivot()` | Cambiar formato Long $\rightarrow$ Wide | `df.pivot(index, columns, values)` |
| `df[[list]]` | Filtrar columnas específicas | `df[['col1', 'col2']]` |
