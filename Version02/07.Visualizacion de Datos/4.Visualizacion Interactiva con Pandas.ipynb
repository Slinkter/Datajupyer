## 1. Introducción

Hasta ahora hemos utilizado Matplotlib para crear gráficos estáticos. Estos son perfectos para publicaciones impresas o reportes en PDF. Sin embargo, durante el **Análisis Exploratorio de Datos (EDA)**, necesitamos herramientas que nos permitan "interrogar" a los datos.

**Cufflinks** es una librería que conecta Pandas con **Plotly**, permitiendo que el método `.iplot()` genere gráficos interactivos directamente desde el DataFrame.

````python
import pandas as pd
import cufflinks as cf
from IPython.display import display, HTML
from pathlib import Path

# Configuración de Cufflinks
# offline=True permite que los gráficos funcionen sin conexión a internet
cf.set_config_file(sharing='public', theme='white', offline=True)

print("Cufflinks configurado correctamente.")
````

### Preparación de los Datos

Cargamos y pivotamos los datos de población para que sean compatibles con las visualizaciones.

````python
DATA_PATH = Path('population_total.csv')
df_raw = pd.read_csv(DATA_PATH).dropna()
df_population = df_raw.pivot(index='year', columns='country', values='population')
df_population = df_population[['United States', 'India', 'China', 'Indonesia', 'Brazil']]
df_population.head()
````

## 2. El Puente Pedagógico: Estático vs Interactivos

### ¿Cuándo usar cada uno?
No se trata de que uno sea mejor que el otro, sino de para qué los necesitas.

#### Comparativa de Experiencia de Usuario
```
   [ Gráfico Estático (Matplotlib) ]       [ Gráfico Interactivo (Cufflinks) ]
   +------------------------------+       +---------------------------------+
   |        Line Chart           |       |          Line Chart              |
   |    (Imagen fija .png)       |       |    (Gráfico Dinámico .html)      |
   |                             |       |                                  |
   |  Cero interacción           |       |  - Zoom In/Out                   |
   |  Solo lectura               |       |  - Hover (Ver valor exacto)      |
   |  Ideal para PDF/Impresión   |       |  - Filtro de leyendas (clic)     |
   +------------------------------+       +---------------------------------+
```

### A. Gráfico de Líneas Interactivo (`kind='line'`)
Prueba a pasar el ratón sobre las líneas para ver los valores exactos por año.

````python
df_population.iplot(kind='line', 
                    xTitle='Años', 
                    yTitle='Población', 
                    title='Evolución Poblacional Interactiva (1955-2020)')
````

### B. Gráfico de Barras Interactivo (`kind='bar'`)
Ideal para comparar magnitudes rápidas.

````python
# Filtrar solo el año 2020
df_2020 = df_population.loc[[2020]].T

df_2020.iplot(kind='bar', 
              color='lightgreen', 
              xTitle='País', 
              yTitle='Población', 
              title='Población en 2020')
````

### C. Box Plots Interactivos (`kind='box'`)
Permite analizar la dispersión y detectar *outliers* (valores atípicos) fácilmente.

````python
df_population.iplot(kind='box', 
                    xTitle='Países', 
                    yTitle='Población')
````

### D. Histogramas Interactivos (`kind='hist'`)
Analizamos la distribución de la población para dos países.

````python
df_population[['United States', 'Indonesia']].iplot(kind='hist', 
                                                     xTitle='Población', 
                                                     title='Distribución de Frecuencia')
````

### E. Gráfico de Pastel Interactivo (`kind='pie'`)
Muestra la proporción relativa de cada país en el año 2020.

````python
# Preparar datos para el pie chart
df_pie = df_2020.reset_index().rename(columns={2020: 'poblacion'})

df_pie.iplot(kind='pie', 
             labels='index', 
             values='poblacion', 
             title='Porcentaje de Población en 2020')
````

### F. Scatter Plots Interactivos (`kind='scatter'`)
Útil para buscar correlaciones.

````python
df_population.iplot(kind='scatter', mode='markers', 
                    xTitle='Años', 
                    yTitle='Población', 
                    title='Dispersión de Población')
````

## 📋 Resumen

| Método | Tipo | Característica Principal | Uso Ideal |
