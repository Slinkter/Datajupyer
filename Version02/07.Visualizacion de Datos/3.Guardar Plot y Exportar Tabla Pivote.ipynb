# Exportación de Datos y Guardado de Visualizaciones

## 🎯 Objetivos
- Exportar un DataFrame procesado a un archivo de Excel (`.xlsx`).
- Guardar visualizaciones generadas por Matplotlib en archivos de imagen (`.png`).
- Comprender la diferencia entre formatos de almacenamiento de datos y formatos de presentación.

## 1. Introducción

En un flujo de trabajo real de Ciencia de Datos, el análisis no termina cuando vemos un gráfico en la pantalla del notebook. Los resultados deben ser exportados para: 
1. **Reportes**: Enviar tablas resumidas a stakeholders (Excel).
2. **Presentaciones**: Insertar gráficos de alta calidad en diapositivas o documentos (PNG/PDF).
3. **Interoperabilidad**: Guardar datos limpios para que otros procesos los utilicen (CSV).

````python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Preparación de datos (igual que en sesiones anteriores)
DATA_PATH = Path('population_total.csv')
df_raw = pd.read_csv(DATA_PATH).dropna()
df_pivot = df_raw.pivot(index='year', columns='country', values='population')
df_pivot_filtered = df_pivot[['United States', 'India', 'China', 'Indonesia', 'Brazil']]
````

## 2. Exportando Datos: De DataFrame a Excel

### El Puente Pedagógico: ¿CSV o Excel?

- **CSV (Comma Separated Values)**: Es el estándar universal. Es texto plano, ligero y compatible con cualquier lenguaje. Ideal para mover datos entre sistemas.
- **Excel (XLSX)**: Es el estándar empresarial. Permite múltiples hojas, formato de celdas y fórmulas. Ideal para usuarios finales.

#### Diagrama de Flujo de Datos
```
  [ DataFrame ] ----------> [ .to_csv() ] ----------> [ CSV File ]
        |                                              (Universal)
        |
        +-----------------> [ .to_excel() ] ---------> [ XLSX File ]
                                                       (Business)
```

````python
# Exportar la tabla pivote filtrada a un archivo Excel
# Nota: Requiere la librería 'openpyxl' instalada
df_pivot_filtered.to_excel('population_pivot_table.xlsx')

print("Archivo 'population_pivot_table.xlsx' exportado exitosamente.")
````

## 3. Guardando Visualizaciones

Cuando creamos un gráfico con `.plot()`, Matplotlib lo renderiza en el notebook. Para guardarlo en el disco, utilizamos la función `plt.savefig()`.

````python
# Crear el gráfico
ax = df_pivot_filtered.plot(kind='line', 
                           figsize=(10, 6), 
                           title='Crecimiento Poblacional (1955-2020)', 
                           xlabel='Año', 
                           ylabel='Población')
plt.grid(True)

# Guardar la figura antes de llamar a plt.show()
# dpi=300 asegura una resolución profesional para impresión
plt.savefig('population_trend_highres.png', dpi=300, bbox_inches='tight')

plt.show()
print("Gráfico guardado como 'population_trend_highres.png'")
````

#### Diagrama de Exportación de Imágenes
```
  [ Matplotlib Figure ] ---> [ plt.savefig() ] ------> [ PNG / PDF / JPG ]
                                                       (Reporting)
```

## 📋 Resumen

| Acción | Método | Formato Recomendado | Propósito |
|---|---|---|---|
| **Exportar Datos** | `df.to_csv()` | `.csv` | Intercambio universal de datos |
| **Exportar Datos** | `df.to_excel()` | `.xlsx` | Reportes ejecutivos y negocio |
| **Guardar Imagen** | `plt.savefig()` | `.png` | Presentaciones y web (raster) |
| **Guardar Imagen** | `plt.savefig()` | `.pdf` | Impresiones de alta calidad (vectorial) |
