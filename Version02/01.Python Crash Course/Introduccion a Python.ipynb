## 2. Tipos de Datos Básicos

💡 **Intuición**: Las computadoras no ven "números" o "texto" de la misma forma. Necesitan saber si un dato es una cantidad entera, un decimal o una cadena de caracteres para saber qué operaciones puede realizar con él.

🛠️ **Diagrama de Tipos de Datos**
```
Tipos Primitivos:
  ├── Numéricos: int (1, 2) ----> Operaciones aritméticas
  │              float (1.5) --> Precisión decimal
  ├── Booleanos: bool (True) ---> Lógica binaria
  └── Texto:     str ("Hola") --> Manipulación de caracteres
```

````python
print(f"Tipo de 1: {type(1)}")
print(f"Tipo de 2.3: {type(2.3)}")
print(f"Tipo de True: {type(True)}")
print(f"Tipo de 'Hola': {type('Hola')}")
````

### Operaciones Aritméticas

| Operación | Símbolo | Ejemplo | Resultado |
|-----------|---------|---------|-----------|
| Adición   | `+`     | `1 + 2` | `3`       |
| Sustracción| `-`     | `3 - 2` | `1`       |
| Multiplicación| `*` | `3 * 2` | `6`       |
| División  | `/`     | `5 / 2` | `2.5`     |
| Exponente | `**`    | `3 ** 2`| `9`       |
| Residuo   | `%`     | `5 % 2` | `1`       |
| Div. Entera| `//`   | `5 // 2`| `2`       |

````python
print(f"Suma: {1 + 2}")
print(f"Potencia: {3 ** 2}")
print(f"División Entera: {5 // 2}")
````

### Métodos de Cadenas (Strings)

Las cadenas en Python son objetos con métodos incorporados para su transformación.

````python
texto = "hello world"
print(f"Mayúsculas: {texto.upper()}")
print(f"Minúsculas: {texto.lower()}")
print(f"Título: {texto.title()}")
print(f"Contar 'l': {texto.count('l')}")
print(f"Reemplazar: {texto.replace('o', 'u')}")
````

## 3. Variables y Formateo de Cadenas

💡 **Intuición**: Una variable es como una etiqueta pegada a un valor en la memoria. En lugar de recordar la dirección de memoria, usamos el nombre de la variable.

🛠️ **Diagrama de Variable**
```
[ nombre_variable ] ----> [ Valor en Memoria ]
   "message"       ---->    "Hola Python"
```

````python
msg_1: str = "Estoy aprendiendo Python"
msg_2: str = "y es divertido!"

# Concatenación clásica
resultado_clásico = msg_1 + " " + msg_2

# f-strings (Estándar moderno Python 3.6+)
resultado_moderno = f"{msg_1} {msg_2}"

print(f"Clásico: {resultado_clásico}")
print(f"Moderno: {resultado_moderno}")
````

## 4. Listas: Colecciones Ordenadas

💡 **Intuición**: Una lista es como una fila de casilleros numerados. Podemos guardar cualquier cosa en ellos y acceder a ellos mediante su posición (índice). Recuerda que en Python, el conteo comienza en **0**.

🛠️ **Diagrama de Listas y Slicing**
```
Índices:   0       1       2       3
Lista: [ "USA", "India", "China", "Brazil" ]
        ^                       ^
     Inicio                   Final
       [--- Slice [0:3] ---] 
```

````python
countries: list[str] = ['United States', 'India', 'China', 'Brazil']

print(f"Primer elemento: {countries[0]}")
print(f"Último elemento: {countries[-1]}")
print(f"Slicing (0 a 3): {countries[0:3]}")
print(f"Desde índice 1 al final: {countries[1:]}")
````

### Operaciones y Métodos de Listas

````python
countries.append('Canada')
countries.insert(0, 'Argentina')
print(f"Lista actualizada: {countries}")

countries.remove('Canada')
eliminado = countries.pop(0)
print(f"Elemento pop: {eliminado}")
print(f"Lista final: {countries}")
````

### Ordenamiento y Copia

Es crucial entender la diferencia entre una **referencia** y una **copia**.

````python
numbers: list[int] = [4, 3, 10, 7, 1, 2]
numbers.sort()
print(f"Ascendente: {numbers}")
numbers.sort(reverse=True)
print(f"Descendente: {numbers}")

# Copia Correcta
copia_lista = numbers.copy()
copia_slice = numbers[:]
print(f"Copia: {copia_lista}")
````

## 5. Diccionarios: Mapeos Clave-Valor

💡 **Intuición**: A diferencia de las listas, los diccionarios funcionan como un directorio telefónico. No buscas por la posición 0 o 1, sino por una "clave" única (nombre) para obtener un "valor" (teléfono).

🛠️ **Diagrama de Diccionario**
```
  Clave (Key)   --->   Valor (Value)
  [ "nombre" ]  --->  [ "Frank" ]
  [ "edad"    ]  --->  [ 26      ]
```

````python
user_data: dict = {'nombre': 'Frank', 'edad': 26}

print(f"Nombre: {user_data['nombre']}")
print(f"Claves: {list(user_data.keys())}")
print(f"Valores: {list(user_data.values())}")
print(f"Items: {list(user_data.items())}")
````

### Manipulación de Diccionarios

````python
user_data['estatura'] = 1.7
user_data.update({'estatura': 1.9, 'lenguajes': ['Ingles', 'Español']})
print(f"Actualizado: {user_data}")

user_data.pop('estatura')
del user_data['lenguajes']
print(f"Simplificado: {user_data}")
````

## 6. Control de Flujo: Condicionales y Bucles

💡 **Intuición**: El flujo de un programa no siempre es lineal. A veces necesitamos tomar decisiones (If) o repetir una tarea muchas veces (For).

🛠️ **Diagrama de Decisión (IF)**
```
       [ Condición ]
       /           \
    (True)       (False)
      |             |
 [ Acción A ]  [ Acción B ]
```

````python
edad: int = 24

if edad >= 18:
    print("Eres mayor de edad")
elif edad >= 13:
    print("Eres un adolescente")
else:
    print("Eres un niño")
````

### Bucles For

El bucle `for` permite iterar sobre cualquier elemento iterable (listas, diccionarios, rangos).

````python
countries = ['United States', 'India', 'China', 'Brazil']

# Iteración simple
for pais in countries:
    print(f"País: {pais}")

# Iteración con índice (enumerate)
for i, pais in enumerate(countries):
    print(f"{i}: {pais}")

# Iteración de diccionarios
user_data = {'nombre': 'Frank', 'edad': 26}
for clave, valor in user_data.items():
    print(f"{clave} -> {valor}")
````

## 7. Funciones Personalizadas

💡 **Intuición**: Las funciones son "recetas" de código. En lugar de escribir los mismos pasos 10 veces, escribes la receta una vez y la "llamas" cuando la necesites.

🛠️ **Diagrama de Función**
```
[ Entradas (Args) ] --> [ Cuerpo de la Función ] --> [ Salida (Return) ]
```

````python
def sumar_numeros(a: int, b: int) -> int:
    """Suma dos números enteros y retorna el resultado."""
    return a + b

resultado = sumar_numeros(10, 5)
print(f"La suma es: {resultado}")
````

## 8. Funciones Integradas (Built-in)

💡 **Intuición**: Python viene con "herramientas preinstaladas" para tareas comunes, evitando que tengamos que reinventar la rueda.

````python
nums = [16, 83, 32, 1, 10]
print(f"Longitud: {len(nums)}")
print(f"Máximo: {max(nums)}")
print(f"Mínimo: {min(nums)}")
print(f"Redondeo: {round(2.333, 2)}")

# Range para bucles
for i in range(1, 10, 2):
    print(f"Número impar: {i}")
````

💡 **Intuición**: Python puede interactuar con el sistema operativo de tu computadora (crear carpetas, leer archivos) a través de módulos como `os`.

````python
import os

print(f"Directorio actual: {os.getcwd()}")
print(f"Contenido del directorio: {os.listdir()}")

# Crear carpeta de ejemplo
os.makedirs("Ejemplo_Opencode", exist_ok=True)
print(f"Carpeta creada. Contenido: {os.listdir()}")
````
