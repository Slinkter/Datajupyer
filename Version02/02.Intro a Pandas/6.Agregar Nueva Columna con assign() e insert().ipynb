# ➕ Expandiendo el DataFrame: Agregar Nuevas Columnas

🎯 **Objetivos**
- Implementar la creación de nuevas columnas utilizando el método `assign()`.
- Controlar la posición de las nuevas columnas mediante el método `insert()`.
- Diferenciar entre operaciones que devuelven una copia y operaciones que modifican el DataFrame original (in-place).

📖 **Introducción**

A medida que limpiamos o procesamos datos, es común necesitar crear nuevas variables (features) basadas en datos existentes o generar datos sintéticos para pruebas. Pandas ofrece diferentes caminos dependiendo de si queremos mantener la inmutabilidad de nuestros datos o modificar la tabla actual.

````python
import pandas as pd
import numpy as np

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. El Método `assign()`: Inmutabilidad y Flexibilidad

💡 **Intuición**: `assign()` es la herramienta ideal para el "method chaining". No modifica el DataFrame original, sino que crea una **copia** con las nuevas columnas añadidas. Es la forma más segura de agregar datos sin riesgo de alterar la fuente original accidentalmente.

🛠️ **Ventajas de `assign()`**
- ✅ Permite agregar múltiples columnas en una sola línea.
- ✅ Es ideal para pipelines de datos.
- ✅ Facilita la sobreescritura controlada de columnas.

````python
# 1. Generar datos aleatorios para simular nuevas métricas
np.random.seed(42) # Para reproducibilidad
score_1 = np.random.randint(1, 100, size=1000)
score_2 = np.random.randint(1, 100, size=1000)

# 2. Convertir a Series para asegurar la alineación de índices
series_1 = pd.Series(score_1)
series_2 = pd.Series(score_2)

# 3. Agregar múltiples columnas en una sola operación
df_updated = df_exams.assign(extra_score_1=series_1, extra_score_2=series_2)

df_updated.head()
````

## 2. El Método `insert()`: Control de Posición

💡 **Intuición**: A diferencia de `assign()`, que siempre añade las columnas al final, `insert()` nos permite colocar la nueva columna en cualquier posición específica (índice). 

⚠️ **Importante**: `insert()` modifica el DataFrame **original** (operación in-place). No devuelve un nuevo objeto, sino que altera el existente.

🛠️ **Sintaxis de `insert()`**
 `df.insert(loc, column, value)`
- `loc`: Posición entera (ej. 0 para el inicio, 1 para la segunda columna).
- `column`: Nombre de la nueva columna (string).
- `value`: Los datos a insertar (Series, lista o valor escalar).

````python
# Insertar una columna en la posición 1 (segunda columna)
df_exams.insert(1, 'test_score', series_1)

df_exams.head()
````

## 3. Comparativa: `assign()` vs `insert()`

💡 **Intuición**: Elegir la herramienta correcta depende de si priorizamos la estructura (posición) o la seguridad (copias).

🛠️ **Diagrama de flujo de decisión**
```
¿Necesito una posición específica?
   |
   +--> SÍ --> [ insert() ] --> Modifica Original (In-place)
   |
   +--> NO --> [ assign() ] --> Devuelve Copia Nueva
```

📋 **Resumen de Métodos**

| Método | Posición | Efecto en Original | Ventaja Principal |
|--------|----------|--------------------|-------------------|
| `assign()` | Siempre al final | Devuelve Copia | Seguridad e Inmutabilidad |
| `insert()` | Personalizable | Modifica In-place | Control de estructura |
