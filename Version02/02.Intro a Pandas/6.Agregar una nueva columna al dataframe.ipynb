# 🛠️ Métodos Rápidos para Agregar Columnas

🎯 **Objetivos**
- Implementar la asignación directa de columnas mediante la notación de corchetes `[]`.
- Aprender a agregar valores constantes y vectores de datos (arrays) a un DataFrame.
- Comprender el requisito de alineación de longitud entre el DataFrame y los datos insertados.

📖 **Introducción**

Aunque existen métodos sofisticados como `assign()` e `insert()`, la forma más común y rápida de añadir datos en un flujo de trabajo exploratorio es la asignación directa. Este método es intuitivo y similar a cómo añadiríamos una clave a un diccionario de Python.

````python
import pandas as pd
import numpy as np

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Asignación de Valores Constantes

💡 **Intuición**: Cuando queremos que todos los registros de una nueva columna tengan el mismo valor (por ejemplo, una etiqueta de categoría o un valor por defecto), simplemente asignamos el escalar directamente.

🛠️ **Diagrama de Valor Constante**
```
DataFrame                      Nueva Columna
+-------+-------+              +------------+
| Col A | Col B |  --->        | Col A|Col B| Valor C
+-------+-------+              +-------+-------+-------+
|  Val  |  Val  |              |  Val  |  Val  |  70    |
|  Val  |  Val  |              |  Val  |  Val  |  70    |
+-------+-------+              +-------+-------+-------+
```

````python
# Agregar una columna donde todos los estudiantes tienen la misma calificación inicial
df_exams['language score'] = 70

df_exams.head()
````

## 2. Asignación de Vectores de Datos (Arrays)

💡 **Intuición**: Para agregar datos únicos por fila, debemos proporcionar una estructura (como una lista o un array de NumPy) que tenga exactamente la misma cantidad de elementos que filas tiene el DataFrame.

⚠️ **La Regla de Oro de la Longitud**: Si el DataFrame tiene $N$ filas, el array insertador DEBE tener exactamente $N$ elementos. De lo contrario, Pandas lanzará un `ValueError`.

````python
# 1. Obtener la cantidad de filas del DataFrame
num_filas = len(df_exams)
print(f"El DataFrame tiene {num_filas} filas.")

# 2. Crear un array de la misma longitud (ej. 0 a 999)
language_score_array = np.arange(0, num_filas)

# 3. Asignar el array a una nueva columna
df_exams['language score'] = language_score_array

df_exams.head()
````

## 3. Comparativa: Asignación Directa vs Métodos Especializados

💡 **Intuición**: Saber cuándo usar `df['col'] = val` frente a `assign()` es clave para escribir código limpio.

🛠️ **Tabla de Decisión**

| Método | Efecto | Uso Ideal |
|---|---|---|
| **Asignación Directa `[]`** | Modifica el original (In-place) | Scripts rápidos y exploratorios |
| **Método `assign()`** | Devuelve copia (Inmutable) | Pipelines de datos y encadenamiento |
| **Método `insert()`** | Modifica el original | Cuando la posición de la columna es crítica |

📋 **Resumen**

La asignación directa `df['nueva_col'] = valores` es la forma más sencilla de expandir un DataFrame. Solo recuerda que los `valores` deben ser un escalar (un solo dato) o un vector con la misma longitud que el eje de las filas del DataFrame.
