# 🐼 Creando un DataFrame en Pandas

🎯 **Objetivos**
- Comprender qué es un DataFrame y su estructura básica.
- Crear DataFrames a partir de diversas fuentes de datos: Arrays, Listas, Diccionarios y Archivos CSV.
- Diferenciar entre índices y columnas.

📖 **Introducción**

Un DataFrame es la estructura de datos fundamental de Pandas. Podemos imaginarlo como una tabla de Excel o una tabla de SQL: una estructura bidimensional con filas y columnas donde cada columna puede tener un tipo de dato diferente.

## 1. Conceptos Básicos: Filas y Columnas

💡 **Intuición**: Un DataFrame es esencialmente un contenedor que organiza datos en una cuadrícula. 
- El **Índice (Index)** identifica cada fila (como la dirección de una casa).
- Las **Columnas (Columns)** identifican cada atributo del dato (como las características de la casa).

🛠️ **Diagrama de un DataFrame**
```
          [ Columnas (Features) ]
               Col A    Col B
Index  +-------+-------+-------+
Row 0  |  Val  |  Val  |  Val  |
Row 1  |  Val  |  Val  |  Val  |
       +-------+-------+-------+
```

````python
import pandas as pd
import numpy as np
````

## 2. Creación desde Arrays (NumPy)

💡 **Intuición**: NumPy es la base de Pandas. Crear un DataFrame desde un array es la forma más eficiente cuando ya tenemos datos numéricos estructurados.

````python
# 1. Definir los datos como un array de NumPy
data_np = np.array([[1, 4], [2, 5], [3, 6]])

# 2. Crear el DataFrame especificando índices y columnas
df_np = pd.DataFrame(
    data=data_np, 
    index=['fila_1', 'fila_2', 'fila_3'],
    columns=['valor_A', 'valor_B']
)

df_np
````

## 3. Creación desde Listas de Listas

💡 **Intuición**: Cuando no usamos NumPy, podemos pasar una lista de listas. Cada lista interna representa una fila de la tabla.

````python
# Datos en formato de lista anidada
data_list = [[1, 4], [2, 5], [3, 6]]

df_list = pd.DataFrame(
    data=data_list, 
    index=['fila_1', 'fila_2', 'fila_3'],
    columns=['valor_A', 'valor_B']
)

df_list
````

## 4. Creación desde Diccionarios

💡 **Intuición**: Esta es la forma más intuitiva de crear DataFrames manualmente. Las claves del diccionario se convierten automáticamente en los nombres de las columnas, y los valores (listas) en los datos de esas columnas.

🛠️ **Diagrama de Mapeo Diccionario $\rightarrow$ DataFrame**
```
{ "Estados": ["CA", "TX"], "Pob": [39M, 29M] }
       |             |
       v             v
  [ Col: Estados ] [ Col: Pob ]
  [ "CA"         ] [ 39M      ]
  [ "TX"         ] [ 29M      ]
```

````python
states = ["California", "Texas", "Florida", "New York"]
population = [39613493, 29730311, 21944577, 19299981]

# Almacenar en un diccionario
dict_states = {'Estado': states, 'Poblacion': population}

# El DataFrame infiere las columnas desde las llaves del diccionario
df_population = pd.DataFrame(dict_states)

df_population
````

## 5. Carga de Datos desde Archivos CSV

💡 **Intuición**: En el mundo real, los datos rara vez se escriben a mano. El formato CSV (Comma Separated Values) es el estándar para intercambiar datos tabulares. Pandas ofrece la función `read_csv()` para cargar estos archivos en segundos.

````python
# Leer el archivo CSV directamente a un DataFrame
df_exams = pd.read_csv('StudentsPerformance.csv')

# Mostrar las primeras 5 filas para verificar la carga
df_exams.head()
````

📋 **Resumen de Métodos de Creación**

| Fuente | Método | Ventaja |
|--------|--------|---------|
| **NumPy Array** | `pd.DataFrame(data)` | Máxima velocidad y eficiencia numérica |
| **Listas** | `pd.DataFrame(list)` | Sencillo para datos pequeños y rápidos |
| **Diccionario** | `pd.DataFrame(dict)` | Mapeo intuitivo de nombres de columna |
| **CSV** | `pd.read_csv(file)` | Estándar para cargar datos externos |
