# 📑 Los métodos `set_index()` y `sort_index()`

🎯 **Objetivos**
- Definir una columna existente como el nuevo índice del DataFrame.
- Restaurar el índice numérico por defecto mediante `reset_index()`.
- Ordenar el DataFrame basándose en las etiquetas del índice.

📖 **Introducción**

El índice es el "nombre" o "identificador" de cada fila en un DataFrame. Por defecto, Pandas asigna números enteros (0, 1, 2...). Sin embargo, en muchos casos es más útil usar una columna con valores únicos (como un ID de estudiante o un nombre) como índice para facilitar la búsqueda y organización de la información.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Definir un Nuevo Índice

💡 **Intuición**: Imagina que cada fila es un libro. Por defecto, los libros están numerados según llegaron a la biblioteca. `set_index()` nos permite re-organizarlos usando su título o su código ISBN como identificador principal.

🛠️ **Diagrama de `set_index()`**
```
Index | Col A | Col B      Index (Col A) | Col B
  0   |  ID1  |  Val          ID1       |  Val
  1   |  ID2  |  Val  ===>    ID2       |  Val
  2   |  ID3  |  Val          ID3       |  Val
```

````python
# 1. Mostrar índice actual (RangeIndex)
print(f"Índice actual: {df_exams.index}")

# 2. Crear un índice artificial para ejemplo (opcional)
df_exams['student_id'] = [f"S_{i}" for i in range(len(df_exams))]

# 3. Fijar la nueva columna como índice
df_exams.set_index('student_id', inplace=True)
df_exams.head()
````

## 2. Ordenar por Índice

💡 **Intuición**: Una vez que tenemos un índice significativo (como nombres o fechas), queremos que estén en orden alfabético o cronológico. `sort_index()` hace exactamente eso, ignorando los valores de las columnas y enfocándose solo en las etiquetas de las filas.

````python
# Ordenar por el ID del estudiante (descendente)
df_exams.sort_index(ascending=False).head()
````

## 3. Restaurar el Índice Original

💡 **Intuición**: Si cometemos un error o simplemente queremos volver al estado inicial de numeración 0, 1, 2..., usamos `reset_index()`. Por defecto, la columna que era índice vuelve a ser una columna normal del DataFrame.

````python
# Volver al índice numérico
df_exams.reset_index(inplace=True)
df_exams.head()
````

📋 **Resumen de Métodos de Índice**

| Método | Propósito | Observación |
|--------|-----------|-------------|
| `.set_index('col')` | Convierte una columna en el índice | La columna desaparece del cuerpo del DF |
| `.reset_index()` | Vuelve al índice numérico (0, 1, 2...) | El índice anterior vuelve a ser columna |
| `.sort_index()` | Ordena filas según sus etiquetas | Útil para índices de tiempo o nombres |
