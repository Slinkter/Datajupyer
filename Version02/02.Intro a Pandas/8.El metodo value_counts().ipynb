# 📊 El método `value_counts()`

🎯 **Objetivos**
- Contar la frecuencia de valores únicos en una columna categórica.
- Obtener frecuencias relativas (porcentajes) mediante la normalización.
- Comprender la diferencia entre contar el total de filas y contar por categorías.

📖 **Introducción**

En el análisis de datos, a menudo trabajamos con variables categóricas (como género, país, nivel educativo). El método `value_counts()` es una de las herramientas más potentes de Pandas para entender la distribución de estas categorías, permitiéndonos ver de un vistazo qué valores son los más comunes y cuáles los más raros.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Conteo de Categorías

💡 **Intuición**: Mientras que `.count()` nos dice cuántas celdas tienen datos, `.value_counts()` desglosa esos datos por cada etiqueta única que encuentra en la columna.

🛠️ **Diagrama de `value_counts()`**
```
Columna: [A, B, A, C, A, B]
             |
             v
Resultado:  A: 3
            B: 2
            C: 1
```

````python
# 1. Contar el total de registros (no desglosado)
total_generos = df_exams['gender'].count()
print(f"Total de registros en 'gender': {total_generos}")

# 2. Desglosar por categoría (conteo real)
conteo_generos = df_exams['gender'].value_counts()
print("\nConteo por género:")
print(conteo_generos)
````

## 2. Frecuencia Relativa (Normalización)

💡 **Intuición**: A veces, los números absolutos (ej. 518 mujeres) son difíciles de interpretar sin contexto. La normalización convierte estos números en proporciones (ej. 0.51 o 51%), facilitando la comparación entre diferentes datasets.

🛠️ **Diagrama de Normalización**
```
[ Valor Categoría ] / [ Suma de Todos los Valores ] = Proporción
      518         /          1000            =   0.518
```

````python
# Obtener porcentajes (proporciones de 0 a 1)
frecuencia_relativa = df_exams['gender'].value_counts(normalize=True)
print("Frecuencia relativa por género:")
print(frecuencia_relativa)

# Ejemplo avanzado: Nivel educativo con redondeo
edu_pct = df_exams['parental level of education'].value_counts(normalize=True).round(2)
print("\nDistribución porcentual de educación parental:")
print(edu_pct)
````

📋 **Resumen de Métodos de Conteo**

| Método / Parámetro | Resultado | Uso Principal |
|---------------------|-----------|---------------|
| `.count()`          | Número entero | Saber cuántos datos no nulos hay |
| `.value_counts()`   | Serie (Categoría: Conteo) | Entender la distribución de datos |
| `normalize=True`    | Serie (Categoría: %) | Comparar proporciones y pesos |
| `.round(n)`         | Serie redondeada | Mejorar la legibilidad de porcentajes |
