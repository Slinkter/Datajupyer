# 🔢 Ordenamiento con `sort_values()`

🎯 **Objetivos**
- Ordenar un DataFrame basándose en los valores de una sola columna.
- Realizar ordenamientos multinivel utilizando múltiples columnas.
- Controlar la dirección del orden (ascendente vs descendente).
- Entender el impacto del parámetro `inplace` en la persistencia de datos.

📖 **Introducción**

El ordenamiento es una de las tareas más comunes en el análisis de datos. Ya sea que necesitemos identificar a los mejores estudiantes, los productos más vendidos o las fechas más recientes, el método `sort_values()` es la herramienta fundamental en Pandas para organizar nuestra información de manera lógica y jerárquica.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Ordenamiento Simple

💡 **Intuición**: Ordenar por una columna es como organizar una fila de personas por su altura. Podemos decidir si queremos al más bajo primero (ascendente) o al más alto primero (descendente).

🛠️ **Diagrama de Ordenamiento**
```
[ Datos Desordenados ]
       |
       v
[ 50 ]  (A)
[ 10 ]  (B) --> [ sort_values ] --> [ 10 ] (B)
[ 30 ]  (C)                         [ 30 ] (C)
                                    [ 50 ] (A)
```

````python
# 1. Orden ascendente (por defecto)
print("Primeras 5 filas ordenadas por puntaje de matemáticas (asc):")
df_exams.sort_values('math score').head()

# 2. Orden descendente
print("\nPrimeras 5 filas ordenadas por puntaje de matemáticas (desc):")
df_exams.sort_values('math score', ascending=False).head()
````

## 2. Ordenamiento Multinivel

💡 **Intuición**: ¿Qué pasa si dos estudiantes tienen la misma nota en matemáticas? Podemos definir un "criterio de desempate" usando una segunda columna. Es como ordenar un directorio por Apellido y luego por Nombre.

🛠️ **Diagrama de Multinivel**
```
Nivel 1: [ math score ] (Desc)
   |-- Nivel 2: [ reading score ] (Desc)
```

````python
# Ordenar por matemáticas descendentemente y luego por lectura descendentemente
df_ordenado = df_exams.sort_values(['math score', 'reading score'], ascending=False)
df_ordenado.head(10)
````

## 3. Persistencia y Transformaciones Avanzadas

💡 **Intuición**: Por defecto, Pandas nos devuelve una *copia* ordenada del DataFrame sin modificar el original. Si queremos que el cambio sea permanente en nuestra variable actual, usamos `inplace=True`.

````python
# 1. Usar inplace para modificar el DataFrame original
df_exams.sort_values(['math score', 'reading score'], ascending=False, inplace=True)

# 2. Uso de funciones 'key' para ordenamientos personalizados (ej. ignorar mayúsculas)
df_exams.sort_values('race/ethnicity', ascending=True, key=lambda col: col.str.lower()).head()
````

📋 **Resumen de Ordenamiento**

| Parámetro | Tipo | Propósito |
|-----------|------|-----------|
| `by`      | str o list | Columna(s) sobre las que se ordena |
| `ascending`| bool o list | `True` para menor a mayor, `False` para mayor a menor |
| `inplace`  | bool | Si es `True`, modifica el DataFrame original |
| `key`      | function | Función aplicada a los valores antes de ordenar |
