# 🎯 Selección de Columnas Únicas: El Objeto Series

🎯 **Objetivos**
- Dominar las dos sintaxis principales para seleccionar una sola columna.
- Entender la diferencia entre un `DataFrame` y una `Series`.
- Identificar las limitaciones de la notación de punto (dot notation).

📖 **Introducción**

Cuando extraemos una sola columna de un DataFrame, Pandas no nos devuelve un "DataFrame más pequeño", sino un objeto diferente llamado **Series**. Una Series es esencialmente una sola columna de datos junto con su índice.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Notación de Corchetes `[]` (Sintaxis Recomendada)

💡 **Intuición**: Esta es la forma más robusta de seleccionar datos. Funciona exactamente como la indexación de un diccionario de Python: pasamos la clave (el nombre de la columna) y obtenemos el valor (la columna completa).

🛠️ **Diagrama de Selección `[]`**
```
DataFrame [ "Columna" ]  ---> [ Series ]
       |
       +---(Busca nombre)-------->+
```

````python
# Selección de la columna 'gender'
gender_col = df_exams['gender']
gender_col
````

## 2. El Objeto Series

💡 **Intuición**: Una Series es el "átomo" de Pandas. Es una estructura unidimensional. Al igual que los DataFrames, las Series tienen sus propios atributos (`.index`) y métodos (`.head()`).

````python
# Verificamos que el resultado es una Series
print(f"Tipo de dato: {type(df_exams['gender'])}")

# Acceder a atributos y métodos de la Series
print(f"Índice de la series: {df_exams['gender'].index}")
df_exams['gender'].head()
````

## 3. Notación de Punto `.` (Atajo)

💡 **Intuición**: Pandas permite acceder a las columnas como si fueran atributos del objeto. Es más rápido de escribir, pero tiene reglas estrictas.

🛠️ **Limitaciones de la Notación de Punto**
- ❌ **Espacios**: Si el nombre de la columna tiene espacios (ej. `"math score"`), la notación de punto fallará ya que Python interpretaría el espacio como el fin de la instrucción.
- ❌ **Conflictos**: Si el nombre de la columna coincide con un método de Pandas (ej. `"head"`), el método tendrá prioridad sobre la columna.

````python
# Selección exitosa (nombre sin espacios)
df_exams.gender
````

````python
# INTENTO FALLIDO: Notación de punto con espacios
try:
    df_exams.math score
except SyntaxError as e:
    print(f"Error detected: {e}")

# SOLUCIÓN: Usar corchetes para nombres con espacios
df_exams['math score']
````

📋 **Resumen: Corchetes vs Punto**

| Característica | Notación `[]` | Notación `.` |
|----------------|---------------|--------------|
| **Soporta Espacios** | ✅ Sí | ❌ No |
| **Soporta Caracteres Especiales** | ✅ Sí | ❌ No |
| **Evita Conflictos con Métodos** | ✅ Sí | ❌ No |
| **Velocidad de Escritura** | Media | Alta |
| **Recomendación** | **Uso General** | **Exploración Rápida** |
