📖 **Introducción**

Una de las mayores fortalezas de Pandas es su capacidad para realizar cálculos complejos sobre grandes volúmenes de datos de forma extremadamente rápida. En lugar de iterar fila por fila (como en una lista de Python), Pandas utiliza **operaciones vectorizadas**, lo que significa que aplica una operación a toda una columna o DataFrame simultáneamente.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Operaciones Estadísticas en Columnas

💡 **Intuición**: Cuando queremos saber la "tendencia" de una columna (¿Cuál es la nota más alta?, ¿Cuál es el promedio?), usamos métodos de agregación. Estos métodos toman muchos valores y devuelven uno solo.

🛠️ **Diagrama de Agregación**
```
[ 10 ]
[ 20 ] --> [ Método (sum, mean, max) ] --> [ Resultado Único ]
[ 30 ]
```

````python
# Operaciones individuales
print(f"Suma de puntajes de matemáticas: {df_exams['math score'].sum()}")
print(f"Puntaje promedio de lectura: {df_exams['reading score'].mean():.2f}")
print(f"Puntaje máximo de escritura: {df_exams['writing score'].max()}")
print(f"Puntaje mínimo de matemáticas: {df_exams['math score'].min()}")
print(f"Desviación estándar (lectura): {df_exams['reading score'].std():.2f}")
print(f"Conteo de elementos: {df_exams['math score'].count()}")
````

## 2. Resumen Estadístico: El Método `describe()`

💡 **Intuición**: En lugar de llamar a cada función por separado, `describe()` genera un reporte completo de las métricas más importantes para todas las columnas numéricas del DataFrame. Es la "navaja suiza" de la exploración de datos inicial.

````python
# Genera estadísticas descriptivas
df_exams.describe()
````

## 3. Operaciones Aritméticas entre Columnas

💡 **Intuición**: Pandas permite sumar, restar, multiplicar o dividir columnas completas como si fueran variables simples. Esto es fundamental para crear indicadores (KPIs) o normalizar datos.

🛠️ **Diagrama de Operación Vectorizada**
```
Col A + Col B = Nueva Col
[ 1 ] + [ 10 ] = [ 11 ]
[ 2 ] + [ 20 ] = [ 22 ]
[ 3 ] + [ 30 ] = [ 33 ]
```

````python
# 1. Sumar columnas para obtener un puntaje total
df_exams['total score'] = df_exams['math score'] + df_exams['reading score'] + df_exams['writing score']

# 2. Calcular el promedio de los tres exámenes
df_exams['average'] = df_exams['total score'] / 3

# Visualizar los resultados
df_exams[['math score', 'reading score', 'writing score', 'total score', 'average']].head()
````

📋 **Resumen de Operaciones**

| Método / Operación | Propósito | Salida |
|---------------------|-----------|--------|
| `.sum()`, `.mean()` | Agregación estadística | Escalar (un solo número) |
| `.max()`, `.min()`  | Valores extremos | Escalar |
| `.describe()`       | Resumen multivariado | DataFrame estadístico |
| `+`, `-`, `*`, `/`  | Aritmética vectorizada | Serie (columna nueva) |
