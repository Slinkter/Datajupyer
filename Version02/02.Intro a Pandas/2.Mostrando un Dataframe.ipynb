# 🔍 Visualizando y Explorando DataFrames

🎯 **Objetivos**
- Aprender a visualizar rápidamente los datos de un DataFrame.
- Utilizar los métodos `head()` y `tail()` para inspeccionar los extremos del dataset.
- Comprender la dimensión de los datos mediante el atributo `shape`.

📖 **Introducción**

Cuando trabajamos con miles o millones de filas, es imposible (y computacionalmente costoso) imprimir todo el conjunto de datos. Pandas proporciona herramientas optimizadas para obtener "muestras" representativas del DataFrame sin cargar toda la información en la consola.

````python
import pandas as pd

# Carga del dataset de ejemplo
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Inspección Rápida: El Método `head()`

💡 **Intuición**: `head()` (cabeza) nos permite ver las primeras filas del DataFrame. Es el primer paso fundamental después de cargar cualquier dataset para verificar que los datos se han importado correctamente y entender la estructura de las columnas.

🛠️ **Diagrama de `head()`**
```
[ Dataset Completo ]
       |
       v
[ Filas 0 a N ] <--- head(N)
       |
[ Resto de Datos ]
```

````python
# Por defecto muestra las primeras 5 filas
df_exams.head()
````

## 2. Inspección Final: El Método `tail()`

💡 **Intuición**: `tail()` (cola) hace lo opuesto a `head()`. Nos muestra las últimas filas. Esto es extremadamente útil para verificar el número total de registros o detectar si hay filas vacías o corruptas al final del archivo.

🛠️ **Diagrama de `tail()`**
```
[ Dataset Completo ]
       |
       v
[ Mayoría de Datos ]
       |
[ Filas Finales ] <--- tail(N)
```

````python
# Muestra las últimas 5 filas por defecto
df_exams.tail()

# Podemos especificar el número de filas
df_exams.tail(10)
````

## 3. Dimensiones del Dataset: El Atributo `shape`

💡 **Intuición**: Antes de realizar cualquier análisis, necesitamos saber el tamaño de nuestra "tabla". El atributo `shape` nos devuelve una tupla con el número de filas y columnas. A diferencia de `head()`, `shape` no es una función (no lleva paréntesis) porque es una propiedad intrínseca del objeto DataFrame.

🛠️ **Diagrama de `shape`**
```
DataFrame:
  +---+---+---+
  | v | v | v |  Row 0
  +---+---+---+
  | v | v | v |  Row 1
  +---+---+---+
  ^     ^     ^
Col 0  Col 1  Col 2

shape --> (num_rows, num_cols)
```

````python
# Retorna una tupla: (filas, columnas)
dimensiones = df_exams.shape
print(f"Dimensiones del dataset: {dimensiones}")
print(f"Número de filas: {dimensiones[0]}")
print(f"Número de columnas: {dimensiones[1]}")
````

📋 **Resumen de Visualización**

| Herramienta | Tipo | Salida | Uso Principal |
|-------------|------|--------|---------------|
| `head(n)`   | Método | DataFrame | Ver inicio del dataset |
| `tail(n)`   | Método | DataFrame | Ver final del dataset |
| `shape`     | Atributo | Tupla (int, int) | Conocer el tamaño total |
