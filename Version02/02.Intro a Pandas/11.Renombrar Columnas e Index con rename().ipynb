# 🏷️ Renombrar Columnas e Índices

🎯 **Objetivos**
- Cambiar el nombre de columnas específicas utilizando diccionarios de mapeo.
- Modificar las etiquetas del índice mediante el método `rename()`.
- Aplicar cambios de forma permanente usando el parámetro `inplace`.

📖 **Introducción**

En el análisis de datos real, los nombres de las columnas suelen venir con espacios, caracteres especiales o nombres poco descriptivos (ej. `var_1`, `col_2`). Pandas nos ofrece el método `rename()`, una forma elegante y flexible de renombrar nuestras etiquetas sin necesidad de reescribir todo el DataFrame.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Renombrar Columnas

💡 **Intuición**: Piensa en `rename()` como un buscador de "buscar y reemplazar". Le pasamos un diccionario donde la clave es el nombre antiguo y el valor es el nombre nuevo.

🛠️ **Diagrama de Mapeo de Columnas**
```
Diccionario: { 'viejo': 'nuevo' }
       |
       v
[ Col: gender ] ----> [ Col: Gender ]
[ Col: math score ] ----> [ Col: MS ]
```

````python
# 1. Renombrar una sola columna (asignando a una nueva variable)
df_exams = df_exams.rename(columns={'gender': 'Gender'})

# 2. Renombrar múltiples columnas (usando inplace para modificar el original)
df_exams.rename(
    columns={
        'math score': 'MS', 
        'reading score': 'RS', 
        'writing score': 'WS'
    }, 
    inplace=True
)

df_exams.head()
````

## 2. Renombrar el Índice

💡 **Intuición**: Al igual que las columnas, las etiquetas de las filas (el índice) también pueden renombrarse. Esto es útil cuando queremos asignar nombres descriptivos a registros específicos o cuando hemos cambiado el índice a algo no numérico.

````python
# Renombrar las etiquetas de las primeras 3 filas
df_exams.rename(index={0: 'A', 1: 'B', 2: 'C'}, inplace=True)

df_exams.head()
````

📋 **Resumen de Renombramiento**

| Parámetro | Tipo de Dato | Propósito |
|-----------|--------------|-----------|
| `columns` | dict         | Mapeo de nombres de columna `{antiguo: nuevo}` |
| `index`   | dict         | Mapeo de etiquetas de fila `{antigua: nueva}` |
| `inplace` | bool         | Si es `True`, modifica el objeto actual sin devolver copia |
| `errors`  | str          | `'ignore'` o `'raise'` si la clave no existe |
