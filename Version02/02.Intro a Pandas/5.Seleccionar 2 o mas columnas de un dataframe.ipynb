# 📑 Selección de Múltiples Columnas: Sub-DataFrames

🎯 **Objetivos**
- Aprender la sintaxis de "doble corchete" `[[]]` para seleccionar varias columnas.
- Diferenciar la salida de una selección única (Series) frente a una selección múltiple (DataFrame).
- Comprender por qué la notación de punto no es viable para selecciones múltiples.

📖 **Introducción**

En el análisis de datos, es común querer aislar un subconjunto de variables para compararlas o procesarlas. Cuando seleccionamos más de una columna, Pandas nos devuelve un **nuevo DataFrame** que contiene únicamente esas columnas, manteniendo la estructura original de filas e índices.

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. La Sintaxis de Doble Corchete `[[]]`

💡 **Intuición**: Para seleccionar varias columnas, pasamos una **lista de nombres de columnas** dentro de los corchetes de indexación del DataFrame. Esto resulta en la sintaxis `df[['Col1', 'Col2']]`.

🛠️ **Diagrama de Selección Múltiple**
```
[ Lista de Columnas ] --> [ DataFrame ] --> [ Sub-DataFrame ]
  ["Col A", "Col B"]         (Original)       (Reducido)
```

````python
# Seleccionamos dos columnas pasando una lista
df_sub = df_exams[['gender', 'math score']]
df_sub.head()
````

## 2. DataFrame vs Series: El Resultado

💡 **Intuición**: Es fundamental notar el cambio de tipo de objeto según la cantidad de columnas seleccionadas.
- **Una columna `df['col']`** $\rightarrow$ Retorna una `Series`.
- **Dos o más columnas `df[['col1', 'col2']]`** $\rightarrow$ Retorna un `DataFrame`.

````python
# Selección única
print(f"Tipo (1 col): {type(df_exams['gender'])}")

# Selección múltiple
print(f"Tipo (2+ cols): {type(df_exams[['gender', 'math score']])}")
````

## 3. Limitaciones de la Notación de Punto

💡 **Intuición**: La notación de punto (`df.columna`) está diseñada estrictamente para el acceso a un solo atributo. No existe una forma de concatenar atributos mediante puntos para obtener múltiples columnas en un solo objeto DataFrame.

````python
# Intento fallido: la notación de punto no soporta listas de columnas
try:
    # Esto causará un error de sintaxis
    df_exams.gender, df_exams['math score']
except Exception as e:
    print(f"Error: {e}")

print("Para selecciones múltiples, SIEMPRE utiliza la notación de corchetes dobles [[]].")
````

📋 **Resumen de Salidas**

| Sintaxis de Selección | Objeto Resultante | Descripción |
|-----------------------|-------------------|-------------|
| `df['col']` | `pd.Series` | Un solo vector de datos |
| `df[['col']]` | `pd.DataFrame` | Un DataFrame de una sola columna |
| `df[['c1', 'c2']]` | `pd.DataFrame` | Un sub-conjunto tabular de datos |
