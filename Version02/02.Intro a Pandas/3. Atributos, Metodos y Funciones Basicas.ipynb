# 🛠️ Atributos, Métodos y Funciones Básicas de Pandas

🎯 **Objetivos**
- Diferenciar entre Atributos y Métodos en un DataFrame.
- Utilizar atributos para obtener metadatos del dataset (forma, índices, tipos).
- Aplicar métodos de exploración estadística y resumen de datos.

📖 **Introducción**

Para analizar datos eficientemente, no necesitamos mirar cada celda, sino entender la "arquitectura" del DataFrame. Pandas nos permite acceder a esta arquitectura mediante **Atributos** (propiedades estáticas) y **Métodos** (acciones que procesan los datos).

````python
import pandas as pd

# Carga del dataset
df_exams = pd.read_csv('StudentsPerformance.csv')
````

## 1. Atributos: La Arquitectura del Dato

💡 **Intuición**: Un atributo es como una etiqueta de información que ya viene con el objeto. No procesa los datos, solo te dice "qué es" el objeto. Los atributos **no llevan paréntesis** `()` al final.

🛠️ **Diagrama de Atributos**
```
DataFrame Object
       |
       +--> .shape    --> (Filas, Columnas)
       +--> .index    --> Rango de índices
       +--> .columns  --> Nombres de columnas
       +--> .dtypes    --> Tipo de dato por columna
```

````python
# Dimensiones del DataFrame
print(f"Shape: {df_exams.shape}")

# Información sobre los índices (filas)
print(f"Index: {df_exams.index}")

# Nombres de las columnas
print(f"Columns: {df_exams.columns}")

# Tipos de datos de cada columna
print("Dtypes:\n", df_exams.dtypes)
````

## 2. Métodos de Exploración: Acciones sobre el Dato

💡 **Intuición**: Un método es una función que pertenece al DataFrame. A diferencia del atributo, el método **realiza una operación** sobre los datos para devolver un resultado. Los métodos **siempre llevan paréntesis** `()`.

🛠️ **Diagrama de Métodos**
```
[ DataFrame ] --(.info())--> [ Resumen Técnico (Memoria, Nulos) ]
       |
       +--(.describe())--> [ Resumen Estadístico (Media, Std, etc.) ]
       |
       +--(.head())    --> [ Muestra de Filas ]
```

````python
# Resumen técnico: tipos, nulos y uso de memoria
df_exams.info()

# Resumen estadístico: media, desviación estándar, cuartiles
df_exams.describe()
````

## 3. Funciones Globales aplicadas a DataFrames

💡 **Intuición**: Además de los métodos propios de Pandas, podemos usar funciones estándar de Python para obtener información rápida del objeto DataFrame.

````python
# Cantidad total de filas usando len()
print(f"Total de filas: {len(df_exams)}")

# Tipo de objeto
print(f"Tipo de objeto: {type(df_exams)}")

# Operaciones sobre el índice
print(f"Índice máximo: {max(df_exams.index)}")
print(f"Índice mínimo: {min(df_exams.index)}")

# Redondeo de valores numéricos
df_rounded = round(df_exams, 2)
df_rounded.head()
````

📋 **Resumen: Atributo vs Método**

| Característica | Atributo | Método |
|----------------|----------|---------|
| **Sintaxis** | `df.atributo` | `df.metodo()` |
| **Naturaleza** | Propiedad / Estado | Acción / Proceso |
| **Ejemplo** | `.shape` | `.describe()` |
| **Resultado** | Valor inmediato | Resultado de un cálculo |
