## 📖 Introducción

En la vida real, los análisis de datos rara vez dependen de una sola variable. Normalmente, necesitamos responder preguntas complejas como: *"¿Cuáles son las laptops de Apple que cuestan más de 2000 euros?"* o *"¿Qué laptops son Dell o HP y tienen pantallas de más de 15 pulgadas?"

Para lograr esto, Pandas nos permite combinar múltiples máscaras booleanas utilizando operadores lógicos.

## 🌉 Puente Pedagógico: `and`/`or` vs `&`/`|`

Un error común es intentar usar las palabras clave de Python `and` y `or` dentro de Pandas. 

- **`and` / `or`**: Están diseñados para evaluar la verdad de un *único* objeto (un escalar).
- **`&` / `|`**: Son operadores **bit-a-bit** (bitwise) que Pandas utiliza para comparar *cada elemento* de dos series booleanas simultáneamente.

### ⚠️ Regla de Oro: Los Paréntesis
Debido a la precedencia de operadores en Python, los operadores `&` y `|` tienen mayor prioridad que los operadores de comparación (`>`, `==`, etc.). Por lo tanto, **SIEMPRE** debes envolver cada condición entre paréntesis.

**Incorrecto**: `df['A'] > 1 & df['B'] < 2`  (Python intentará hacer `1 & df['B']` primero)
**Correcto**: `(df['A'] > 1) & (df['B'] < 2)`

### Flujo de Operadores
```
[ Condición 1 ] --( Operador )--> [ Condición 2 ] --> [ Máscara Final ]
       |                               |                      |
  (Máscara A)                     (Máscara B)          (Resultado Lógico)
       |
       +-->  & (AND): Solo True si AMBAS son True
       +-->  | (OR): True si AL MENOS UNA es True
       +-->  ~ (NOT): Invierte True <-> False
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('laptop_price.csv')
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. El Operador AND (`&`)

Se utiliza cuando queremos que **todas** las condiciones se cumplan simultáneamente.

**Caso**: Laptops de la marca 'Apple' **Y** con un precio superior a 2000 euros.

````python
# Definir condiciones
cond_apple = df_laptops['Company'] == 'Apple'
cond_expensive = df_laptops['Price_euros'] > 2000

# Combinar con AND
df_apple_expensive = df_laptops[cond_apple & cond_expensive]
df_apple_expensive.head()
````

### 2. El Operador OR (`|`)

Se utiliza cuando queremos que **al menos una** de las condiciones se cumpla.

**Caso**: Laptops que sean de la marca 'Apple' **O** de la marca 'Dell'.

````python
# Combinar con OR
df_apple_or_dell = df_laptops[(df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')]
df_apple_or_dell['Company'].value_counts()
````

### 3. Combinaciones Complejas

Podemos anidar operadores para crear filtros muy específicos. 

**Caso**: Laptops que sean ('Apple' o 'Dell') **Y** que además cuesten más de 2000 euros.

````python
# Filtro complejo con paréntesis anidados
filtro_complejo = ((df_laptops['Company'] == 'Apple') | (df_laptops['Company'] == 'Dell')) & (df_laptops['Price_euros'] > 2000)

df_premium_brands = df_laptops[filtro_complejo]
df_premium_brands.head()
````

## 📋 Resumen Rápido

| Operador | Nombre | Descripción | Ejemplo |
| :--- | :--- | :--- | :--- |
| `&` | AND | True si todas las condiciones son True | `(cond1) & (cond2)` |
| `|` | OR | True si al menos una condición es True | `(cond1) | (cond2)` |
| `~` | NOT | Invierte el valor booleano (True $\rightarrow$ False) | `~(cond1)` |
| `( )` | Paréntesis | Obligatorios para separar comparaciones | `(df['A'] > 1)` |
