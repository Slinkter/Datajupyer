# Análisis de Cardinalidad: Valores Únicos con `.unique()` y `.nunique()`

## 🎯 Objetivos
- Extraer la lista de valores distintos presentes en una columna de datos.
- Calcular la cardinalidad (número de valores únicos) de una variable.
- Comprender la diferencia entre obtener los valores y contar la cantidad de los mismos.

## 📖 Introducción

Cuando exploramos un nuevo dataset, una de las primeras preguntas que nos hacemos es: *"¿Cuántas categorías diferentes hay en esta columna?"* o *"¿Cuáles son exactamente esas categorías?"*.

Este concepto se conoce en bases de datos como **Cardinalidad**. Una columna con alta cardinalidad (como un ID de usuario) tiene muchos valores únicos, mientras que una de baja cardinalidad (como el género o el país) tiene pocos. Saber esto nos ayuda a decidir si una variable debe tratarse como categórica o numérica.

## 🌉 Puente Pedagógico: El Conjunto Matemático

En matemáticas, un **Conjunto (Set)** es una colección de elementos donde no se permiten duplicados. Si tienes la lista `[Apple, Apple, HP, Apple]`, el conjunto resultante es simplemente `{Apple, HP}`.

El método `.unique()` de Pandas aplica exactamente esta lógica a una columna entera, devolviendo un array de NumPy con los elementos en el orden en que aparecieron por primera vez.

### Flujo de Extracción
```
[ Columna con Repetidos ] --> [ .unique() ] --> [ Array de Valores Únicos ]
       (Apple)                                      (Apple)
       (Apple)       -------------------------->   (HP)
       (HP)                                        (Acer)
       (Apple)                                     (Lenovo)
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('laptop_price.csv')
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. Obtener los valores con `.unique()`

Este método devuelve un array con los valores únicos. Es ideal para validar que no haya errores de escritura (ej. 'Apple' y 'apple' serían detectados como dos valores distintos).

````python
# Extraer las marcas únicas presentes en el dataset
unique_companies = df_laptops['Company'].unique()
print(f"Marcas únicas encontradas:\n{unique_companies}")
````

### 2. Contar los valores con `.nunique()`

Si no nos interesan los nombres de los valores, sino solo **cuántos** hay, utilizamos `.nunique()`. Este método devuelve un número entero.

````python
# Calcular la cardinalidad de la columna 'Company'
num_companies = df_laptops['Company'].nunique()
print(f"Hay {num_companies} marcas diferentes en el dataset.")
````

### 3. Comparativa: `len(.unique())` vs `.nunique()`

Ambos métodos pueden dar el mismo resultado, pero hay una diferencia técnica clave: **el manejo de los valores nulos (NaN)**.

- `.unique()` incluye los valores `NaN` en el array resultante.
- `.nunique()` **excluye** los `NaN` por defecto.

````python
# Ejemplo comparativo (Simulando un valor nulo)
import numpy as np
test_series = pd.Series(['Apple', 'Apple', 'HP', np.nan])

print(f"Valores únicos (.unique()): {test_series.unique()}")
print(f"Conteo con len(.unique()): {len(test_series.unique())}")
print(f"Conteo con .nunique(): {test_series.nunique()}")
````
