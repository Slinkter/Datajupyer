## 📖 Introducción

En la Ciencia de Datos, a menudo necesitamos simplificar la información. En lugar de analizar cada precio individual, puede ser más útil saber si un producto es "Caro" o "Barato". Este proceso de transformar una variable continua en una categórica se conoce como **discretización** o **binning**.

La herramienta más eficiente en Python para realizar esta tarea en DataFrames es `np.where()`.

## 🌉 Puente Pedagógico: De `if-else` a `np.where()`

Si estuviéramos usando Python puro, escribiríamos un ciclo `for` con un `if-else` para cada fila. Sin embargo, esto es extremadamente lento en datasets grandes.

`np.where()` es la versión **vectorizada** de un `if-else`. En lugar de procesar fila por fila, procesa toda la columna al mismo tiempo utilizando instrucciones optimizadas de CPU (SIMD).

### Lógica de Funcionamiento
```
       [ Columna de Entrada ] 
                  |
         [ ¿Cumple Condición? ] 
          /               \ 
      ( SÍ )             ( NO )
        |                  |
 [ Valor A (True) ]   [ Valor B (False) ]
        \                  /
       [ Nueva Columna Resultante ]
```

````python
import pandas as pd
import numpy as np
from pathlib import Path

# Configuración de ruta
DATA_PATH = Path('laptop_price.csv')

# Carga de datos
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. Categorización de Precios

Queremos crear una columna llamada `Price_tier` donde:
- Si el precio es **> 2000**, la etiqueta sea `'Expensive'`.
- De lo contrario, la etiqueta sea `'Inexpensive'`.

**Sintaxis de `np.where`**: `np.where(condición, valor_si_true, valor_si_false)`

````python
# Crear la nueva columna basándose en el precio
df_laptops['Price_tier'] = np.where(
    df_laptops['Price_euros'] > 2000, 
    'Expensive', 
    'Inexpensive'
)

df_laptops[['Price_euros', 'Price_tier']].head()
````

### 2. Validación de Resultados

Es fundamental verificar que la distribución de las etiquetas sea coherente con nuestros datos utilizando `value_counts()`.

````python
# Contar la cantidad de laptops en cada categoría
distribucion = df_laptops['Price_tier'].value_counts()
print(f"Distribución de precios:\n{distribucion}")
````

## 📋 Resumen Rápido

| Componente | Descripción |
| :--- | :--- |
| **Condición** | Una expresión booleana que se evalúa para cada fila (ej. `df['col'] > 10`). |
| **Valor True** | El valor que se asigna cuando la condición es verdadera. |
| **Valor False** | El valor que se asigna cuando la condición es falsa. |
| **Eficiencia** | Alta (Vectorizada), ideal para millones de filas. |
