## 📖 Introducción

En el análisis de datos, rara vez necesitamos trabajar con el conjunto de datos completo. El **filtrado** es el proceso de seleccionar un subconjunto de filas que cumplen con un criterio específico. Esta capacidad es fundamental para realizar análisis segmentados, limpiar datos o aislar casos atípicos (outliers).

## 🌉 Puente Pedagógico: ¿Cómo funciona el filtrado?

Imagina que tienes una lista de productos y quieres solo los que son 'Rojos'. En lugar de recorrer la lista con un ciclo `for` y añadir los rojos a una nueva lista, Pandas utiliza un concepto llamado **Máscara Booleana**.

Una máscara es esencialmente una columna de valores `True` (si cumple la condición) y `False` (si no la cumple). Cuando pasamos esta máscara al DataFrame original, Pandas "deja pasar" solo las filas marcadas como `True`.

### Flujo de Filtrado
```
[ DataFrame ]  -->  [ Condición (Col == Valor) ]  -->  [ Máscara Booleana ]  -->  [ DataFrame Filtrado ]
      |                        |                            |                           |
 (Datos Totales)       (Criterio de Filtro)           (True/False por fila)       (Solo filas True)
```

````python
import pandas as pd
from pathlib import Path

# Definir la ruta al archivo de datos
DATA_PATH = Path('laptop_price.csv')

# Carga eficiente del dataset
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. Filtrado por Igualdad

Para encontrar filas donde una columna tiene un valor exacto, utilizamos el operador `==`. 

**Paso A: Crear la máscara.** 
Primero, generamos la serie de valores booleanos.

````python
# Crear máscara para laptops de la compañía 'Apple'
mask_apple = df_laptops['Company'] == 'Apple'
mask_apple.head()
````

**Paso B: Aplicar la máscara al DataFrame.**
Ahora usamos esa máscara para extraer las filas correspondientes.

````python
# Aplicar el filtro
df_apple = df_laptops[mask_apple]
df_apple.head()
````

### 2. Filtrado Numérico (Comparaciones)

Podemos utilizar operadores como `>` (mayor que), `<` (menor que), `>=` o `<=` para filtrar datos cuantitativos. 

Ejemplo: Laptops con un precio superior a 2000 euros.

````python
# Filtrar laptops caras (> 2000 euros)
df_expensive = df_laptops[df_laptops['Price_euros'] > 2000]
df_expensive.head()
````

### 3. Filtrado por Desigualdad

El operador `!=` nos permite excluir valores específicos. 

Ejemplo: Todas las laptops que **no** sean de la compañía 'HP'.

````python
# Filtrar excluyendo a 'HP'
df_not_hp = df_laptops[df_laptops['Company'] != 'HP']
df_not_hp['Company'].value_counts()
````

## 📋 Resumen Rápido

| Operador | Significado | Caso de Uso |
| :--- | :--- | :--- |
| `==` | Igual a | Buscar un valor específico (Categorías) |
| `!=` | Diferente de | Excluir un valor específico |
| `>` / `<` | Mayor / Menor que | Rangos numéricos (Precios, Edades) |
| `>=` / `<=` | Mayor o Igual / Menor o Igual | Límites inclusivos |
