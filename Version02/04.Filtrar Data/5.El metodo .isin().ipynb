# El Método `.isin()`: Simplificando la Selección Múltiple

## 🎯 Objetivos
- Implementar filtros basados en listas de valores utilizando el método `.isin()`.
- Reducir la complejidad del código al eliminar el uso excesivo de operadores OR (`|`).
- Combinar `.isin()` con otros filtros lógicos para consultas avanzadas.

## 📖 Introducción

A menudo necesitamos filtrar un DataFrame basándonos en varios valores posibles para una misma columna. Por ejemplo, si queremos extraer laptops de 'Apple', 'Dell', 'HP' y 'Lenovo', escribir la condición de la siguiente manera sería agotador:

```python
df[(df['Company'] == 'Apple') | (df['Company'] == 'Dell') | (df['Company'] == 'HP') | (df['Company'] == 'Lenovo')]
```

Pandas ofrece el método `.isin()`, que permite pasar una **lista** de valores y devuelve `True` si el elemento de la celda se encuentra en dicha lista. Es la forma más limpia y eficiente de realizar selecciones múltiples.

## 🌉 Puente Pedagógico: Membresía de Listas

En Python básico, para saber si un elemento está en una lista, usamos el operador `in`:
```python
marcas = ['Apple', 'HP']
if 'Apple' in marcas: print("Encontrado")
```

El método `.isin()` es exactamente lo mismo, pero diseñado para trabajar con **Series completas** de Pandas. En lugar de preguntar por un solo valor, pregunta por miles de valores al mismo tiempo.

### Flujo de Funcionamiento
```
[ Columna DataFrame ] --> [ .isin([Val1, Val2, ...]) ] --> [ Máscara Booleana ]
        |                               |                           |
 (Datos de Marca)           (¿Está en la lista?)            (True si coincide)
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('laptop_price.csv')
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. Filtrado Simple con `.isin()`

Vamos a extraer todas las laptops que pertenezcan a las compañías 'Apple' o 'HP'.

````python
# Definir la lista de valores deseados
marcas_objetivo = ['Apple', 'HP']

# Crear la máscara usando .isin()
mask_marcas = df_laptops['Company'].isin(marcas_objetivo)

# Aplicar el filtro
df_filtered = df_laptops[mask_marcas]
df_filtered['Company'].value_counts()
````

### 2. Filtrado Múltiple y Combinado

Podemos utilizar `.isin()` en varias columnas y combinar los resultados con operadores lógicos (`&` o `|`).

**Caso**: Laptops que sean de 'Apple' o 'HP' **Y** que sean de tipo 'Notebook' o 'Ultrabook'.

````python
# Filtro 1: Marcas específicas
filter_company = df_laptops['Company'].isin(['Apple', 'HP'])

# Filtro 2: Tipos de laptop específicos
filter_type = df_laptops['TypeName'].isin(['Notebook', 'Ultrabook'])

# Combinar ambos filtros con AND (&)
df_final = df_laptops[filter_company & filter_type]
df_final.head()
````

## 📋 Resumen Rápido

| Método / Operador | Sintaxis | Uso Ideal |
| :--- | :--- | :--- |
| **Operador `|`** | `(df['A']==1) | (df['A']==2)` | 2 o 3 condiciones simples. |
| **Método `.isin()`** | `df['A'].isin([1, 2, 3, 4])` | 3 o más valores posibles en la misma columna. |
| **Combinación** | `df[mask1 & mask2]` | Filtros complejos basados en múltiples columnas. |
