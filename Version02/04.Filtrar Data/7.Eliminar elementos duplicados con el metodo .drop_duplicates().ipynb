# Eliminación de Duplicados: Limpiando Datos con `.drop_duplicates()`

## 📖 Introducción

En la etapa de limpieza de datos (Data Cleaning), una vez que hemos identificado los duplicados con `.duplicated()`, el siguiente paso lógico es eliminarlos. 

El método `.drop_duplicates()` permite limpiar la base de datos eliminando las filas que contienen información repetida. Esto es crucial para evitar el sobreajuste (overfitting) en modelos de Machine Learning y para obtener métricas estadísticas precisas.

## 🌉 Puente Pedagógico: El Filtro de Representantes

Imagina que tienes un álbum de cromos con muchos repetidos. Para ahorrar espacio, decides quedarte solo con **un cromo de cada personaje**. 

`.drop_duplicates()` hace exactamente eso: analiza el DataFrame y, cuando encuentra un grupo de filas idénticas, descarta todas menos una. Tú decides cuál conservar: el primero que llegó, el último, o borrar el grupo completo si hay cualquier repetición.

### Flujo de Limpieza
```
[ DataFrame con Duplicados ] --> [ .drop_duplicates() ] --> [ DataFrame Limpio ]
   Fila A (Original)    \                                     Fila A (Conservada)
   Fila A (Repetida)    --->   (Filtro de Unicidad)   --->    Fila B (Conservada)
   Fila B (Original)    /                                    Fila C (Conservada)
   Fila B (Repetida)    /
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('laptop_price.csv')
df_laptops = pd.read_csv(DATA_PATH)
df_laptops.head()
````

## 🛠️ Implementación

### 1. Eliminación Básica

Si queremos una lista de las compañías presentes en el dataset sin repeticiones, podemos eliminar duplicados basados en la columna `Company`.

````python
# Eliminar duplicados de la columna 'Company' y conservar solo la primera aparición
df_unique_companies = df_laptops.drop_duplicates(subset=['Company'])

print(f"Filas originales: {len(df_laptops)}")
print(f"Filas únicas por compañía: {len(df_unique_companies)}")
df_unique_companies[['Company', 'Price_euros']].head()
````

### 2. Control de Conservación con `keep`

El parámetro `keep` define qué fila sobrevive a la limpieza:
- `keep='first'`: Conserva la primera fila encontrada (elimina las siguientes).
- `keep='last'`: Conserva la última fila encontrada (elimina las anteriores).
- `keep=False`: Elimina **todas** las filas que tengan duplicados.

````python
# Ejemplo: Eliminar todas las laptops que tengan un nombre de producto repetido
df_no_repeats = df_laptops.drop_duplicates(subset=['Product'], keep=False)
print(f"Filas sin ningún tipo de repetición de producto: {len(df_no_repeats)}")
````

### 3. El Patrón "Representante Único"

Una técnica poderosa es combinar `sort_values()` con `drop_duplicates()`. Esto nos permite elegir qué registro conservar basándonos en un valor (como el precio más bajo).

**Objetivo**: Encontrar la laptop más barata de cada marca.

````python
# 1. Ordenar por Marca (asc) y Precio (asc)
df_sorted = df_laptops.sort_values(['Company', 'Price_euros'])

# 2. Eliminar duplicados de marca conservando el primero (que es el más barato)
df_cheapest_per_brand = df_sorted.drop_duplicates(subset=['Company'], keep='first')

df_cheapest_per_brand[['Company', 'Product', 'Price_euros']].head()
````

### 4. Parámetros Adicionales

- `inplace=True`: Modifica el DataFrame original directamente en lugar de devolver una copia.
- `ignore_index=True`: Reinicia el índice del DataFrame resultante (0, 1, 2...), evitando huecos causados por las filas eliminadas.

````python
# Ejemplo con ignore_index
df_clean = df_laptops.drop_duplicates(subset=['Product'], ignore_index=True)
df_clean.head()
````

## 📋 Resumen Rápido

| Parámetro | Valor | Efecto |
| :--- | :--- | :--- |
| `subset` | `['col1', 'col2']` | Define qué columnas analizar para buscar duplicados. |
