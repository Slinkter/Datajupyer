## 📖 Introducción

En Pandas, existen dos formas principales de acceder a los datos: mediante la posición numérica (`.iloc`) o mediante la etiqueta del índice (`.loc`).

El uso de `.loc` es la práctica estándar en la industria porque hace que el código sea **semántico**. Es mucho más claro leer `df.loc['L. Messi']` que `df.iloc[0]`, especialmente cuando el orden de los datos puede cambiar después de un filtrado o un ordenamiento.

## 🌉 Puente Pedagógico: El Sistema de Direcciones

Imagina que estás buscando una casa en una calle. Tienes dos opciones:
1. **Posición**: "Ve a la quinta casa desde la esquina". (Si alguien construye una casa nueva al principio, la quinta casa ahora es otra diferente).
2. **Dirección**: "Ve a la casa número 123". (No importa cuántas casas se construyan alrededor, la dirección 123 siempre llevará a la misma casa).

`.loc` funciona como la **dirección**. No le importa en qué posición está la fila; busca la etiqueta exacta que definimos en el índice.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('players_20.csv')
df = pd.read_csv(DATA_PATH)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
df.head()
```

## 🛠️ Implementación

### 1. Acceso a Valores Específicos

La sintaxis básica es: `df.loc[fila, columna]`.

```python
# Obtener toda la información de un jugador específico (devuelve una Serie)
messi_data = df.loc['L. Messi']
print("Datos de Messi:\n", messi_data)

# Obtener un dato puntual (celda específica)
messi_height = df.loc['L. Messi', 'height_cm']
print(f"\nLa altura de Messi es: {messi_height} cm")
```

### 2. Selección de Múltiples Elementos

Podemos pasar una lista de etiquetas para obtener varias filas o columnas a la vez.

```python
# Seleccionar varios jugadores
estrellas = ['L. Messi', 'Cristiano Ronaldo']
df_estrellas = df.loc[estrellas]
df_estrellas
```

### 3. Rebanadas (Slicing) con Etiquetas

A diferencia de las rebanadas normales de Python, en `.loc` **el límite final está incluido**.

```python
# Seleccionar un rango de jugadores y un rango de columnas
# Desde 'L. Messi' hasta 'M. Salah' en el índice
# Desde la columna 'age' hasta 'club'
df_slice = df.loc['L. Messi':'M. Salah', 'age':'club']
df_slice.head()
```

### 4. Filtrado Booleano + `.loc`

La combinación más poderosa es usar una máscara booleana en las filas y una lista de columnas. Esto nos permite decir: "Dame las columnas X e Y de todas las filas que cumplan la condición Z".

```python
# Objetivo: Jugadores argentinos con más de 180cm, solo queremos su nombre y club
mask = (df['nationality'] == 'Argentina') & (df['height_cm'] > 180)
cols = ['long_name', 'club']

df_argentinos_altos = df.loc[mask, cols]
df_argentinos_altos.head()
```

## 📋 Resumen Rápido

| Característica | `.loc` | `.iloc` |
| :--- | :--- | :--- | 
| **Referencia** | Etiquetas / Nombres | Posiciones Enteras |
| **Slicing** | Incluye el final `[start:stop]` | Excluye el final `[start:stop)` |
| **Uso** | Acceso semántico y legible | Acceso posicional y genérico |
