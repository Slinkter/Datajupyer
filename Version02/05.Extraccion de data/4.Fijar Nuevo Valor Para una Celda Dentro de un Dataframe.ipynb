# Modificación de Datos: Asignación de Valores con `.loc`

## 🎯 Objetivos

## 📖 Introducción

En la práctica, los datos rara vez llegan perfectos. Es común encontrar errores de digitación, valores desactualizados o la necesidad de crear "marcas" (flags) para identificar ciertos grupos de datos.

Para modificar la información, la herramienta más segura y explícita es `.loc`. A diferencia de otras formas de asignación que pueden generar el aviso `SettingWithCopyWarning`, `.loc` garantiza que estamos modificando la vista correcta del DataFrame original.

## 🌉 Puente Pedagógico: La Edición Quirúrgica

Imagina que el DataFrame es un gran tablero de celdas. Modificar un dato con `.loc` es como realizar una cirugía: 

1. **Localizas** el punto exacto (Fila $\rightarrow$ Columna).
2. **Sustituyes** el valor antiguo por el nuevo.
3. **Cierras** la operación sin alterar ninguna otra celda del tablero.

### Flujo de Modificación
```
   Sintaxis: df.loc[ fila, columna ] = nuevo_valor
               |           |            |
               V           V            V
   [ "L. Messi" ] [ "height_cm" ] [   175   ]
                                      ^
                             (Sustituye el 170)
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('players_20.csv')
df = pd.read_csv(DATA_PATH)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
df.head()
````

## 🛠️ Implementación

### 1. Actualización de una Celda Única

Cuando detectamos un error en un dato puntual, podemos corregirlo directamente.

````python
# Corregir la altura de L. Messi de 170 a 175
df.loc['L. Messi', 'height_cm'] = 175

print(f"Nueva altura de Messi: {df.loc['L. Messi', 'height_cm']} cm")
````

### 2. Asignación a Columna Completa

Si queremos resetear o estandarizar una columna entera, podemos asignar un valor único a todas las filas.

````python
# Establecer la altura de TODOS los jugadores a 190 (Ejemplo hipotético)
df.loc[:, 'height_cm'] = 190
df[['height_cm']].head()
````

### 3. Actualización Basada en Condiciones

Esta es la función más potente de `.loc`. Nos permite actualizar solo las filas que cumplen un criterio específico.

**Caso**: Queremos marcar con el valor `0` todas las variables físicas de los jugadores que miden más de 180cm para analizarlos por separado.

````python
# Definir las columnas a modificar
cols_to_reset = ['age', 'dob', 'height_cm', 'weight_kg']

# Aplicar el cambio solo si height_cm > 180
df.loc[df['height_cm'] > 180, cols_to_reset] = 0
df.head()
````

## 📋 Resumen Rápido

| Tipo de Cambio | Sintaxis | Resultado |
| :--- | :--- | :--- |
| **Puntual** | `df.loc[fila, col] = val` | Modifica una sola celda. |
| **Masivo** | `df.loc[:, col] = val` | Modifica toda la columna. |
| **Condicional** | `df.loc[mask, col] = val` | Modifica solo las filas que cumplen la condición. |
