## 📖 Introducción

Mientras que `.loc` busca "nombres", `.iloc` (Integer-location) busca "coordenadas". No le importa si el índice de la fila es "L. Messi" o "Z. Wei"; para `.iloc`, simplemente es la fila `0` o la fila `18277`.

Este método es indispensable cuando estamos automatizando procesos mediante ciclos, realizando muestreos aleatorios o cuando trabajamos con datasets que no tienen un índice con nombres significativos.

## 🌉 Puente Pedagógico: El Sistema de Coordenadas de Excel

Imagina que el DataFrame es una hoja de cálculo de Excel. En lugar de buscar la pestaña llamada "Ventas", decides buscar la **celda B5**. 

No importa si cambias el nombre de la pestaña o el título de la columna; la celda en la fila 5, columna 2 siempre estará en la misma posición física. `.iloc` es el GPS que nos lleva a esa coordenada exacta.

### Mapa de Acceso
```
           Col 0    Col 1    Col 2    Col 3
         +--------+--------+--------+--------+
 Row 0   | (0,0)  | (0,1)  | (0,2)  | (0,3)  |
         +--------+--------+--------+--------+
 Row 1   | (1,0)  | (1,1)  | (1,2)  | (1,3)  |
         +--------+--------+--------+--------+
 Row 2   | (2,0)  | (2,1)  | (2,2)  | (2,3)  |
         +--------+--------+--------+--------+
```

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('players_20.csv')
df = pd.read_csv(DATA_PATH)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
df.head()
````

## 🛠️ Implementación

### 1. Acceso a Valores Puntuales

La sintaxis es `df.iloc[indice_fila, indice_columna]`. Recuerda que en Python, el conteo comienza siempre en `0`.

````python
# Obtener el valor de la primera fila y cuarta columna (altura de Messi)
height = df.iloc[0, 3]
print(f"El valor en la posición [0, 3] es: {height} cm")

# Obtener el valor de la segunda fila y quinta columna (peso de Cristiano)
weight = df.iloc[1, 4]
print(f"El valor en la posición [1, 4] es: {weight} kg")
````

### 2. Selección de Series (Filas o Columnas Completas)

Utilizamos el operador `:` para indicar que queremos "todos" los elementos de esa dimensión.

````python
# Obtener la fila completa de la posición 0 (L. Messi)
row_0 = df.iloc[0, :]
print("Fila 0:\n", row_0)

# Obtener la columna completa de la posición 3 (height_cm)
col_3 = df.iloc[:, 3]
print(f"\nPrimeros 5 valores de la columna 3:\n{col_3.head()}")
````

### 3. Slicing Numérico

A diferencia de `.loc`, el slicing de `.iloc` sigue la regla estándar de Python: **el límite inferior se incluye, pero el superior se excluye**.

````python
# Seleccionar las primeras 5 filas y las columnas de la 1 a la 4 (excluyendo la 5)
df_slice = df.iloc[0:5, 1:5]
df_slice
````

### 4. `.iloc` con Máscaras Booleanas

Un punto crítico: `.iloc` **no acepta** Series booleanas directamente porque espera enteros. Para usar un filtro booleano con `.iloc`, debemos convertir la Serie a una lista de Python utilizando `.tolist()` o `list()`.

````python
# Objetivo: Jugadores con altura > 180cm, extrayendo solo la columna de nombre (pos 0)
mask = df['height_cm'] > 180
df_altos = df.iloc[list(mask), 0]
print(f"Jugadores altos (primeros 5):\n{df_altos.head()}")
````

## 📋 Resumen Rápido: `.loc` vs `.iloc`

| Característica | `.loc` | `.iloc` |
| :--- | :--- | :--- | 
| **Referencia** | Etiquetas / Nombres | Posiciones Enteras |
| **Slicing** | Incluye el final `[start:stop]` | Excluye el final `[start:stop)` |
| **Lógica** | "Búscame a Messi" | "Dame la fila 0" |
| **Uso Principal** | Análisis basado en datos | Automatización, ciclos, muestreo |
