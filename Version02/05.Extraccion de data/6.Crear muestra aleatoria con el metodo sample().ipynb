# Creación de Muestras Aleatorias con `.sample()`

## 🎯 Objetivos
En este notebook, aprenderás a extraer subconjuntos aleatorios de tus datos, una técnica fundamental para el análisis estadístico y el entrenamiento de modelos de Machine Learning. Al finalizar, serás capaz de:
1. Extraer una cantidad específica de elementos usando el parámetro `n`.
2. Obtener un porcentaje del dataset utilizando el parámetro `frac`.
3. Garantizar que tus resultados sean reproducibles mediante `random_state`.
4. Implementar el muestreo con reemplazo para generar datasets sintéticos o balancear clases.

## 💡 Introducción

Cuando trabajamos con "Big Data", procesar millones de filas puede ser computacionalmente costoso y lento. El **muestreo aleatorio** nos permite trabajar con una fracción representativa de los datos para obtener intuiciones rápidas o validar hipótesis sin necesidad de procesar todo el volumen de información.

Además, en Machine Learning, el muestreo es la base para dividir los datos en conjuntos de **Entrenamiento (Train)** y **Prueba (Test)**, evitando que el modelo simplemente "memorice" los datos.

````python
import pandas as pd
from pathlib import Path

# Configuración del dataset
file_path = Path('players_20.csv')
df = pd.read_csv(file_path)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]

df.head()
````

## 🛠️ El Concepto de Muestreo Aleatorio

Muestrear es como tomar un "puñado" de datos al azar de una bolsa. Si el puñado es lo suficientemente grande y aleatorio, debería representar las características de toda la bolsa.

### 🌉 Puente Pedagógico: Visualización del Proceso
```
  [ POBLACIÓN TOTAL (DataFrame) ]
  +----------------------------+
  | Fila 1 | Fila 2 | Fila 3 ... |
  | Fila 4 | Fila 5 | Fila 6 ... |
  | Fila 7 | Fila 8 | Fila 9 ... |
  +----------------------------+
               |
               v  <-- .sample()
               |
  [ MUESTRA ALEATORIA (Subset) ]
  +----------------------------+
  | Fila 2 | Fila 7 | Fila 12 ... |
  +----------------------------+
```

### 1. Muestreo por Cantidad (`n`)

Si sabes exactamente cuántos elementos necesitas, utiliza el parámetro `n`. Esto funciona tanto en DataFrames como en Series.

````python
# Extraer 10 nacionalidades al azar
muestra_n = df['nationality'].sample(n=10, random_state=99)

print("Muestra de 10 nacionalidades:")
print(muestra_n)
````

### 2. Muestreo por Proporción (`frac`)

Cuando no necesitas un número fijo, sino un porcentaje del total, utiliza `frac` (fracción). Por ejemplo, `frac=0.2` extraerá el 20% de los datos.

````python
# Extraer el 10% del DataFrame
muestra_frac = df.sample(frac=0.1, random_state=99)

print(f"Tamaño original: {len(df)}")
print(f"Tamaño de la muestra (10%): {len(muestra_frac)}")
muestra_frac.head()
````

### 3. Reproducibilidad con `random_state`

El muestreo es, por definición, aleatorio. Esto significa que cada vez que corras la celda, obtendrás resultados diferentes. En ciencia de datos, esto es un problema porque otros no podrán replicar tus resultados.

El parámetro `random_state` actúa como una "semilla". Si usas el mismo número (ej. `99`), pandas generará siempre la **misma secuencia aleatoria**.

### 4. Muestreo con Reemplazo (`replace`)

Por defecto, `replace=False`, lo que significa que una vez que un elemento es seleccionado, no puede volver a salir en la misma muestra.

Si establecemos `replace=True`, permitimos que los elementos se repitan. Esto es útil para:
1. **Over-sampling**: Aumentar la cantidad de datos sintéticamente.
2. **Bootstrapping**: Técnica estadística para estimar la precisión de un modelo.

````python
# Crear una muestra que sea el DOBLE del tamaño del dataset original
# Esto es imposible sin replace=True
muestra_gigante = df.sample(frac=2.0, replace=True, random_state=99)

print(f"Tamaño original: {len(df)}")
print(f"Tamaño de la muestra con reemplazo: {len(muestra_gigante)}")
````

## 📋 Resumen Rápido

| Parámetro | Tipo | Descripción |
| :--- | :--- | :--- |
| `n` | Entero | Número exacto de elementos a extraer. |
| `frac` | Float | Proporción del dataset (0.1 = 10%). |
| `replace` | Booleano | `True` permite repetir elementos; `False` no. |
| `random_state` | Entero | Semilla para asegurar resultados idénticos en cada ejecución. |
