# Eliminar Filas y Columnas de un DataFrame

## 🎯 Objetivos
En este notebook, aprenderás a limpiar tu conjunto de datos eliminando información irrelevante o redundante. Al finalizar, serás capaz de:
1. Utilizar el método `.drop()` para remover filas y columnas.
2. Comprender y aplicar el parámetro `axis` para navegar la estructura del DataFrame.
3. Gestionar la persistencia de los cambios mediante el parámetro `inplace`.

## 💡 Introducción

En el análisis de datos, es común encontrarse con columnas que no aportan valor al modelo (como IDs internos) o filas que contienen errores o datos irrelevantes. La capacidad de "podar" el DataFrame es esencial para optimizar la memoria y mejorar la calidad del análisis.

El método principal para lograr esto es `.drop()`, que actúa como una herramienta de precisión para extraer elementos específicos de la tabla.

````python
import pandas as pd
from pathlib import Path

# Configuración del dataset
file_path = Path('players_20.csv')
df = pd.read_csv(file_path)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]

df.head()
````

## 🛠️ El Concepto de `axis` (El Eje)

Para indicarle a pandas qué queremos eliminar, debemos especificar el eje (`axis`). Esta es una de las fuentes de confusión más comunes para los principiantes.

### 🌉 Puente Pedagógico: Intuición del Eje
Imagina que el DataFrame es una rejilla de coordenadas:
- **Axis 0 (Filas)**: Se mueve verticalmente. Si eliminas en el eje 0, estás "cortando" una franja horizontal (una fila).
- **Axis 1 (Columnas)**: Se mueve horizontalmente. Si eliminas en el eje 1, estás "cortando" una franja vertical (una columna).

```
             AXIS 1 (Columnas) -->
           +-------+-------+-------+
           | Col A | Col B | Col C |
 AXIS 0   +-------+-------+-------+
 (Filas)   | Fila 0|  ...  |  ...  |
   |      +-------+-------+-------+
   v      | Fila 1|  ...  |  ...  |
           +-------+-------+-------+
```

### 1. Eliminar Filas

Para eliminar filas, podemos usar el nombre del índice o el parámetro `index`.

````python
# Opción A: Usando el nombre del índice y axis=0
df_sin_messi = df.drop('L. Messi', axis=0)

# Opción B: Usando explícitamente el parámetro index (más legible)
df_sin_messi_alt = df.drop(index='L. Messi')

print(f"Tamaño original: {len(df)}")
print(f"Tamaño nuevo: {len(df_sin_messi)}")
````

### 2. Eliminar Múltiples Filas y el parámetro `inplace`

Por defecto, `.drop()` devuelve una **copia** del DataFrame. Si quieres que el cambio se aplique directamente al objeto original sin asignar el resultado a una variable, usa `inplace=True`.

````python
# Eliminamos varias filas simultáneamente
filas_a_eliminar = ['L. Messi', 'Cristiano Ronaldo']
df.drop(index=filas_a_eliminar, inplace=True)

df.head()
````

### 3. Eliminar Columnas

Para eliminar columnas, cambiamos el eje a `axis=1` o utilizamos el parámetro `columns`.

````python
# Opción A: Usando axis=1
df_sin_nombre = df.drop('long_name', axis=1)

# Opción B: Usando el parámetro columns (recomendado por claridad)
df_sin_nombre_alt = df.drop(columns=['long_name'])

df_sin_nombre.head()
````

#### Eliminar por posición
A veces no conocemos el nombre de la columna, pero sí su posición. Podemos combinar `df.columns` con `.drop()`.

````python
# Eliminar la última columna
columna_final = df.columns[-1]
df_sin_ultima = df.drop(columns=[columna_final])

print(f"Columna eliminada: {columna_final}")
````

### 4. Eliminación Masiva de Columnas
Podemos pasar una lista de nombres para limpiar el dataset rápidamente.

````python
columnas_irrelevantes = ['long_name', 'dob']
df.drop(columns=columnas_irrelevantes, inplace=True)

df.head()
````

## 📋 Resumen Rápido

| Objetivo | Método | Parámetro Clave |
| :--- | :--- | :--- |
| **Eliminar Fila** | `.drop()` | `axis=0` o `index=...` |
| **Eliminar Columna** | `.drop()` | `axis=1` o `columns=...` |
| **Modificar Original** | `.drop()` | `inplace=True` |
| **Mantener Original** | `.drop()` | `inplace=False` (Default) |
