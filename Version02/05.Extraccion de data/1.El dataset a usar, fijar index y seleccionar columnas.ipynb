# Fundamentos de Extracción: Indexación y Selección de Columnas

## 🎯 Objetivos
- Cargar datasets de forma eficiente utilizando `pandas` y `pathlib`.
- Transformar columnas identificadoras en índices para mejorar la búsqueda de datos.
- Realizar proyecciones de datos seleccionando únicamente las columnas necesarias para el análisis.

## 📖 Introducción

Cargar un archivo CSV es el primer paso de cualquier proyecto de datos, pero los datasets reales suelen venir con cientos de columnas y miles de filas. Trabajar con toda la información puede saturar la memoria RAM y hacer que el código sea ilegible.

Para optimizar nuestro flujo de trabajo, aplicamos dos técnicas fundamentales: la **Indexación** (para acceder a filas por nombre en lugar de posición) y la **Selección** (para reducir el ruido eliminando columnas irrelevantes).

## 🌉 Puente Pedagógico: El Índice como "Tabla de Contenidos"

Por defecto, Pandas asigna un número entero a cada fila (0, 1, 2...). Esto es útil, pero imagina que quieres buscar la información de "L. Messi". Tendrías que saber que él está en la fila 0.

Al usar `.set_index()`, convertimos una columna (como el nombre del jugador) en la etiqueta de la fila. Ahora, la fila no se identifica por el número 0, sino por la etiqueta "L. Messi". Es exactamente como el índice de un libro: no buscas la página 45, buscas el capítulo "Introducción".
### Flujo de Organización de Datos
```
[ DataFrame Base ] --> [ .set_index('Nombre') ] --> [ DataFrame Indexado ]
  Fila 0: Messi          ( la etiqueta )            Messi: [Datos...]
  Fila 1: Ronaldo  --->  ( se vuelve )    --->      Ronaldo: [Datos...]
  Fila 2: Neymar         ( el índice )             Neymar: [Datos...]
```

````python
import pandas as pd
from pathlib import Path

# Definimos la ruta al archivo usando pathlib para asegurar compatibilidad
DATA_PATH = Path('players_20.csv')

# Carga del dataset
df = pd.read_csv(DATA_PATH)
df.head()
````

## 🛠️ Implementación

### 1. Establecer el Índice

Utilizaremos la columna `short_name` como índice, ya que es un identificador humano legible y único para la mayoría de los casos.

````python
# Establecer el índice
# Nota: usamos inplace=True para modificar el objeto original sin crear una copia
df.set_index('short_name', inplace=True)

df.head()
````

### 2. Selección de Columnas (Proyección)

El dataset original tiene más de 100 columnas. Para nuestro análisis, solo necesitamos algunas variables demográficas y profesionales. 

Para seleccionar múltiples columnas, pasamos una **lista de strings** dentro de los corchetes del DataFrame: `df[['col1', 'col2']]`.

````python
# Definimos las columnas de interés
cols_interes = ['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']

# Creamos un nuevo DataFrame solo con esas columnas
df_perfil = df[cols_interes]

df_perfil.head()
````

| Operación | Comando | Propósito |
