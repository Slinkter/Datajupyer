# Filtrado Eficiente con el Método `.query()`

## 🎯 Objetivos
El filtrado es la tarea más común en el análisis de datos. Aunque el slicing booleano es poderoso, pandas ofrece el método `.query()` para hacer que el código sea más legible y cercano al lenguaje natural (estilo SQL). En este notebook aprenderás a:
1. Implementar filtros básicos utilizando expresiones de cadena.
2. Comparar la sintaxis de `.query()` frente al slicing booleano tradicional.
3. Construir filtros complejos combinando múltiples condiciones (`and`, `or`, `not`).
4. Integrar variables externas en tus consultas utilizando el operador `@`.

## 💡 Introducción

Cuando trabajamos con DataFrames grandes, escribir `df[(df['columna'] > 10) & (df['otra_columna'] == 'Valor')]` puede volverse tedioso y difícil de leer debido a la repetición del nombre del DataFrame.

El método `.query()` resuelve esto permitiéndonos escribir la condición como una cadena de texto. Internamente, pandas evalúa esta expresión, lo que a menudo resulta en un código más limpio y, en algunos casos, más eficiente.

````python
import pandas as pd
from pathlib import Path

# Configuración del dataset
file_path = Path('players_20.csv')
df = pd.read_csv(file_path)
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]

df.head()
````

## 🛠️ Cómo funciona `.query()`

A diferencia del slicing booleano, donde creamos una máscara de True/False, `.query()` envía una instrucción a un motor de evaluación interno.

### 🌉 Puente Pedagógico: El Flujo de la Consulta
```
  [ DataFrame ]  +  [ "Expresión de Filtro" ]
          \               / 
           \             / 
            v           v
          [ Motor de query() ]  <-- Analiza la cadena
                    |
                    v
          [ Nuevo DataFrame ]   <-- Solo filas que cumplen la condición
```

### 1. Filtros Básicos

Probemos un filtro simple: jugadores mayores de 34 años.

````python
# Usando .query()
mayores_34 = df.query("age > 34")

print(f"Jugadores encontrados: {len(mayores_34)}")
mayores_34.head()
````

### 2. `.query()` vs Slicing Booleano

Es importante conocer ambas formas. Observa la diferencia en la verbosidad:

````python
# Slicing Booleano (Tradicional)
resultado_slice = df[df['age'] > 34]

# Método .query() (Moderno)
resultado_query = df.query("age > 34")

print("¿Son idénticos?", resultado_slice.equals(resultado_query))
````

### 3. Filtros Complejos (Operadores Lógicos)

Podemos combinar condiciones usando `and` (Y), `or` (O) y `not` (NO).

````python
# Jugadores mayores de 34 años que sean de Italia
italianos_veteranos = df.query("age > 34 and nationality == 'Italy'")

# Jugadores que NO sean de Argentina ni de Brasil
no_latino_top = df.query("not (nationality == 'Argentina' or nationality == 'Brazil')")

print(f"Veteranos italianos: {len(italianos_veteranos)}")
print(f"No Argentina/Brasil: {len(no_latino_top)}")
````

### 4. Filtros Dinámicos con el Operador `@`

Una de las funciones más potentes de `.query()` es la capacidad de referenciar variables de Python externas utilizando el símbolo `@`.

````python
pais_objetivo = 'Argentina'
edad_minima = 30

# Filtramos usando las variables externas
filtro_dinamico = df.query("nationality == @pais_objetivo and age > @edad_minima")

print(f"Jugadores de {pais_objetivo} mayores de {edad_minima}:")
filtro_dinamico
````

## 📋 Resumen Rápido

| Operador | Uso en `.query()` | Ejemplo |
| :--- | :--- | :--- |
| **Y** | `and` | `"age > 20 and age < 30"` |
| **O** | `or` | `"nationality == 'Italy' or nationality == 'Spain'"` |
| **NO** | `not` | `"not (age < 18)"` |
| **Variable** | `@` | `"nationality == @mi_variable"` |
