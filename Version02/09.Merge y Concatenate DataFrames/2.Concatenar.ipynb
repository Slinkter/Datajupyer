## 📖 Introducción

La concatenación es la operación de "pegar" DataFrames uno junto al otro. A diferencia del `merge`, que busca coincidencias en valores de columnas, la concatenación se basa principalmente en la **posición** o en el **índice**.

Es la herramienta ideal cuando tienes datos del mismo formato que provienen de diferentes fuentes (por ejemplo, ventas de enero en un archivo y ventas de febrero en otro) o cuando quieres añadir nuevas características (columnas) a un conjunto de datos ya existente.

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path(".")

# Carga rápida para ejemplos
df_movies = pd.read_csv(DATA_PATH / "IMDb movies.csv", low_memory=False)
df_movies = df_movies[['imdb_title_id', 'title', 'year', 'genre', 'country']]
````

## 1. Concatenación Vertical (`axis=0`)

### 💡 Intuición
La concatenación vertical es como apilar cajas una sobre otra. Se utiliza cuando los DataFrames tienen las **mismas columnas** y queremos añadir más filas (observaciones).

### 🛠️ Diagrama de Flujo
```
 [ DF 1 (Filas 1-10) ]       [ DF 2 (Filas 11-20) ]
         |                              |
         +-----------------+-----------+ 
                            |
                            v
                   [ Combined DF (1-20) ]
                   +-------------------+
                   |      Datos DF 1    |
                   +-------------------+
                   |      Datos DF 2    |
                   +-------------------+
```

````python
# Creación de DataFrames de ejemplo
df_jan = pd.DataFrame({'id': ['A', 'B'], 'ventas': [100, 200]})
df_feb = pd.DataFrame({'id': ['C', 'D'], 'ventas': [150, 250]})

print("--- Enero ---")
display(df_jan)
print("\n--- Febrero ---")
display(df_feb)
````

````python
# Concatenación Vertical
# ignore_index=True es vital para evitar índices duplicados (0,1,0,1...)
df_total = pd.concat([df_jan, df_feb], axis=0, ignore_index=True)

print("--- Ventas Totales (Concatenado) ---")
display(df_total)
````

## 2. Concatenación Horizontal (`axis=1`)

### 💡 Intuición
La concatenación horizontal es como poner dos hojas de papel una al lado de la otra. Se utiliza cuando tenemos el mismo número de filas y queremos añadir nuevas columnas (características).

**Importante**: Para que la concatenación horizontal tenga sentido, ambos DataFrames deben compartir el mismo **índice**. Si los índices no coinciden, Pandas creará filas adicionales con valores `NaN`.

````python
# DataFrames con el mismo índice
df_info = pd.DataFrame({'nombre': ['Ana', 'Luis'], 'ciudad': ['Madrid', 'Lima']})
df_stats = pd.DataFrame({'puntos': [85, 92], 'nivel': ['Pro', 'Expert']})

print("--- Info ---")
display(df_info)
print("\n--- Stats ---")
display(df_stats)
````

````python
# Concatenación Horizontal
df_full = pd.concat([df_info, df_stats], axis=1)

print("--- Perfil Completo ---")
display(df_full)
````

## ⚖️ ¿Concatenar (`concat`) o Unir (`merge`)?

Es común confundir estas dos operaciones. Aquí la guía rápida:

| Característica | `pd.concat()` | `pd.merge()` |
| :--- | :--- | :--- |
| **Lógica** | Basado en posición/índice | Basado en valores de columnas |
| **Dirección** | Vertical u Horizontal | Principalmente Horizontal |
| **Uso Principal** | Unir datasets idénticos | Relacionar tablas diferentes |
| **Analogía** | Pegar hojas de papel | Cruce de bases de datos (SQL Join) |

## 📋 Resumen Rápido

1. **`axis=0`**: Apila filas (Vertical). Útil para agregar más datos del mismo tipo.
2. **`axis=1`**: Pega columnas (Horizontal). Útil para agregar más atributos.
3. **`ignore_index=True`**: Limpia el índice después de una concatenación vertical.
4. **`pd.concat([list_of_dfs])`**: La función principal para unificar estructuras similares.
