- Transformar datos crudos en tablas de reporte profesionales.

## 📖 Introducción

En el análisis de datos, a menudo nos encontramos con datos en **formato largo** (donde cada fila es una observación única), pero para presentar resultados o crear reportes, necesitamos un **formato ancho** (donde las categorías se convierten en columnas).

Esta operación es esencial para crear matrices de correlación, tablas comparativas y dashboards ejecutivos. Pandas nos ofrece dos herramientas principales para esto: `pivot()` y `pivot_table()`.

````python
import pandas as pd
from pathlib import Path

# Configuración de rutas
DATA_PATH = Path(".")
````

## 1. El método `pivot()`

### 💡 Intuición
Imagina que tienes una lista de compras donde cada fila es un producto. `pivot()` es como tomar esa lista y reorganizarla para que los productos sean las columnas y las fechas sean las filas. 

**Regla de Oro**: `pivot()` NO puede manejar duplicados. Si tienes dos valores para la misma combinación de índice y columna, lanzará un error.

### 🛠️ Diagrama de Flujo
```
  [ Formato Largo (Long) ]                [ Formato Ancho (Wide) ]
  +---------+---------+-------+           +---------+-------+-------+
  |  Índice | Columna | Valor |
  +---------+---------+-------+
  |   1970  |   AFG   |  709  |  ------>  |   1970  |  709  |  522  |
  |   1970  |   ALB   |  522  |           +---------+-------+-------+
  +---------+---------+-------+
```

````python
# Carga de datos
df_gdp = pd.read_csv(DATA_PATH / "gdp.csv")

# Mostramos una muestra del formato largo
print("--- Formato Largo ---")
display(df_gdp[['country', 'year', 'gdppc']].head(10))
````

````python
# Reestructuración con pivot()
# index: Lo que queremos en las filas
# columns: Lo que queremos que se convierta en columnas
# values: El dato que llenará las celdas
df_pivot = df_gdp.pivot(index="year", columns="country", values="gdppc")

print("--- Formato Ancho (Pivot) ---")
display(df_pivot.head())
````

## 2. El método `pivot_table()`

### 💡 Intuición
A diferencia de `pivot()`, `pivot_table()` es una herramienta de **agregación**. Es ideal cuando tienes datos duplicados o quieres resumir la información (sumar, promediar, contar).

Si `pivot()` es una simple reorganización, `pivot_table()` es un resumen estadístico.

### 🛠️ Diagrama de Flujo
```
  [ Datos Crudos ]          [ Proceso de Agregación ]          [ Tabla Pivote ]
  +-------+-------+---+     +-----------------------+          +---------+-------+
  | Gen   | Prod  | Val|    | Calcular: sum(Val)    |          | Gen     | Suma  |
  +-------+-------+---+
  | Fem   | P1    | 10 |    | Agrupado por:         |  ------>  | Fem     |  50   |
  | Fem   | P1    | 40 |    | Gen y Prod            |          | Masc    |  40   |
  +-------+-------+---+
                                                               +---------+-------+
```

````python
# Carga de datos de ventas
df_sales = pd.read_excel(DATA_PATH / "supermarket_sales.xlsx")

print("--- Vista Previa de Ventas ---")
display(df_sales.head())
````

````python
# Caso 1: Suma total de ventas por género
# aggfunc='sum' indica que queremos sumar los valores
pivot_sum = df_sales.pivot_table(index="Gender", 
                                values=["Quantity", "Total"],
                                aggfunc="sum")

print("--- Suma de Cantidad y Total por Género ---")
display(pivot_sum)
````

````python
# Caso 2: Análisis multidimensional
# ¿Cuánto gastaron hombres y mujeres en cada línea de producto?
pivot_complex = df_sales.pivot_table(index="Gender",
                                    columns="Product line",
                                    values="Total",
                                    aggfunc="sum")

print("--- Gasto Total por Género y Categoría de Producto ---")
display(pivot_complex)
````

## ⚖️ Comparativa: `pivot()` vs `pivot_table()`

| Característica | `pivot()` | `pivot_table()` |
| :--- | :--- | :--- |
| **Propósito** | Reorganización simple | Resumen y Agregación |
| **Duplicados** | Lanza Error $\text{ValueError}$ | Los agrupa (agrega) |
| **Función** | No tiene `aggfunc` | Requiere/Permite `aggfunc` (sum, mean, etc.) |
| **Uso Ideal** | Datos ya limpios y únicos | Datos crudos con múltiples entradas |

## 📋 Resumen Rápido

1. **`df.pivot(index, columns, values)`**: Cambia la forma de los datos. Solo si no hay duplicados.
2. **`df.pivot_table(index, columns, values, aggfunc)`**: Resume los datos. Maneja duplicados mediante una función de agregación.
3. **Formato Largo $\rightarrow$ Ancho**: Proceso clave para reportes y análisis comparativos.
