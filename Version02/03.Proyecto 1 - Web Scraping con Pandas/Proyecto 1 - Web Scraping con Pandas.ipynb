# 🕸️ Proyecto 1: Web Scraping Automático con Pandas

🎯 **Objetivos**
- Extraer datos tabulares directamente de URLs utilizando `read_csv` y `read_html`.
- Manipular múltiples DataFrames provenientes de fuentes externas.
- Automatizar la descarga de datos de sitios web estructurados.
- Realizar una limpieza básica de datos obtenidos mediante scraping.

📖 **Introducción**

El Web Scraping es la técnica de extraer información de sitios web de forma automatizada. Pandas simplifica enormemente este proceso cuando los datos ya están en formatos tabulares (como archivos CSV remotos o tablas HTML). En este proyecto, aprenderemos a conectar nuestro código directamente con fuentes de datos en vivo en internet.

## 🌉 Puente Pedagógico: Scraping con Pandas

Tradicionalmente, el scraping requiere librerías complejas como BeautifulSoup o Selenium. Sin embargo, si el sitio web ofrece un enlace directo a un archivo o contiene tablas estándar, Pandas puede hacer todo el trabajo pesado en una sola línea de código.

### Flujo de Trabajo del Proyecto
```
[ URL del Sitio ] --> [ Pandas (read_csv / read_html) ] --> [ Limpieza de Datos ] --> [ Análisis Final ]
       |                         |
 (Identificar Datos)       (Extracción Automática)
```

````python
import pandas as pd
import numpy as np

# Configuración para mostrar más columnas si es necesario
pd.set_option('display.max_columns', None)
````

## 1. Extracción de CSV desde URLs

💡 **Intuición**: Muchos sitios de estadísticas (como football-data.co.uk) publican sus bases de datos en archivos CSV accesibles mediante una dirección web. Pandas puede tratar esa URL como si fuera un archivo local en tu computadora.

````python
# 1. Definir la URL del dataset (Estadísticas de la Premier League 21/22)
url_premier = 'https://www.football-data.co.uk/mmz4281/2122/E0.csv'

# 2. Leer directamente el CSV desde la web
df_premier = pd.read_csv(url_premier)

# 3. Inspección inicial
print(f"Partidos cargados: {len(df_premier)}")
df_premier.head()
````

## 2. Renombrado y Selección de Datos

💡 **Intuición**: Los datos de scraping suelen tener nombres de columnas crípticos (ej. `FTHG` para Full Time Home Goals). Para que nuestro análisis sea legible, debemos renombrar las columnas clave y seleccionar solo lo que necesitamos.

````python
# Renombrar columnas para mejor legibilidad
df_premier.rename(columns={
    'Date': 'Fecha',
    'HomeTeam': 'Local',
    'AwayTeam': 'Visitante',
    'FTHG': 'Goles_Local',
    'FTAG': 'Goles_Visitante'
}, inplace=True)

# Seleccionar subconjunto de columnas relevantes
df_resumen = df_premier[['Fecha', 'Local', 'Visitante', 'Goles_Local', 'Goles_Visitante']]
df_resumen.head()
````

## 3. Extracción de Tablas HTML (Simplificado)

💡 **Intuición**: Wikipedia y otros sitios de noticias usan tablas `<table>` de HTML. El método `read_html()` de Pandas escanea toda la página y devuelve una lista con todas las tablas encontradas convertidas a DataFrames.

````python
# Ejemplo: Extraer tablas de Wikipedia (Copa del Mundo)
try:
    url_wiki = 'https://en.wikipedia.org/wiki/List_of_FIFA_World_Cup_finals'
    tablas = pd.read_html(url_wiki)
    
    # Mostrar cuántas tablas se encontraron
    print(f"Se encontraron {len(tablas)} tablas en la página.")
    
    # La tabla de finales suele ser la número 3 o 4 (ajustar según la página)
    df_finales = tablas[3]
    df_finales.head()
except Exception as e:
    print(f"No se pudo realizar el scraping: {e}")
````

📋 **Resumen de Scraping**

| Función | Uso Ideal | Ventaja |
|---------|-----------|---------|
| `pd.read_csv(url)` | Archivos de datos planos en servidores remotos | Carga instantánea y directa |
| `pd.read_html(url)` | Tablas visibles en páginas web (Wikipedia, Forbes) | No requiere parseo manual de HTML |
| `pd.read_json(url)` | APIs que entregan datos en formato JSON | Ideal para integración con web services |
