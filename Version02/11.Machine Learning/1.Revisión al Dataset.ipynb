## 1. Introducción al EDA

Antes de aplicar cualquier algoritmo de Machine Learning, es imperativo realizar un **EDA** (*Exploratory Data Analysis*). El EDA nos permite responder preguntas críticas:
- ¿Tienen sentido los valores? (Ej: ¿Hay casas con 0 habitaciones?)
- ¿Cuál es la escala de los datos? (¿Estamos hablando de miles o millones?)
- ¿Existe una correlación visual entre las variables?

Sin esta fase, corremos el riesgo de entrenar modelos sobre datos sesgados o erróneos.

````python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('Boston House Prices.csv')
df_boston = pd.read_csv(DATA_PATH)
df_boston.head()
````

## 3. El Puente Pedagógico: Estadísticas Descriptivas

### ¿Qué nos dice `.describe()`?
La función `.describe()` es la "radiografía" del dataset. Nos entrega métricas clave:
- **Mean (Media)**: El promedio. Nos da una idea del centro de los datos.
- **Std (Desviación Estándar)**: Qué tan dispersos están los datos respecto a la media.
- **Min/Max**: Los límites extremos. Útil para detectar errores de captura (ej: precios negativos).
- **25%, 50%, 75% (Cuartiles)**: Nos ayudan a entender la distribución y detectar la presencia de *outliers*.

#### Flujo de Trabajo de EDA
```
  [ Raw Data ] --> [ Data Audit ] --> [ Statistical Summary ] --> [ Insights ]
                        |                       |
                    shape/dtypes            describe()
                                         (Mean, Std, Min, Max)
```

````python
df_boston.describe()
````

## 📋 Resumen de Auditoría

| Métrica | Propósito | Interpretación |
|---|---|---|
| `.shape` | Tamaño | Cantidad de ejemplos y variables |
