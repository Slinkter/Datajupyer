# Regresión Lineal Múltiple con Statsmodels

## 🎯 Objetivos
- Evolucionar el modelo de regresión simple a uno múltiple incorporando más variables predictoras.
- Comprender el concepto de "Hiperplano de Regresión".
- Comparar la capacidad predictiva ($R^2$) entre modelos simples y múltiples.
- Analizar la significancia individual de cada variable predictora.

## 1. Introducción

En la vida real, un fenómeno rara vez depende de una sola causa. El precio de una casa no depende solo de cuántas habitaciones tiene, sino también de su ubicación, la antigüedad, la zona, etc.

La **Regresión Lineal Múltiple** extiende la ecuación simple a:
$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_n x_n + \epsilon$$

En nuestro caso, utilizaremos:
- $x_1$: `Rooms` (Habitaciones)
- $x_2$: `Distance` (Distancia al centro)

````python
import pandas as pd
import statsmodels.api as sm
from pathlib import Path

DATA_PATH = Path('Boston House Prices.csv')
df_boston = pd.read_csv(DATA_PATH)
````

## 2. Definición de Variables

A diferencia de la regresión simple, nuestra matriz de características $X$ ahora es un DataFrame con múltiples columnas.

````python
y = df_boston['Value']
X = df_boston[['Rooms', 'Distance']]

print("Variables predictoras seleccionadas:\n", X.columns.tolist())
````

## 3. El Puente Pedagógico: De la Línea al Plano

### La Dimensión de la Predicción
Cuando tenemos una sola variable, buscamos la **línea** que mejor ajusta los puntos. Cuando tenemos dos o más, ya no buscamos una línea, sino un **plano** (o un hiperplano en dimensiones superiores) que minimice la distancia a todos los puntos en el espacio 3D.

#### Representación Conceptual
```
  Regresión Simple                     Regresión Múltiple
     (2D Plane)                            (3D Space)
        ^
        |          /                      ^
        |        /                        |      / 
        |      /                          |    /  (Regression Plane)
        |    /                            |  / 
        |  /                              |/ 
        +--------> Rooms (X)              +------------> Rooms (X1)
                                                   \
                                                    \--> Distance (X2)
```

````python
# Agregar la constante para el intercepto beta_0
X_with_const = sm.add_constant(X)

# Ajustar el modelo OLS
model_mult = sm.OLS(y, X_with_const).fit()

print(model_mult.summary())
````

### Análisis de Resultados

1. **R-squared**: Notarás que el $R^2$ aumentó (ej: de 0.48 a 0.49). Esto indica que añadir la variable `Distance` mejoró la capacidad predictiva del modelo.
2. **Coeficientes**: 
   - El coeficiente de `Rooms` sigue siendo positivo (más habitaciones $\rightarrow$ más valor).
   - El coeficiente de `Distance` es probablemente positivo o negativo dependiendo del dataset, indicando cómo afecta la ubicación al precio.
3. **Significancia**: Verificamos que el p-value de ambas variables sea $< 0.05$.

````python
# Ejemplo de predicción para una casa con 7 habitaciones y 3 unidades de distancia
nueva_casa = [1, 7, 3] # [constante, rooms, distance]
prediccion = model_mult.predict(nueva_casa)

print(f"El valor predicho para la casa es: {prediccion[0]:.2f}")
````

## 📋 Resumen: Simple vs Múltiple

| Característica | Regresión Simple | Regresión Múltiple |
|---|---|---|
| **Predictoras** | 1 sola ($x$) | Varias ($x_1, x_2, ...$)| 
| **Geometría** | Línea recta | Plano / Hiperplano |
| **Complejidad** | Baja, fácil de visualizar | Alta, requiere análisis de colinealidad |
| **Poder Predictivo**| Limitado | Generalmente mayor (si las variables son relevantes) |
