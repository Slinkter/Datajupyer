# Regresión Lineal con Scikit-Learn

## 🎯 Objetivos
- Implementar un modelo de regresión lineal utilizando la librería `scikit-learn` (sklearn).
- Comprender la diferencia entre un enfoque estadístico (`statsmodels`) y un enfoque de aprendizaje automático (`sklearn`).
- Realizar predicciones sobre datos nuevos y evaluar el desempeño del modelo mediante la métrica $R^2$.

## 1. Introducción

En el ecosistema de Python, existen dos herramientas principales para la regresión lineal:

1. **Statsmodels**: Enfocada en la inferencia estadística. Proporciona tablas detalladas con p-values, intervalos de confianza y análisis de residuos. Es ideal para investigadores.
2. **Scikit-Learn**: Enfocada en la predicción y la producción. Su API es consistente (`fit` $ightarrow$ `predict`) y está optimizada para manejar grandes volúmenes de datos y pipelines complejos. Es el estándar para ingenieros de ML.

````python
import pandas as pd
from sklearn import linear_model
from pathlib import Path

DATA_PATH = Path('Boston House Prices.csv')
df_boston = pd.read_csv(DATA_PATH)
````

## 2. Definición de Datos

Al igual que en los notebooks anteriores, definimos nuestra variable dependiente ($y$) y nuestras variables independientes ($X$).

````python
y = df_boston['Value']
X = df_boston[['Rooms', 'Distance']]

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
````

## 3. El Puente Pedagógico: El Flujo de Trabajo de ML

A diferencia de `statsmodels`, `sklearn` no requiere que añadamos manualmente la constante (intercepto). El objeto `LinearRegression` se encarga de ello internamente.

#### Diagrama del Flujo de Predicción
```
  [ Features (X) ]  +  [ Target (y) ]
          |                  |
          v                  v
  +---------------------------------+
  |         SKLearn Model           |
  | (Optimiza beta_0 y beta_1... )  |
  +---------------------------------+
                   |
                   v
  [ New Data (X_new) ] --> [ .predict() ] --> [ Prediction (y_pred) ]
```

````python
# 1. Instanciar el modelo
lm = linear_model.LinearRegression()

# 2. Entrenar el modelo (ajustar la línea a los datos)
lm.fit(X, y)

print("Modelo entrenado exitosamente.")
````

## 4. Evaluación y Predicción

Ahora que el modelo "conoce" la relación entre las variables, podemos evaluarlo y usarlo para predecir valores.

````python
# Medir el desempeño con el coeficiente de determinación R^2
r2_score = lm.score(X, y)
print(f"R-squared: {r2_score:.4f}")

# Acceder a los parámetros aprendidos
print(f"Intercepto (beta_0): {lm.intercept_:.4f}")
print(f"Coeficientes (beta_n): {lm.coef_}")
````

````python
# Realizar predicciones para los primeros 5 registros
predictions = lm.predict(X)[:5]
print("Predicciones:\n", predictions)
````

## 📋 Resumen: Statsmodels vs Scikit-Learn

| Característica | Statsmodels | Scikit-Learn |
|---|---|---|
| **Objetivo** | Inferencia Estadística | Predicción / Producción |
| **Intercepto** | Debe añadirse manualmente | Automático |
| **Resultados** | Tabla detallada (`.summary()`) | Métricas simples (`.score()`) |
| **API** | Orientada a modelos | Orientada a objetos (`fit`/`predict`) |
