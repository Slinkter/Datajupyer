## 1. Introducción

La **Regresión Lineal Simple** es una técnica de Machine Learning supervisado que busca modelar la relación entre una variable independiente ($X$) y una variable dependiente ($Y$) mediante una línea recta.

La ecuación fundamental es:
$$y = \beta_0 + \beta_1 x + \epsilon$$

Donde:
- $\beta_0$ (**Intercepto**): El valor de $y$ cuando $x = 0$.
- $\beta_1$ (**Pendiente**): Cuánto cambia $y$ por cada unidad que aumenta $x$.
- $\epsilon$ (**Error**): La diferencia entre el valor real y la predicción.

````python
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path('Boston House Prices.csv')
df_boston = pd.read_csv(DATA_PATH)
````

## 2. Preparación de Variables

Para este ejemplo, queremos predecir el **Valor** (`Value`) de la casa basándonos únicamente en el número de **Habitaciones** (`Rooms`).

````python
y = df_boston['Value'] # Variable dependiente (Target)
X = df_boston['Rooms'] # Variable independiente (Feature)

df_boston.plot.scatter(x='Rooms', y='Value', alpha=0.5)
plt.title('Relación Habitaciones vs Valor')
plt.show()
````

## 3. El Puente Pedagógico: Ajustando el Modelo

### La Constante en Statsmodels
A diferencia de otras librerías como `scikit-learn`, `statsmodels` no añade automáticamente el intercepto ($\beta_0$) al modelo. Si no lo hacemos manualmente, la línea de regresión pasaría obligatoriamente por el origen $(0,0)$, lo cual sería incorrecto en la mayoría de los casos reales.

#### Concepto de la Línea de Mejor Ajuste
```
  Value (Y)
     ^
     |          /  <-- Línea de Regresión (y = mx + b)
     |        / 
     |  *   /  *    <-- Datos Reales (Observaciones)
     |    / * 
     |  / *   * 
     |/________________> Rooms (X)
```
El objetivo del algoritmo **OLS** (*Ordinary Least Squares*) es minimizar la suma de los cuadrados de las distancias verticales entre los puntos reales y la línea.

````python
# 1. Agregar la constante para calcular el intercepto beta_0
X_with_const = sm.add_constant(X)

# 2. Ajustar el modelo OLS
model = sm.OLS(y, X_with_const).fit()

# 3. Mostrar la tabla de resultados
print(model.summary())
````

### Interpretando la Tabla de Regresión

- **R-squared**: Indica qué porcentaje de la varianza de $Y$ es explicada por $X$. Un $R^2 = 0.48$ significa que el número de habitaciones explica el 48% del precio de la casa.
- **Coef (Rooms)**: La pendiente. Si el coeficiente es $9.1$, significa que por cada habitación adicional, el valor de la casa sube, en promedio, $9.1$ unidades.
- **P>|t| (p-value)**: Si es $< 0.05$, la variable es estadísticamente significativa.

````python
# Realizar predicciones sobre el mismo dataset
y_pred = model.predict(X_with_const)

plt.figure(figsize=(8, 5))
sns.scatterplot(x=X, y=y, alpha=0.5, label='Datos Reales')
sns.lineplot(x=X, y=y_pred, color='red', linewidth=2, label='Línea de Regresión')
plt.title('Regresión Lineal: Rooms vs Value')
plt.xlabel('Número de Habitaciones')
plt.ylabel('Valor de la Vivienda')
plt.legend()
plt.show()
````

## 📋 Resumen de Regresión Lineal

| Concepto | Significado | Impacto en el Modelo |
|---|---|---|
| **Intercepto** | Valor de Y cuando X=0 | Punto de partida de la recta |
| **Pendiente** | Cambio en Y por cada unidad de X | Dirección y fuerza de la relación |
| **R-Squared** | Coeficiente de Determinación | Calidad del ajuste del modelo |
| **p-value** | Significancia estadística | Determina si la variable es útil o es ruido |
