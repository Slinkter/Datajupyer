```python
c1 = df_laptop3["Price_euros"]>3000
c2 = (df_laptop3["Price_euros"]>2000) & (df_laptop3["Price_euros"]<=3000)
c3 = (df_laptop3["Price_euros"]>800) & (df_laptop3["Price_euros"]<=2000)
c4 = df_laptop3["Price_euros"]<=800
```

```python
conditions = [c1,c2,c3,c4]
choices = ["Too Expensive","Expensive","Affordable","Cheap"]
```

```python
df_laptop3["Price_tier"] = np.select(conditions, choices, default="Unknown")
```

```python
df_laptop3["Price_tier"].value_counts()
```
