```python
dataFramePopulation_raw  = pd.read_csv("./population_total.csv")
print(dataFramePopulation_raw)
```

```python
dataFramePopulation_clean = dataFramePopulation_raw.dropna(inplace=False) 
print(dataFramePopulation_raw)
print(dataFramePopulation_clean)
```

```python
pivot01 =  dataFramePopulation_clean.pivot(index="year",columns="country",values="population")
pivot01
```

```python
countryFilter = ["Ecuador","Peru","Bolivia","Chile"]
filterPivot01 = pivot01[countryFilter]
filterPivot01 
```

```python
filterPivot01.plot(kind="line" ,xlabel="Year",ylabel="Population",title="Poupulation 1955-2020",figsize=(12,4))
```

```python
filterPivot01_2020 = filterPivot01[filterPivot01.index.isin([2020])]
```

```python
filterPivot01_2020
```

```python
filterPivot01_2020 = filterPivot01_2020.T
```

```python
filterPivot01_2020
```

```python
filterPivot01_2020.plot(kind="bar" ,xlabel="Country",ylabel="population" ,title="Population in 2020" ,color="red")
```

```python
df_pivot_sample = filterPivot01[filterPivot01.index.isin([1980, 1990, 2000, 2010, 2020])]
```
