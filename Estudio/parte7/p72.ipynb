```python
import pandas as pd
```

```python
dataFramePopulation = pd.read_csv("./population_total.csv")
```

```python
dataFramePopulation
```

```python
dataFramePopulation_raw = dataFramePopulation.dropna(inplace=False) 
```

```python
df_pivot =  dataFramePopulation_raw.pivot(index="year",columns="country",values="population")
df_pivot
```

```python
selectedCountry = ["Ecuador","Peru","Bolivia","Chile"]
df_pivot = df_pivot[selectedCountry]
df_pivot 
```

```python
df_pivot.plot(kind="line" ,
              xlabel="Year",ylabel="Population",
              title="Poupulation 1955-2020",figsize=(8,4))
```

```python
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python
df_pivot_2020
```

```python
filterPivot01_2020 = df_pivot_2020.T
```

```python
filterPivot01_2020
```

```python
filterPivot01_2020.plot(kind="bar" ,
                        xlabel="Country",ylabel="population" ,
                        title="Population in 2020" ,color="orange")
```
