```python
import time
```

```python
a = time.time() + 0.2
b = time.time()
resultado = b-a
print("--------")
print(f" {resultado:.10f} seg.")
```

```python
# Valor
valor = 1000000
# normal list

lista_normal = list(range(valor))
startTime = time.time()
lista_normal = [x * 2 for x in lista_normal]
endTime = time.time()
resultTime = endTime - startTime
print(f"Lista: {resultTime:.4f} seg.")
# numpy list 
arr = np.arange(valor)
startTime2 = time.time()
arr = arr * 2
endTime2 = time.time()
print(f"Lista: {endTime2 - startTime2:.4f} seg.")


```
