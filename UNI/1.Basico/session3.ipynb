```python
i= 0.0375
MontoInicial = float(input("Ingresa el monto inicial: "))
tiempo = int(input("ingresa el tiempo , en años: "))
#
print("El Monto incial es : ", MontoInicial)
interes = MontoInicial * i * tiempo
print("El interes es: ", interes)
#
MontoFinal = MontoInicial + interes
print("El monto final es : ", MontoFinal)


```
