```python
l1 = float(input("ingresa lado 1 "))
l2 = float(input("ingresa lado 2 "))
l3 = float(input("ingresa lado 3 "))

s = (l1+l2+l3)/2
area = (s*(s-l1)*(s-l2)*(s-l3))**(1/2)

print("Calculo del area de un triangulo (Formula de Heron)")
print("---------------------------------------------------")
print()
print("Primer lado ", l1)
print("Segundo lado ", l2)
print("Tercer lado ", l3)
print("Area : %.3f" %(area))
print("---------------------------------------------------")
print("s : %.3f" %(s))

```

```python
# Ejercicio 1
entrada_peso = input("Ingresa tu peso en kilogramos (kg): ")
entrada_altura = input("Ingresa tu altura en metros (m): ")

try:
    # Normalizar coma decimal a punto
    peso = float(entrada_peso.replace(",", "."))
    altura = float(entrada_altura.replace(",", "."))

    if peso > 0 and altura > 0:
        imc = peso / (altura ** 2)

        # Rangos continuos sin huecos ni condiciones redundantes
        if imc < 18.5:
            msj = "BAJO PESO"
        elif imc < 25.0:
            msj = "NORMAL"
        elif imc < 30.0:
            msj = "SOBREPESO"
        else:
            msj = "OBESIDAD"

        # Salida única y limpia
        print(f"""
    Tu IMC es    : {imc:.2f}
    Tu peso es   : {peso:.2f} kg
    Tu altura es : {altura:.2f} m
    Tu estado es : {msj}
        """)

    else:
        print("El peso y la altura deben ser valores mayores a 0.")

except ValueError:
    print("Error: Ingresaste un valor no numérico o inválido. Usa números (ej. 70 o 1.75).")
```
