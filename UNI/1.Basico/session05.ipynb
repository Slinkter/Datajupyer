```python
#crear tamano de la lista
# ingresa datos de la lista

listaN = []
n = int(input('ingresar en tamanio de la lista'))
i =0
while  i < n:
    dato = int(input('ingresa tu dato  rango[1-6]'))
    if (dato<=0 and dato>=7):
        print('fuera de rango')
        continue
    listaN.append(dato)
    i=i+1


print(listaN)
listaN.clear()
print(listaN)
del listaN
print(listaN)
```
