```python
palabras_clave = ["error", "warning", "timeout", "failed", "exception"]
conteo_eventos = dict.fromkeys(palabras_clave,0)
print(conteo_eventos)
```

```python
logs_crudos = [
    "Sistema iniciado correctamente",
    "ERROR  : fallo de conexion en el modulo principal",
    "Advertencia : Uso de memoria alta",
    "Timeout en la solicitud a la base de datos",
    "Proceso completado con exito",
    "Failed : Autenticacion denegada para usuario Arellano",
    "warning : El sistema de almacenamiento esta aun 95% de capacidad",
    "Excepcion no controlada en el hilo 4",
    "Error : falta de token"
]
```

```python
for mesaje in logs_crudos:
    msj_clean = mesaje.lower()
    for clave in conteo_eventos:
        if clave in msj_clean:
            conteo_eventos[clave]+=1

print(conteo_eventos)

func_print={k:v for k,v in conteo_eventos.items() if v>0}
print(func_print)

for evento,cantidad in func_print.items():
    print(f"[{cantidad}] incidencias de tipo {evento.upper()}")

```

```python
import random as rnd 
lanzamiento  =[rnd.randint(1,6) for _ in range(200)]
print(lanzamiento)
print(len(lanzamiento))
# forma no adecuado
conteo = {}
for number in lanzamiento:
        if number in conteo:
            conteo[number] +=1
        else:
            conteo[number] = 1

print(conteo)

```

```python
conteo_avanzado = {}
for numbertemp in lanzamiento:
    conteo_avanzado[numbertemp] = conteo_avanzado.get(numbertemp,0) +1

print(conteo_avanzado)
```

```python
config_antigua = {"host": "localhost", "puerto":8080, "debug":True, "timeout":30}
config_nueva = {"host": "192.168.1.10", "puerto" : 8080, "debug":False, "reintentos":3}
print(config_antigua)
print(config_nueva)
```

```python
interseccion = config_antigua & config_nueva
print(interseccion)
```

```python
interseccionKey = config_antigua.keys() & config_nueva.keys()
print(interseccionKey)
```

```python
interseccionKey = config_antigua.keys() - config_nueva.keys()
print(interseccionKey)
```

```python
interseccionKey = config_nueva.keys() - config_antigua.keys()
print(interseccionKey)
```

```python
lecturas_sensores = {
    "sensorA" : 25.4,
    "sensorB" : -999.0,
    "sensorC" : 26.1,
    "sensorD" : -999.0
}
```
