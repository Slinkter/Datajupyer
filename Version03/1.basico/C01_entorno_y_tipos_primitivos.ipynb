## Analogía inicial: ¿Por qué importan los tipos de datos?

Imagina que viajas con maletas de distintos tamaños: una caja de zapatos, una mochila y un contenedor marítimo. Cada una guarda cosas de un tipo y tamaño: no meterías un elefante en una caja de zapatos ni usarías un contenedor para guardar un lápiz.

Python funciona igual. Cada **tipo de dato** es como una maleta con reglas propias:

- Los **números** (`int`, `float`) guardan cantidades y se pueden sumar o multiplicar.
- Los **strings** guardan texto y se pueden concatenar o buscar dentro de ellos.
- Los **booleanos** solo admiten dos valores: `True` o `False` (sí o no).

Si intentas sumar un número con un texto sin convertirlo, Python te lanzará un error, igual que si intentaras meter el elefante en la caja de zapatos. Conocer los tipos te permite elegir la "maleta" correcta para cada tarea y escribir programas que funcionen a la primera.

## 1. El entorno Jupyter

Jupyter es un entorno interactivo donde puedes ejecutar código por celdas. Cada celda es un bloque independiente y el resultado se muestra justo debajo.

### 1.1 `print()`

La función `print()` muestra texto o valores en pantalla. Sin ella, solo se muestra el resultado de la **última** expresión de la celda.

````python
print("Hola, esto es Python")
print("Puedes imprimir varios valores", 42, 3.14, "juntos")
````

### 1.2 Comentarios

Los comentarios comienzan con `#` y son ignorados por Python. Son útiles para documentar tu código.

````python
# Esto es un comentario: no se ejecuta
edad = 30  # los comentarios también pueden ir al final de una línea

# Puede combinar Python 3.12+ con type hints
nombre: str = "Ana"
print(f"{nombre} tiene {edad} años")
````

### 1.3 F-strings con `=` para depurar (Python 3.8+)

Agregar `=` al final dentro de una f-string imprime el nombre de la variable y su valor, muy útil para depurar sin escribir dos veces la variable.

````python
precio = 99
cantidad = 4

print(f"{precio = }")
print(f"{cantidad = }")
````

## 2. Tipos numéricos

Python maneja tres tipos numéricos principales:

- **`int`**: números enteros (sin decimales).
- **`float`**: números decimales (punto flotante).
- **`complex`**: números complejos (parte real + imaginaria, ej. `2+3j`).

````python
entero: int = 42
decimal: float = 3.14159
complejo: complex = 2 + 3j

print(type(entero))   # <class 'int'>
print(type(decimal))  # <class 'float'>
print(type(complejo)) # <class 'complex'>
````

### 2.1 Operaciones aritméticas

Los operadores básicos son `+`, `-`, `*`, `/`, `**` (potencia) y `%` (módulo). La precedencia sigue las reglas matemáticas clásicas.

````python
a = 10
b = 3

print(f"Suma: {a + b}")
print(f"Resta: {a - b}")
print(f"Multiplicación: {a * b}")
print(f"División: {a / b}")
print(f"Potencia: {a ** b}")
print(f"Módulo: {a % b}")
````

### 2.2 Tipos de división

- `/` siempre devuelve un `float`.
- `//` (división entera) redondea hacia abajo y devuelve un `int`.
- `%` (módulo) devuelve el resto de la división entera.

````python
print(f"División normal: {7 / 2}  -> {type(7 / 2)}")
print(f"División entera: {7 // 2} -> {type(7 // 2)}")
print(f"Módulo: {7 % 2} -> {type(7 % 2)}")
print(f"Negativos: {-7 // 2} (redondea hacia abajo)")
````

### 2.3 El operador walrus `:=` (Python 3.8+)

El operador walrus `:=` asigna un valor a una variable dentro de una expresión. Es ideal para evitar calcular algo dos veces.

````python
# Sin walrus: calculamos len dos veces
texto = "murcielago"
if len(texto) > 5:
    print(f"El texto tiene {len(texto)} letras")

# Con walrus: calculamos una sola vez y reutilizamos
if (n := len(texto)) > 5:
    print(f"El texto tiene {n} letras (calculado una sola vez)")
````

## 3. Strings

Los strings (cadenas de texto) se crean con comillas simples, dobles o triples. Son **inmutables**: no se pueden cambiar carácter a carácter, pero sí crear versiones nuevas.

````python
simple = 'comillas simples'
doble = "comillas dobles"
triple = """texto
multilínea
con saltos"""

print(simple)
print(doble)
print(triple)
````

### 3.1 F-strings avanzados

Los f-strings permiten insertar expresiones dentro del texto y dar formato (anchura, decimales, alineación).

````python
precio = 19.995
nombre = "café"

# Ancho, alineación y decimales
print(f"{nombre:>10}")   # alinea a la derecha (10 espacios)
print(f"{nombre:<10}|")  # alinea a la izquierda
print(f"{precio:.2f}")   # dos decimales
print(f"{precio:08.2f}") # rellena con ceros

# Expresiones directas
print(f"El total es {precio * 2:.2f} euros")
````

### 3.2 Métodos útiles

Los strings tienen muchos métodos incorporados: `.upper()`, `.lower()`, `.strip()`, `.split()`, `.join()` y muchos más.

````python
frase = "  Hola Mundo, Python es Genial  "

print(frase.upper())
print(frase.lower())
print(frase.strip())          # elimina espacios al inicio y final
print(frase.replace("Python", "Data"))
print(frase.split())          # separa por espacios -> lista
print(", ".join(["a", "b", "c"]))
print(frase.count("o"))      # cuántas veces aparece la 'o'
````

### 3.3 Slicing (cortar cadenas)

El slicing `[inicio:fin:paso]` permite extraer partes de un string. `inicio` y `fin` son opcionales y los índices negativos cuentan desde el final.

````python
palabra = "pythonprogramming"

print(palabra[0])       # 'p' primer carácter
print(palabra[-1])      # 'g' último carácter
print(palabra[0:6])     # 'python' primeros 6
print(palabra[6:])      # desde el índice 6
print(palabra[::2])     # cada 2 caracteres
print(palabra[::-1])    # invertida
````

### 3.4 Caracteres especiales y raw strings

Algunos caracteres necesitan escapar con `\` (`\n` salto de línea, `\t` tabulación). Los **raw strings** (prefijo `r`) no interpretan escapes, ideal para rutas y expresiones regulares.

````python
print("Línea 1\nLínea 2")
print("Columna 1\tColumna 2")
print("Comillas dentro: \"así\"")

# Raw string: no interpreta \n ni \t
ruta = r"C:\Users\datos\archivo.csv"
print(ruta)
````

## 4. Booleanos y truthy/falsy

El tipo `bool` solo tiene dos valores: `True` y `False`. En Python, **casi cualquier valor** puede interpretarse como verdadero o falso según su "truthiness".

````python
verdadero = True
falso = False

print(type(verdadero))
print(bool(1))    # True
print(bool(0))    # False
print(bool(""))   # False (string vacío)
print(bool("a"))   # True (string no vacío)
print(bool([]))   # False (lista vacía)
print(bool([1]))  # True (lista con elementos)
````

### 4.1 `is` vs `==`

- `==` compara **valores** (contenido).
- `is` compara **identidad** (si son el mismo objeto en memoria).

Para objetos pequeños Python puede reutilizar memoria (internado), así que usar `is` con números puede dar resultados engañosos. **Siempre usa `==` para comparar valores.**

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True  -> mismos valores
print(a is b)  # False -> objetos distintos en memoria

# Con None, `is` es la forma correcta
valor = None
print(valor is None)  # True
print(valor is not None)  # False
````

## 5. Conversión de tipos

Las funciones `int()`, `float()`, `str()` y `bool()` convierten un valor a otro tipo. Las conversiones automáticas pueden sorprender; conviene ser explícito para evitar errores.

````python
print(int(3.9))            # 3  (trunca, no redondea)
print(int("42"))           # 42
print(float("3.14"))       # 3.14
print(str(123))            # "123"
print(str(3.14))           # "3.14"
print(bool(0))             # False
print(bool(1))             # True
````

### 5.1 Errores comunes

Convertir un texto que no es numérico lanza `ValueError`. Es importante leer y entender los mensajes de error.

````python
# Este bloque está comentado para no detener la ejecución
# print(int("abc"))  # ValueError
# print(int("3.14")) # ValueError: no es un entero válido

# La forma segura: validar con try/except
try:
    numero = int("42")
    print(f"Conversión exitosa: {numero}")
except ValueError as e:
    print(f"Error: {e}")
````

## 6. Operadores

````python
# Aritméticos
print(5 + 3, 5 - 3, 5 * 3, 5 / 3, 5 // 3, 5 % 3, 5 ** 3)

# Comparación
print(5 == 5, 5 != 3, 5 > 3, 5 < 3, 5 >= 5, 5 <= 4)
````

````python
# Lógicos
print(True and False)   # False
print(True or False)    # True
print(not True)         # False

# Identidad
x = 10
y = 10
print(x is y)           # True (enteros pequeños están internados)
print(x is not y)       # False

# Pertenencia
print("py" in "python")  # True
print("z" not in "python") # True
print(2 in [1, 2, 3])   # True
````

Escribe código que:

1. Pida una temperatura en grados Celsius (puedes asignar una variable en lugar de usar `input`).
2. La convierta a Fahrenheit con la fórmula `F = C * 9/5 + 32`.
3. Use un f-string con el modificador `=` para mostrar el valor de Celsius y el resultado con 1 decimal.
4. Imprima `"Hace frío"` si es menor a 10, `"Hace calor"` si es mayor a 30, y `"Temperatura agradable"` en caso contrario.

````python
# Tu solución aquí (independiente)
celsius = 25

# Escribe tu código debajo
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius=} -> {fahrenheit:.1f} °F")

if celsius < 10:
    print("Hace frío")
elif celsius > 30:
    print("Hace calor")
else:
    print("Temperatura agradable")
````
