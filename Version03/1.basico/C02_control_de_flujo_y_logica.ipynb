## Analogía: el programa como un semáforo y una receta

Imagina que estás conduciendo y llegas a un cruce con un **semáforo**:

```
               ┌─────────────┐
    luz roja → │ PARAR       │
               └─────────────┘
    luz ámbar → ┌──────────────────┐
               │ DECELERAR        │
               └──────────────────┘
    luz verde →┌───────────────────┐
               │ AVANZAR          │
               └───────────────────┘
```

El semáforo es una **decisión**: según el color de la luz, tomas una acción distinta. Eso es exactamente un `if/elif/else`.

Ahora piensa en una **receta de cocina**:

```
1. Precalienta el horno
2. Mezcla los ingredientes
3. Repite: bate durante 3 minutos
4. Vierte en el molde
5. Hornea a 180° por 30 minutos
```

Las recetas tienen **pasos que se repiten** (bate *hasta que* esté cremoso) y **condiciones** (si no tienes harina, usa maicena).

En resumen:
- **Decisiones** (semáforo) → `if`, `elif`, `else`, `match-case`.
- **Repeticiones** (receta) → `for` y `while`.

Un programa es una serie de estas decisiones y repeticiones ordenadas lógicamente.

## 1. Condicionales

### 1.1 `if`, `elif`, `else`

La estructura básica: se evalúa una condición; si es verdadera se ejecuta ese bloque, si no se prueban las demás ramas.

```
if condicion1:
    accion A
elif condicion2:
    accion B
else:
    accion C
```

Ejemplo con un semáforo:

````python
color_semaforo = "verde"

if color_semaforo == "rojo":
    print("🚦 Detente.")
elif color_semaforo == "ambar":
    print("🚦 Decelera, se acerca el rojo.")
else:
    print("🚦 Avanza con cuidado.")
````

**Importante:** los bloques se delimitan por **indentación** (4 espacios), no por llaves. La condición debe ser evaluable como `True` o `False`.

````python
edad = 17

if edad >= 18:
    print("Mayor de edad: puedes votar.")
elif edad >= 13:
    print("Adolescente: tienes restricciones.")
else:
    print("Menor de edad.")
````

### 1.2 Operador ternario

Un `if/else` en una sola línea. Útil para asignaciones simples:

```
valor = resultado_si_verdadero if condicion else resultado_si_falso
```

````python
edad = 21
tipo = "adulto" if edad >= 18 else "menor"
print(f"Edad {edad}: {tipo}")
````

### 1.3 `match-case` (Python 3.10+)

Equivalente moderno del `switch/case`. Compara un valor con varios patrones.

````python
dia = "viernes"

match dia:
    case "lunes":
        print("Inicio de semana.")
    case "viernes":
        print("Fin de semana laboral.")
    case "sabado" | "domingo":
        print("Descanso.")
    case _:
        print("Día normal.")
````

### Diagrama ASCII: flujo `if/elif/else`

```
                 ┌─────────────┐
                 │   Inicio    │
                 └──────┬──────┘
                        │
                        ▼
                 ┌─────────────┐
        ┌────────┤ condicion 1 │────────┐
        │  Sí    └─────────────┘   No   │
        ▼                                ▼
   ┌────────────┐                 ┌─────────────┐
   │  Acción A  │        Sí ┌────┤ condicion 2 │────┐  No
   └────────────┘            ▼    └─────────────┘    ▼
                      ┌────────────┐          ┌────────────┐
                      │  Acción B  │          │  Acción C  │
                      └────────────┘          └────────────┘
                             │                     │
                             ▼                     ▼
                       ┌────────────────────┐
                       │      Contínua       │
                       └────────────────────┘
```

Solo **una** de las ramas se ejecuta. Si ninguna condición se cumple, se ejecuta el `else` (si existe).

## 2. Operadores lógicos y cortocircuito

Los operadores `and`, `or` y `not` combinan condiciones. Su importancia radica en el **cortocircuito** (evaluación perezosa):

- `a and b`: si `a` es `False`, **no evalúa `b`** (resultado `False`).
- `a or b`: si `a` es `True`, **no evalúa `b`** (resultado `True`).
- `or` devuelve el **último valor evaluado**, no solo `True/False`.

````python
edad = 25
tiene_ine = True

if edad >= 18 and tiene_ine:
    print("Puede votar.")
else:
    print("No puede votar.")
````

````python
usuario = ""  # cadena vacía es falsy
nombre = usuario or "Invitado"
print(f"Hola, {nombre}")
````

### Cortocircuito en acción

Las funciones `a()` y `b()` imprimen y devuelven `False` y `True` respectivamente. Observa cuál se evalúa con cada operador.

````python
def a():
    print("  Evaluando a (False)")
    return False

def b():
    print("  Evaluando b (True)")
    return True

print("and:")
resultado_and = a() and b()
print(f"  Resultado: {resultado_and}")

print("or:")
resultado_or = a() or b()
print(f"  Resultado: {resultado_or}")
````

Observa que en `a() and b()`, como `a()` es `False`, **nunca se llama a `b()`**. En cambio en `a() or b()`, al ser `a()` `False`, **sí** se evalúa `b()` para decidir el resultado. Eso es el cortocircuito (evaluación perezosa).

**Aplicación práctica:** usa cortocircuito para evitar errores, por ejemplo comprobar que una lista no está vacía antes de acceder a su primer elemento:

````python
lista = [10, 20, 30]

# La condición `lista` evita evaluar lista[0] si está vacía (IndexError)
if lista and lista[0] > 15:
    print(f"El primer elemento {lista[0]} es mayor que 15.")
else:
    print("Lista vacía o primer elemento <= 15.")
````

### Operador `not`

Invierte el valor de verdad de una condición.

````python
activo = False

if not activo:
    print("La cuenta está inactiva.")

print(not True)   # False
print(not 0)      # True (0 es falsy)
````

## 3. Bucle `for`

El bucle `for` itera sobre una secuencia: una lista, una cadena, un rango, un diccionario, etc. Por cada elemento ejecuta el bloque de código.

````python
frutas = ["manzana", "pera", "uva"]

for fruta in frutas:
    print(f"Me gusta la {fruta}")
````

### 3.1 `range()`

`range(inicio, fin, paso)` genera una secuencia de números. El `fin` es **exclusivo** (no se incluye).

````python
# de 0 a 4 (5 números)
for i in range(5):
    print(i, end=" ")
print()

# de 2 a 9, de 2 en 2
for i in range(2, 10, 2):
    print(i, end=" ")
print()

# cuenta regresiva
for i in range(3, 0, -1):
    print(i, end=" ")
print("¡Fuego!")
````

### 3.2 `enumerate()`

Necesitamos a menudo el **índice** junto con el elemento. `enumerate` lo entrega como pares `(índice, valor)`.

````python
alumnos = ["Ana", "Luis", "Carla", "Pedro"]

for indice, alumno in enumerate(alumnos):
    print(f"{indice + 1}. {alumno}")
````

### 3.3 `zip()`

`zip` empareja elementos de dos o más secuencias por posición.

````python
nombres = ["Ana", "Luis", "Carla"]
notas = [9.5, 7.0, 8.2]

for nombre, nota in zip(nombres, notas):
    print(f"{nombre}: {nota}")
````

### 3.4 `for-else`

El bloque `else` de un `for` se ejecuta **solo si** el bucle termina sin ser interrumpido por `break`. Es útil para búsquedas: si no encontramos nada, hacemos algo.

````python
numeros = [2, 4, 6, 7, 8]
objetivo = 11

for n in numeros:
    if n == objetivo:
        print(f"¡Encontrado {objetivo}!")
        break
else:
    print(f"No se encontró {objetivo} en la lista.")
````

El bucle repite el cuerpo mientras queden elementos por recorrer, y sale cuando la secuencia se agota.

## 4. Bucle `while`

A diferencia de `for`, el `while` repite un bloque **mientras una condición sea verdadera**. Se usa cuando no sabemos de antemano cuántas iteraciones habrá.

````python
contador = 1

while contador <= 5:
    print(f"Tercio {contador}")
    contador += 1

print("¡Hora de la cena!")
````

**¡Cuidado con los bucles infinitos!** Si la condición nunca se vuelve falsa, el bucle no termina. Asegúrate de modificar algo dentro del bucle que eventualmente haga falsa la condición.

### 4.1 `break`, `continue`, `pass`

- `break`: **termina** el bucle inmediatamente.
- `continue`: **salta a la siguiente iteración** (ignora el resto del cuerpo).
- `pass`: no hace nada; sirve como marcador de posición para código futuro.

````python
print("break: detiene al encontrar el 3")
for i in range(1, 8):
    if i == 3:
        break
    print(i, end=" ")

print("\ncontinue: omite los números pares")
for i in range(1, 8):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
````

### 4.2 `while-else`

Igual que `for-else`: el `else` se ejecuta si el `while` termina con su condición volviéndose falsa, **no** por un `break`.

````python
objetivo = 100
suma = 0
n = 1

while suma < objetivo:
    suma += n
    n += 1
else:
    print(f"La suma acumulada {suma} alcanzó el objetivo {objetivo}.")
````

## 5. Comprensiones básicas

Las **comprensiones de listas** permiten crear y filtrar listas de forma compacta y legible.

```
[expresión for elemento in secuencia if condición]
```

````python
numeros = [1, 2, 3, 4, 5]

cuadrados = [n ** 2 for n in numeros]
print("Cuadrados:", cuadrados)
````

````python
# Filtrar: solo pares, elevados al cubo
pares_cubo = [n ** 3 for n in range(1, 11) if n % 2 == 0]
print("Pares al cubo:", pares_cubo)
````

### `filter` y `map`

- `map(función, iterable)`: aplica una función a cada elemento.
- `filter(condición, iterable)`: conserva solo los elementos que cumplen la condición.

En Python moderno, las comprensiones suelen ser más legibles que `map`/`filter`, pero es útil conocerlos.

````python
notas = [5.5, 8.0, 9.2, 6.5, 3.0]

# map: redondear cada nota
redondeadas = list(map(lambda n: round(n), notas))
print("Redondeadas:", redondeadas)

# filter: aprobados (>= 6)
aprobados = list(filter(lambda n: n >= 6, notas))
print("Aprobados:", aprobados)
````

## 6. Patrones comunes

Estos patrones aparecen constantemente en programación real. Los construimos paso a paso.

### 6.1 Buscar un elemento

Recorrer una lista y detenernos cuando encontramos el objetivo.

````python
contactos = ["Ana", "Luis", "Carla", "Pedro", "Sofía"]
buscado = "Carla"
encontrado = False

for c in contactos:
    if c == buscado:
        encontrado = True
        break

print(f"¿Está {buscado}? {encontrado}")
````

### 6.2 Contar ocurrencias

Contar cuántas veces aparece un valor.

````python
respuestas = ["si", "no", "si", "si", "tal vez", "no"]
contador_si = 0

for r in respuestas:
    if r == "si":
        contador_si += 1

print(f"Veces que respondieron 'si': {contador_si}")
````

### 6.3 Acumular valores

Sumar o combinar valores a lo largo de una iteración.

````python
ventas = [120.5, 80.25, 210.0, 45.75]
total = 0.0

for v in ventas:
    total += v

print(f"Ventas totales: ${total:.2f}")
````

### 6.4 Validación

Comprobar que todos (o algunos) los elementos cumplen una propiedad.

````python
edades = [12, 15, 9, 14]

todos_menores = True
for e in edades:
    if e >= 18:
        todos_menores = False
        break

print(f"¿Todos son menores de edad? {todos_menores}")
````

## 7. Tabla de referencia: sentencias de control

| Sentencia | Uso | Ejemplo breve |
|-----------|-----|---------------|
| `if` / `elif` / `else` | Decidir entre ramas | `if x > 0: print("positivo")` |
| Ternario | `if/else` en una línea | `tipo = "par" if n % 2 == 0 else "impar"` |
| `match-case` | Comparar patrones | `match op: case "+": ...` |
| `and` / `or` / `not` | Combinar condiciones | `if a and not b: ...` |
| `for` | Iterar sobre secuencia | `for x in lista: ...` |
| `range(inicio, fin, paso)` | Generar números | `for i in range(10): ...` |
| `enumerate(sec)` | Índice + valor | `for i, v in enumerate(l): ...` |
| `zip(sec1, sec2)` | Emparejar secuencias | `for a, b in zip(x, y): ...` |
| `while` | Repetir mientras condición | `while n > 0: n -= 1` |
| `break` | Salir del bucle | `if x: break` |
| `continue` | Saltar a la siguiente iteración | `if x: continue` |
| `pass` | Marcador vacío | `pass` |
| `else` (bucle) | Se ejecuta si no hubo `break` | `for ... else: ...` |
| Comprensión de lista | Crear/filtrar listas compactas | `[x*2 for x in l if x>0]` |

### Ejercicio independiente: Número primo

Escribe un programa que determine si un número dado es primo. Un número primo solo es divisible entre 1 y sí mismo.

**Requisitos:**
- Usa un bucle `for` con `range`.
- Usa `break` cuando detectes un divisor.
- Usa `for-else` para confirmar que es primo.

Pista: prueba divisores desde 2 hasta `numero - 1`.

````python
numero = 29

# Tu código aquí
es_primo = True
for div in range(2, numero):
    if numero % div == 0:
        es_primo = False
        break

if numero > 1 and es_primo:
    print(f"{numero} es primo")
else:
    print(f"{numero} no es primo")
````
