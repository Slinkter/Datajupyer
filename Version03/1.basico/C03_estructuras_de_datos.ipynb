## 1. Listas

💡 **Intuición**: Una lista es una **secuencia ordenada y mutable** de elementos. La llamamos "caja abierta" porque podemos agregar, eliminar y reorganizar sus contenidos libremente.

### Creación

```python
numeros = [1, 2, 3, 4, 5]
mixta = [1, "Python", 3.14, True]
vacia = []
```

````python
numeros = [10, 20, 30, 40, 50]
mixta = [1, "Python", 3.14, True]
vacia = []

print(numeros)
print(mixta)
print(vacia, "tamaño:", len(vacia))
````

### Índices y slicing

Los índices empiezan en `0`. Los índices negativos cuentan desde el final. El **slicing** `lista[inicio:fin:paso]` extrae un subconjunto (el índice `fin` es excluyente).

````python
numeros = [10, 20, 30, 40, 50]

print("Elemento 0:", numeros[0])       # 10
print("Último:", numeros[-1])          # 50
print("Slice 1:3:", numeros[1:3])     # [20, 30]
print("Slice cada 2:", numeros[::2])  # [10, 30, 50]
print("Invertida:", numeros[::-1])    # [50, 40, 30, 20, 10]
````

🛠️ **Diagrama: Slicing de una lista**
```
Índices:     0    1    2    3    4
           [ 10,  20,  30,  40,  50 ]
               |----numeros[1:3]----|
                      => [20, 30]
```

### Métodos de lista

| Método | Acción | Complejidad |
|--------|--------|-------------|
| `append(x)` | Agrega `x` al final | O(1) amortizado |
| `extend(iter)` | Agrega todos los elementos de `iter` | O(k) |
| `insert(i, x)` | Inserta `x` en la posición `i` | O(n) |
| `pop()` | Elimina y devuelve el último elemento | O(1) |
| `pop(i)` | Elimina el elemento en la posición `i` | O(n) |
| `remove(x)` | Elimina la primera aparición de `x` | O(n) |
| `sort()` | Ordena en el lugar | O(n log n) |
| `reverse()` | Invierte el orden en el lugar | O(n) |
| `index(x)` | Posición de la primera aparición de `x` | O(n) |
| `count(x)` | Cuenta apariciones de `x` | O(n) |

````python
frutas = ["manzana", "pera"]

frutas.append("uva")          # agrega al final
frutas.extend(["kiwi", "mango"])  # agrega varios
print("Tras append+extend:", frutas)

ultimo = frutas.pop()          # elimina y devuelve el último
print("pop() devolvió:", ultimo)
print("Lista restante:", frutas)

frutas.sort()
print("Ordenada:", frutas)

frutas.reverse()
print("Invertida:", frutas)
````

### Comprensiones de listas

Una **comprensión de lista** construye una lista nueva en una sola línea con una sintaxis declarativa: `[expresión for elemento in iterable if condición]`.

````python
numeros = [1, 2, 3, 4, 5]

cuadrados = [n ** 2 for n in numeros]
pares = [n for n in numeros if n % 2 == 0]
etiquetas = [f"Item {n}" for n in numeros]

print("Cuadrados:", cuadrados)
print("Pares:", pares)
print("Etiquetas:", etiquetas)
````

### Mutabilidad: modificar en el lugar vs. crear copia

La **mutabilidad** significa que el objeto puede cambiar sin crear una copia nueva. Es crucial para entender referencias y memoria.

````python
a = [1, 2, 3]
b = a          # b es UNA REFERENCIA al mismo objeto
b.append(4)

print("a:", a)  # ¡Cambió! Porque a y b apuntan a la misma lista
print("b:", b)
print("¿Son el mismo objeto?", a is b)

c = a.copy()   # copia independiente
c.append(99)
print("a (sin cambio):", a)
print("c (copia modificada):", c)
````

### Complejidad O(n) vs O(1)

💡 **Intuición**: Acceder a un elemento por índice es **O(1)** (instantáneo, sin importar el tamaño), pero **insertar o eliminar** en el medio requiere **O(n)** porque hay que desplazar los elementos posteriores.

```
O(1):  numeros[0]         -> acceso directo (puntero)
        numeros.append(x) -> añadido al final
        numeros.pop()     -> quitar el final

O(n):  numeros.insert(0, x)  -> desplaza todos los elementos
        numeros.pop(0)       -> desplaza todos los elementos
        x in numeros         -> búsqueda lineal
```

````python
grande = list(range(1_000_000))

# O(1): acceso por índice
print("Último elemento:", grande[-1])

# O(n): marcar si un valor está presente (búsqueda secuencial)
print("¿Está 999_999?", 999_999 in grande)
print("¿Está -5?", -5 in grande)
````

## 2. Tuplas

💡 **Intuición**: Una tupla es una **secuencia ordenada pero inmutable**. La llamamos "caja sellada": no puedes agregar, eliminar ni reordenar elementos una vez creada. Esta garantía de inmutabilidad la hace segura como clave de diccionario o para representar registros fijos.

```
list  = [ 1, 2, 3 ]   mutable -> admite cambios
tuple = ( 1, 2, 3 )   inmutable -> solo lectura
```

````python
punto = (3, 4)
print("Tupla:", punto)
print("Tipo:", type(punto))
print("Acceso por índice:", punto[0])

# Desempaquetado (unpacking)
x, y = punto
print(f"Coordenadas: x={x}, y={y}")

# Tupla de un solo elemento: requiere coma
un_solo = (5,)
print("Un elemento:", un_solo, type(un_solo))
````

````python
try:
    punto[0] = 99
except TypeError as e:
    print("Error esperado:", e)
````

### Named tuples

Un **named tuple** combina la inmutabilidad de la tupla con el acceso por nombre de campo, ideal para representar registros pequeños y legibles.

### Packing y unpacking

- **Packing**: agrupar varios valores en una tupla (`a, b, c = 1, 2, 3`).
- **Unpacking**: extraer los valores (`x, y = punto`).

````python
from collections import namedtuple

Punto = namedtuple("Punto", ["x", "y", "z"])
p = Punto(1, 2, 3)

print(p)
print("Acceso por nombre:", p.x, p.y, p.z)
print("Sigue siendo una tupla:", isinstance(p, tuple))
````

### ¿Cuándo usar tuplas vs listas?

| Criterio | Tupla | Lista |
|----------|-------|-------|
| Mutabilidad | Inmutable | Mutable |
| Velocidad | Ligeramente más rápida | Igual acceso, más costosa al modificar |
| Uso de memoria | Menor | Mayor |
| Clave de dict | ✅ Permitido | ❌ No permitido |
| Uso típico | Coordenadas, fechas, registros fijos | Colecciones que crecen o cambian |
| Ejemplo | `(lat, lon)`, `(año, mes, día)` | `ventas`, `nombres`, `items` |

**Regla práctica**: si sabes que el contenido no cambiará (como una fecha o coordenadas), usa tupla. Si el conjunto crecerá o se modificará, usa lista.

````python
# Tuplas como claves de diccionario (seguro por ser inmutables)
posiciones = {
    (0, 0): "origen",
    (1, 2): "casilla A",
    (3, 1): "casilla B",
}
print(posiciones[(1, 2)])

# Registro fijo de venta (no debería cambiar)
venta = ("PROD-001", "laptop", 2, 1250.00)
producto_id, nombre, cantidad, precio = venta
print(f"Total: {cantidad * precio:.2f} €")
````

## 3. Conjuntos (set)

💡 **Intuición**: Un conjunto es una **colección desordenada de elementos únicos**. La llamamos "estuche de fichas únicas" porque no admite duplicados y no tiene un orden definido. Su gran ventaja es la **búsqueda O(1)**: preguntar si un elemento existe es instantáneo, sin importar el tamaño.

```
list = [ 1, 2, 2, 3 ]  -> permite duplicados, mantiene orden
set  = { 1, 2, 3 }     -> SIN duplicados, SIN orden garantizado
```

````python
conjunto = {3, 1, 2, 2, 1}
print("Conjunto:", conjunto)          # {1, 2, 3} -> sin duplicados
print("Tipo:", type(conjunto))

conjunto.add(4)
conjunto.discard(2)   # elimina sin error si no existe
print("Tras add/discard:", conjunto)
````

### Operaciones de conjunto

| Operación | Operador | Método | Descripción |
|-----------|----------|--------|-------------|
| Unión | `A \| B` | `A.union(B)` | Elementos de A o B |
| Intersección | `A & B` | `A.intersection(B)` | Elementos en ambos |
| Diferencia | `A - B` | `A.difference(B)` | En A pero no en B |
| Diferencia simétrica | `A ^ B` | `A.symmetric_difference(B)` | En A o B, no en ambos |
| Subconjunto | `A <= B` | `A.issubset(B)` | ¿A dentro de B? |
| Superconjunto | `A >= B` | `A.issuperset(B)` | ¿A contiene a B? |

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Unión (A | B):", A | B)
print("Intersección (A & B):", A & B)
print("Diferencia (A - B):", A - B)
print("Dif. simétrica (A ^ B):", A ^ B)
````

````python
# 🛠️ Diagrama: Operaciones de conjunto
#        A = {1,2,3,4}        B = {3,4,5,6}
#
#   A - B: {1, 2}        A & B: {3, 4}        A | B: {1,2,3,4,5,6}
#
#            A      B             A    B             A    B
#          ┌────┐ ┌────┐       ┌────┐┌────┐       ┌────┐┌────┐
#          │1 2│3│4│ │5 6│       │1 2│3 4││5 6│      │1 2│3 4│5 6│
#          └────┘ └────┘       └────┘└────┘       └────┘└────┘
print("Ver diagrama en la celda de markdown")
````

### frozenset

Un **frozenset** es la versión inmutable de un set. No admite `add()`, `discard()` ni `clear()`, pero puede usarse como clave de diccionario o elemento de otro set (los sets normales no son hashables).

````python
fs = frozenset([1, 2, 3])
print("frozenset:", fs, type(fs))

try:
    fs.add(4)
except AttributeError as e:
    print("Error esperado:", e)

# Útil como clave de diccionario (es inmutable/hashable)
grupos = {frozenset({"Ana", "Luis"}): "equipo A", frozenset({"Sofía"}): "equipo B"}
print(grupos[frozenset({"Luis", "Ana"})])
````

### Deduplicación

Convertir una lista a set elimina los duplicados de forma eficiente. Recuerda que el orden se pierde; reconviene con `sorted()` si lo necesitas.

````python
datos = ["Madrid", "Barcelona", "Madrid", "Valencia", "Barcelona", "Madrid"]

unicos = list(set(datos))
print("Únicos (sin orden):", unicos)
print("Únicos ordenados:", sorted(unicos))
print("Cantidad única:", len(unicos))
````

### Complejidad O(1) de búsqueda

Preguntar `x in set` es **O(1)** gracias al *hash table*, mientras que `x in list` es **O(n)**. Para conjuntos grandes de datos sin orden, un set es mucho más rápido.

```
Buscar en lista:  x in [ ... n elementos ... ]  -> O(n)  recorre uno a uno
Buscar en set:    x in { ... n elementos ... }  -> O(1)  va directo por el hash
```

````python
import random

datos = list(range(500_000))
conjuntillo = set(datos)
blanco = 499_999

# O(n) en lista vs O(1) en set
print("En lista (O(n)):", blanco in datos)
print("En set (O(1)):", blanco in conjuntillo)
````

## 4. Diccionarios

💡 **Intuición**: Un diccionario mapea **claves únicas a valores**. Lo llamamos "archivador con etiquetas": cada clave (etiqueta) te lleva directamente a su valor (documento) en O(1). Es la estructura de datos más importante para el análisis de datos en Python.

````python
usuario = {
    "nombre": "Ana",
    "edad": 28,
    "ciudad": "Lima",
    "activo": True,
}

print("Diccionario:", usuario)
print("Acceso con []:", usuario["nombre"])
print("Acceso con .get():", usuario.get("ciudad"))
print("get con valor por defecto:", usuario.get("no_existe", "valor_generico"))
````

### Acceso: `[]` vs `.get()`

| Método | Comportamiento | Cuándo usar |
|--------|----------------|-------------|
| `d["clave"]` | Lanza `KeyError` si no existe | Cuando **debe** existir |
| `d.get("clave")` | Devuelve `None` si no existe | Cuando puede faltar |
| `d.get("clave", def)` | Devuelve `def` si no existe | Con valor por defecto |
| `d.setdefault(k, def)` | Devuelve valor y lo crea si falta | Para inicializar |

````python
votos = {}
nombres = ["Ana", "Luis", "Ana", "Sofía", "Luis", "Ana"]

for nombre in nombres:
    votos[nombre] = votos.get(nombre, 0) + 1

print("Conteo de votos:", votos)
````

### Métodos de diccionario

| Método | Descripción |
|--------|-------------|
| `d.keys()` | Vista de las claves |
| `d.values()` | Vista de los valores |
| `d.items()` | Vista de pares (clave, valor) |
| `d.update(otro)` | Fusiona/actualiza con otro dict |
| `d.pop(clave)` | Elimina y devuelve el valor |
| `d.popitem()` | Elimina y devuelve el último par |
| `d.clear()` | Vacía el diccionario |

````python
stock = {"laptop": 5, "mouse": 20, "teclado": 12}

print("Claves:", list(stock.keys()))
print("Valores:", list(stock.values()))
print("Pares:", list(stock.items()))

stock.update({"monitor": 8})
stock.update({"laptop": 4})   # actualiza el valor existente
print("Tras update:", stock)

vendido = stock.pop("mouse")
print("Pop mouse ->", vendido, "| Stock:", stock)
````

````python
# Iterar sobre pares (clave, valor)
stock = {"laptop": 5, "mouse": 20, "teclado": 12}

for articulo, cantidad in stock.items():
    print(f"- {articulo}: {cantidad} unidades")
````

### Comprensiones de diccionario

Al igual que las listas, los diccionarios soportan comprensiones: `{clave: valor for ... in ... if ...}`.

````python
nombres = ["ana", "luis", "sofia"]
capitalizar = {n: n.capitalize() for n in nombres}
print("Capitalizar:", capitalizar)

numeros = [1, 2, 3, 4, 5, 6]
cuadrados_pares = {n: n ** 2 for n in numeros if n % 2 == 0}
print("Cuadrados de pares:", cuadrados_pares)
````

### Diccionarios anidados

Un valor de un diccionario puede ser, a su vez, otro diccionario. Es una forma natural de modelar datos jerárquicos como usuarios, catálogos o registros JSON.

````python
clientes = {
    "C001": {"nombre": "Ana", "saldo": 150.5, "membresia": "premium"},
    "C002": {"nombre": "Luis", "saldo": 40.0, "membresia": "basico"},
}

print("Nombre de C001:", clientes["C001"]["nombre"])
print("Membresía de C002:", clientes["C002"]["membresia"])

for codigo, info in clientes.items():
    print(f"{codigo}: {info['nombre']} (saldo {info['saldo']:.2f})")
````

### OrderedDict

Desde Python 3.7 los diccionarios **preservan el orden de inserción** de forma nativa. `OrderedDict` (del módulo `collections`) sigue existiendo para compatibilidad y ofrece métodos como `move_to_end()`, útiles cuando necesitas reordenar explícitamente.

````python
from collections import OrderedDict

od = OrderedDict()
od["a"] = 1
od["b"] = 2
od["c"] = 3

print("Orden de inserción:", list(od))
od.move_to_end("a")   # mueve 'a' al final
print("Tras move_to_end('a'):", list(od))
````

## 5. Desempaquetado extendido

💡 **Intuición**: El desempaquetado extendido permite "desarmar" una colección de forma flexible usando `*` (para posicionales) y `**` (para por nombre), tanto al definir funciones como al pasar argumentos.

### `*args` en definición y llamada

`*args` recoge un número variable de argumentos posicionales en una tupla. Al llamar, el `*` expande una lista/tupla en argumentos posicionales.

````python
def total_acumulado(*numeros):
    return sum(numeros)

print("Con varios args:", total_acumulado(1, 2, 3))
print("Con muchos args:", total_acumulado(10, 20, 30, 40, 50))

# Desempaquetado en la llamada (* expande la colección)
valores = [1, 2, 3, 4]
print("Con *lista:", total_acumulado(*valores))
````

### `**kwargs` en definición y llamada

`**kwargs` recoge argumentos por nombre en un diccionario. Al llamar, `**` expande un diccionario en argumentos nombrados.

````python
def perfil(**datos):
    for clave, valor in datos.items():
        print(f"{clave} = {valor}")

perfil(nombre="Ana", edad=28, ciudad="Lima")

print("---")
info = {"nombre": "Luis", "edad": 30}
perfil(**info)
````

### Merge de diccionarios con `**`

En Python 3.5+ puedes fusionar diccionarios con el operador `**` dentro de `{}`. Desde Python 3.9 también existe el operador `|`.

````python
base = {"nombre": "Ana", "edad": 28}
extra = {"ciudad": "Lima", "activo": True}

fusionado = {**base, **extra}
print("Merge con **:", fusionado)

fusionado2 = base | extra   # Python 3.9+
print("Merge con |:", fusionado2)

# Los valores de la derecha ganan en caso de colisión
a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}
print("Colisión (gana el de la derecha):", {**a, **b})
````

## 6. Tipo hints para colecciones

💡 **Intuición**: Los **type hints** documentan qué tipo de datos espera una función o variable. Mejoran la legibilidad y permiten a las herramientas (mypy, IDE) detectar errores antes de ejecutar.

En Python 3.9+ puedes usar los tipos genéricos directamente sobre las colecciones (`list[int]` en lugar de `typing.List[int]`). Esto es más limpio y conciso.

````python
numeros: list[int] = [1, 2, 3]
punto: tuple[int, int] = (3, 4)
coordenadas: tuple[float, ...] = (1.5, 2.5, 3.5)  # tamaño variable
name2s: set[str] = {"ana", "luis"}
edades: dict[str, int] = {"Ana": 28, "Luis": 30}

print(numeros, punto, coordenadas, name2s, edades)
````

````python
def resumen_ventas(ventas: list[dict[str, float]]) -> dict[str, float]:
    """Suma las ventas por producto (list de dicts -> dict agregado)."""
    resultado: dict[str, float] = {}
    for v in ventas:
        for producto, monto in v.items():
            resultado[producto] = resultado.get(producto, 0.0) + monto
    return resultado


registros = [{"laptop": 1000, "mouse": 20}, {"laptop": 1500, "teclado": 50}]
print(resumen_ventas(registros))
````

### Ejercicio 1 (Guiado): Analizar ventas con lista y dict

Dado el siguiente inventario de ventas, calcula el **total por producto** y el **producto más vendido** usando un diccionario acumulador.

````python
ventas = ["laptop", "mouse", "laptop", "teclado", "mouse", "laptop", "teclado", "mouse"]

# TODO: contar cuántas veces aparece cada producto en un dict
conteo = {}
for producto in ventas:
    conteo[producto] = conteo.get(producto, 0) + 1

print("Conteo:", conteo)

mas_vendido = max(conteo, key=conteo.get)
print(f"Producto más vendido: {mas_vendido} ({conteo[mas_vendido]} ventas)")
````

### Ejercicio 2 (Guiado): Deduplicar emails con set

Una lista contiene emails con duplicados. **Elimina los duplicados preservando el orden** (pista: usa un set como registro de vistos y una lista para el resultado).

````python
emails = ["a@correo.com", "b@correo.com", "a@correo.com", "c@correo.com", "b@correo.com"]

vistos = set()
unicos_orden = []
for email in emails:
    if email not in vistos:
        vistos.add(email)
        unicos_orden.append(email)

print("Emails únicos (en orden):", unicos_orden)
````

### Ejercicio 3 (Guiado): Desempaquetado y merge de dicts

Crea una función `crear_usuario` que acepte un `id`, datos obligatorios y `**extras`, y devuelva un diccionario completo fusionando todo.

````python
def crear_usuario(uid: int, nombre: str, **extras: object) -> dict[str, object]:
    base = {"id": uid, "nombre": nombre}
    return {**base, **extras}


u = crear_usuario(1, "Ana", edad=28, ciudad="Lima", activo=True)
print(u)
````

### Ejercicio Independiente: Dashboard de puntuaciones

Usa lo aprendido para:
1. Crear un `dict` donde cada clave sea un estudiante y el valor una `list` de sus puntuaciones.
2. Calcular el **promedio** de cada estudiante y guardarlo en un nuevo dict.
3. Con un `set`, obtener el **conjunto de todas las puntuaciones** que aparecen al menos una vez (sin duplicados).
4. Imprimir el estudiante con el promedio más alto usando `max(dict, key=...)`.

Pista: combina comprensiones, `sum()`/`len()` y operaciones de set.

````python
puntuaciones = {
    "Ana": [90, 85, 100],
    "Luis": [75, 80, 82],
    "Sofía": [95, 92, 88],
}

# 1) Promedios
promedios = {estudiante: sum(notas) / len(notas) for estudiante, notas in puntuaciones.items()}
print("Promedios:", promedios)

# 2) Conjunto de todas las puntuaciones únicas
todas = set().union(*puntuaciones.values())
print("Puntuaciones únicas:", sorted(todas))

# 3) Mejor promedio
mejor = max(promedios, key=promedios.get)
print(f"Mejor promedio: {mejor} con {promedios[mejor]:.2f}")
````

## Resumen

✅ **Lo que aprendiste en este notebook:**

- **Listas**: secuencias ordenadas y mutables con `append`, `extend`, `pop`, `sort`, slicing y comprensiones. Acceso O(1), inserción/eliminación central O(n).
- **Tuplas**: secuencias ordenadas e inmutables. Ideales como claves de dict y para registros fijos. Desempaquetado y *named tuples*.
- **Sets**: colecciones sin duplicados ni orden, con búsqueda O(1) y operaciones `&`, `|`, `-`, `^`. `frozenset` para la versión inmutable.
- **Diccionarios**: mapeo clave-valor con acceso O(1), métodos `keys/values/items/update/pop`, comprensiones, anidamiento y `OrderedDict`.
- **Desempaquetado extendido**: `*args` y `**kwargs`, y merge de dicts con `**` o `|`.
- **Type hints**: `list[int]`, `dict[str, int]`, `tuple[int, ...]`, `set[str]` para código claro y seguro.

💡 **Regla de oro**: elige la estructura según **mutabilidad, orden, duplicados y tipo de acceso** necesarios. Para búsquedas frecuentes usa `set`/`dict`; para datos ordenados que cambian usa `list`; para registros fijos usa `tuple`.
