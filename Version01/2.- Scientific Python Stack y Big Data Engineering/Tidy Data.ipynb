# Módulo 2.7: Datos Tidy (Ordenados) - Principios y Práctica con Pandas

En el análisis de datos, la forma en que se organizan los datos tiene un impacto profundo en la facilidad con la que se pueden analizar, visualizar y modelar. El concepto de **Datos Tidy (Ordenados)**, popularizado por Hadley Wickham, propone un marco estandarizado para estructurar conjuntos de datos que simplifica enormemente el flujo de trabajo.

## 1. ¿Qué son los Datos Tidy?

Un conjunto de datos se considera 'tidy' si cumple con tres principios fundamentales:

1.  **Cada variable forma una columna.**
2.  **Cada observación forma una fila.**
3.  **Cada tipo de unidad observacional forma una tabla.**

Cuando los datos no están en formato 'tidy', decimos que están 'untidy' o desordenados. Transformar datos desordenados a 'tidy' es una de las tareas más comunes y cruciales en la limpieza y preprocesamiento de datos.

## 2. El Problema de los Datos Desordenados: Columnas que son Valores

Un patrón común de datos desordenados ocurre cuando los valores de una o más variables se almacenan como nombres de columnas. Considere un conjunto de datos donde las mediciones de 'frecuencia cardíaca' para diferentes 'fármacos' (`a` y `b`) se presentan como columnas separadas. Esto viola el principio de que cada variable (el tipo de fármaco) debe ser una columna y sus valores (la medición de la frecuencia cardíaca) deben estar en otra columna.

```python
import pandas as pd

# Datos 'Untidy': Las columnas 'a' y 'b' son en realidad valores de una variable 'fármaco'
pacientes = ['Ricardo', 'Marielena', 'Miguel']
frecuencia_a = [67, 80, 64]
frecuencia_b = [56, 90, 50]

df_untidy = pd.DataFrame({
    'pacientes': pacientes,
    'frecuencia_farmaco_A': frecuencia_a, # Aquí 'frecuencia_farmaco_A' es el valor 'A' de la variable 'fármaco'
    'frecuencia_farmaco_B': frecuencia_b  # Y 'frecuencia_farmaco_B' es el valor 'B'
})

print("DataFrame Original (Untidy):\n" )
print(df_untidy)
```

## 3. `pandas.melt()`: Transformando Datos de 'Ancho' a 'Largo' (Gathering)

La función `pd.melt()` es la herramienta principal en Pandas para transformar datos de un formato 'ancho' (wide) a un formato 'largo' (long), también conocido como 

`gathering` o `unpivoting`. Su objetivo es llevar los nombres de las columnas a una nueva columna de variables y los valores de esas columnas a otra nueva columna de valores.

```python
# Usamos pd.melt para transformar df_untidy a formato tidy
df_tidy = pd.melt(
    df_untidy,
    id_vars=['pacientes'],                   # Columnas que queremos mantener como identificadores
    value_vars=['frecuencia_farmaco_A', 'frecuencia_farmaco_B'], # Columnas que queremos 'fundir' en una variable
    var_name='farmaco',                      # Nuevo nombre para la columna de variables ('frecuencia_farmaco_A', 'frecuencia_farmaco_B')
    value_name='frecuencia_cardiaca'         # Nuevo nombre para la columna de valores (67, 80, 64, 56, 90, 50)
)

print("\nDataFrame Tidy:\n")
print(df_tidy)

# Podemos ver que ahora 'farmaco' es una variable (columna) con valores 'frecuencia_farmaco_A' y 'frecuencia_farmaco_B'
# Y 'frecuencia_cardiaca' contiene los valores correspondientes.
```

## 4. `df.pivot_table()` / `df.pivot()`: Transformando Datos de 'Largo' a 'Ancho' (Spreading)

La operación inversa a `melt` es `pivot` o `pivot_table`, que permite transformar datos de un formato 'largo' a un formato 'ancho' (spreading). Esto es útil cuando queremos que los valores únicos de una columna se conviertan en nuevas columnas. `pivot_table` es más flexible que `pivot` ya que permite funciones de agregación y puede manejar índices duplicados.

```python
# Ahora tomaremos el DataFrame tidy y lo transformaremos de nuevo a un formato wide
df_wide = df_tidy.pivot_table(
    index='pacientes',                         # La columna que se convertirá en el nuevo índice
    columns='farmaco',                         # La columna cuyos valores se convertirán en nuevas columnas
    values='frecuencia_cardiaca'               # La columna cuyos valores llenarán las nuevas celdas
).reset_index() # reset_index() para convertir el índice 'pacientes' de nuevo en una columna regular

# Renombrar las columnas para que sean como las originales, si se desea
df_wide.columns.name = None # Eliminar el nombre 'farmaco' del índice de columnas
df_wide = df_wide.rename(columns={
    'frecuencia_farmaco_A': 'frecuencia_farmaco_A_pivot',
    'frecuencia_farmaco_B': 'frecuencia_farmaco_B_pivot'
})

print("\nDataFrame Transformado de nuevo a 'Wide' (con pivot_table):\n")
print(df_wide)
```

## 5. Casos de Uso Avanzados y Buenas Prácticas

-   **Análisis Multivariado:** Los datos tidy son ideales para algoritmos de machine learning y librerías de visualización como Seaborn, que esperan una columna por variable.
-   **Consistencia:** Al tener los datos en un formato estándar, se reduce la ambigüedad y se facilita la automatización de procesos de análisis.
-   **Consideraciones de Rendimiento:** Para conjuntos de datos extremadamente grandes, las transformaciones pueden consumir mucha memoria. Es importante considerar la eficiencia de las operaciones de `melt` y `pivot`.

La capacidad de reestructurar datos de manera efectiva es una habilidad crucial en el conjunto de herramientas de cualquier científico de datos.
