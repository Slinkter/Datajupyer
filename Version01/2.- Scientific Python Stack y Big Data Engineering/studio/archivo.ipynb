```python
import numpy as np

# Definición del arreglo 3D
arr_3d = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])

# Salida estructurada y limpia
print("="*30)
print("📊 REPORTE DE ESTRUCTURA NUMPY")
print("="*30)

# Información de dimensiones
print(f"🔹 Dimensiones (Shape): {arr_3d.shape}")
print(f"🔹 Rango (NDim):        {arr_3d.ndim}")
print(f"🔹 Total Elementos:     {arr_3d.size}")
print("-" * 30)

# Visualización del contenido
print("📦 CONTENIDO DEL ARREGLO:")
print(arr_3d)
print("="*30)
```

```python
a = np.array([1,2,3])
b = np.array([10,20,30])

print(f"a + b = {a+b}")
print(f"a * b = {a*b}")
print(f"a ** 2 = {a**2}")

print(f"sum(a)  = {np.sum(a)} -----  {np.sum(b)}")
print(f"mean(a) = {np.mean(a)} ----- {np.mean(b)}")
print(f"std(a)  = {np.std(a)} ----- {np.std(b)}")
```
