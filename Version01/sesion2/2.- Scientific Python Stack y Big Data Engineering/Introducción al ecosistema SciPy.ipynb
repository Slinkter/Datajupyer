## 🌉 Puente Pedagógico: La Pirámide Científica de Python

### ¿Por qué existe un ecosistema y no una sola mega-librería?
Python por sí solo es un lenguaje interpretado dinámico, lo cual es amigable para el desarrollador pero lento para cálculos matemáticos masivos. El stack científico se construyó sobre librerías en C/Fortran vectorizadas, organizadas en capas modulares según su nivel de abstracción.

````python
import numpy as np
from scipy import stats, optimize
import matplotlib.pyplot as plt

# Ejemplo 1: Prueba de hipótesis estadística con scipy.stats
# Comparación de dos grupos (Test t de Student)
np.random.seed(42)
grupo_control = np.random.normal(loc=50, scale=10, size=100)
grupo_tratamiento = np.random.normal(loc=54, scale=10, size=100)

t_stat, p_valor = stats.ttest_ind(grupo_control, grupo_tratamiento)
print(f"Estadístico t: {t_stat:.4f}, p-valor: {p_valor:.4e}")
if p_valor < 0.05:
    print("Resultado estadísticamente significativo (rechazamos hipótesis nula)")
else:
    print("No hay evidencia suficiente para rechazar la hipótesis nula")
````

## 1. Optimización con `scipy.optimize`

SciPy permite encontrar mínimos de funciones matemáticas complejas mediante algoritmos avanzados (BFGS, Nelder-Mead).

````python
# Función de costo cuadrática: f(x) = (x - 3)^2 + 5
def costo(x):
    return (x - 3.0) ** 2 + 5.0

resultado_opt = optimize.minimize(costo, x0=[0.0])
print(f"Mínimo encontrado en x = {resultado_opt.x[0]:.4f} con valor f(x) = {resultado_opt.fun:.4f}")
````

## 2. Mapa y Propósito de las Librerías Científicas

| Librería | Propósito Primario | Cuándo Usarla |
|:---|:---|:---|
| **NumPy** | Arrays N-dimensionales y álgebra lineal básica | Vectores, matrices, transformaciones matemáticas rápidas |
| **SciPy** | Algoritmos científicos avanzados | Optimización, pruebas estadísticas, integración, procesamiento de señales |
| **Pandas** | Tablas estructuradas con etiquetas (DataFrames) | Carga de CSV/SQL, limpieza de nulos, agrupaciones, series de tiempo |
| **Matplotlib / Seaborn** | Visualización 2D/3D estática e interactiva | Gráficos exploratorios, histogramas, heatmaps, dispersión |
| **Scikit-Learn** | Machine Learning predictivo clásico | Clasificación, regresión, clustering, reducción de dimensionalidad |
| **Statsmodels** | Modelado estadístico riguroso | Regresiones OLS con tablas de significancia p, pruebas econométricas |

### Ejercicio 1 (Guiado): Intervalo de Confianza con `scipy.stats`

````python
datos = np.array([22, 24, 25, 23, 26, 29, 21, 24, 27, 25])
media = np.mean(datos)
intervalo_95 = stats.t.interval(0.95, df=len(datos)-1, loc=media, scale=stats.sem(datos))
print(f"Media: {media:.2f}")
print(f"Intervalo de confianza del 95%: [{intervalo_95[0]:.2f}, {intervalo_95[1]:.2f}]")
````
