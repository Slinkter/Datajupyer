## 📋 Cheat Sheet: Métodos Clave de la Fase 2

| Módulo | Función / Método | Propósito Principal |
|:---|:---|:---|
| **Ingesta** | `pd.read_csv()`, `pd.read_parquet()` | Lectura eficiente de tablas |
| **Scraping** | `soup.select()`, `soup.find_all()` | Búsqueda por selectores CSS y etiquetas en el DOM |
| **Regex** | `re.findall()`, `df[col].str.extract()` | Extracción estructurada de substrings por patrones |
| **SQL** | `pd.read_sql_query(query, conn)` | Ejecución directa de consultas relacionales a DataFrame |
| **Tidy Data** | `pd.melt()`, `df.pivot()` | Transformación bidireccional entre formato ancho y largo |
| **Estadística** | `scipy.stats.ttest_ind()` | Contraste de hipótesis paramétrico |
| **Optimización** | `scipy.optimize.minimize()` | Minimización de funciones objetivo continuas |

````python
import pandas as pd
import numpy as np

# Mini proyecto integrador: Pipeline relacional a Tidy format
datos_crudos = pd.DataFrame({
    "id": [1, 2, 3],
    "info_contacto": ["Juan (ventas@tienda.com) - CDMX", "Ana (soporte@corp.org) - MTY", "Luis (luis@ia.com) - GDL"],
    "ventas_2025": [15000, 24000, 18500],
    "ventas_2026": [19000, 31000, 22000]
})

# 1. Regex para extraer nombre y correo
datos_crudos[["nombre", "email"]] = datos_crudos["info_contacto"].str.extract(r"^([A-Za-z]+)\s*\(([^)]+)\)")

# 2. Melt para formato Tidy por año
df_resumen_tidy = pd.melt(
    datos_crudos,
    id_vars=["id", "nombre", "email"],
    value_vars=["ventas_2025", "ventas_2026"],
    var_name="periodo",
    value_name="monto"
)
df_resumen_tidy["periodo"] = df_resumen_tidy["periodo"].str.replace("ventas_", "")

print("--- Resultado del Pipeline Integrador ---")
display(df_resumen_tidy)
````
