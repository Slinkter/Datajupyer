```python
import matplotlib.pyplot as plt
plt.title("Grafica de puntos")
plt.xlabel("puntos x")
plt.ylabel("puntos y")
plt.xlim(-20,20)
plt.ylim(0,20)
plt.grid(True)
plt.plot([19,15,3,8,1,8,3,15,19],color="red",marker="o",linestyle="--")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.plot([1,5,3], marker="o", linestyle="--", color="gray")
plt.title("Ejemplo simple de gráfica")
plt.xlabel("Índice")
plt.ylabel("Valor")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Datos de ejemplo
plt.plot([1,5,3], marker="o", linestyle="--", color="red")

# Ajustar límites de los ejes
plt.xlim(0, 5)   # eje X de 0 a 10
plt.ylim(0, 10)   # eje Y de 0 a 10

# Añadir detalles
plt.title("Gráfica con ejes limitados")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)

plt.show()
```
