#### ¿Qué es Conda?

- Conda es un gestor de paquetes (¡Sí, es como pip!) que sirve para instalar, correr y actualizar paquetes y sus dependencias.

- Conda también es un gestor de ambiente (¡Sí, como virtualenv!). En un entorno de Conda es un directorio que contiene una colección específica de paquetes que se han instalado. Si se cambia entre un ambiente y otro los demás entornos no se ven afectados y se pueden activar o desactivar fácilmente.

#### Array slicing

- a[comienzo:fin] # elementos desde el índice del comienzo hasta el índice fin-1
- a[comienzo:] # del número en comienzo hasta el fin
- a[:fin] # desde el principio hasta fin-1
- a[:] # todo el arreglo
- a[comienzo:fin:paso] # elementos desde el índice del comienzo hasta el índice fin-1, por paso

Usando el dataframe anterior, crear un gráfico de barra donde se vea la cantidad de sobrevivientes en proporción al total de pasajeros por clase.

### RESPUESTA

```python
tabla = pd.crosstab(index=data["Pclass"], columns=data["Survived"], margins=True) #Agregando margins=True nos da los totales
tabla
```

```python
#Obtenemos la proporción dividiento entre el total por filas
tabla = tabla.div(tabla["All"], axis=0)
tabla
```
