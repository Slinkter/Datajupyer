## Ejercicio Práctico: Dataset de Riesgo Crediticio

````python
import pandas as pd
import numpy as np
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
````

### Crear Dataset de Ejemplo

````python
np.random.seed(42)
n_samples = 1000

data = {
    'edad': np.random.randint(18, 70, n_samples),
    'ingreso_anual': np.random.normal(50000, 15000, n_samples),
    'anos_empleo': np.random.randint(0, 30, n_samples),
    'deuda_actual': np.random.exponential(5000, n_samples),
    'score_credito': np.random.normal(650, 100, n_samples),
    'num_tarjetas': np.random.randint(0, 10, n_samples),
    'alquila': np.random.randint(0, 2, n_samples),
    'estados': np.random.randint(0, 5, n_samples)
}

df = pd.DataFrame(data)
df['incumplido'] = ((df['deuda_actual'] / df['ingreso_anual'] > 0.3) & 
                   (df['score_credito'] < 600)).astype(int)

df.head()
````

### Aplicar Feature Selection

````python
X = df.drop('incumplido', axis=1)
y = df['incumplido']

selector = SelectKBest(score_func=f_classif, k=5)
X_selected = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]
print(f"Features seleccionados: {list(selected_features)}")
````

### Entrenar Modelo con Features Seleccionados

````python
X_train, X_test, y_train, y_test = train_test_split(
    X_selected, y, test_size=0.2, random_state=42
)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
````
