## El Puente Pedagógico: De los datos al conocimiento

```
    DATOS ──────► PREPROCESAMIENTO ──────► TRANSFORMACIÓN ──────► MINING ──────► EVALUACIÓN
      │              │                    │                  │              │
      ▼              ▼                    ▼                  ▼              ▼
   raw data      cleaning           conversion        patterns       knowledge
   (crudo)       (limpieza)        (transform)     (patrones)      (saber)
```

## ¿Qué es KDD?

**Knowledge Discovery in Databases (KDD)** es el proceso de extraer conocimiento útil de grandes volúmenes de datos.

No es solo "usar Machine Learning". Es un **proceso integral** que incluye:
- Selección de datos relevantes
- Limpieza y preprocesamiento
- Transformación de datos
- Mining (búsqueda de patrones)
- Evaluación e interpretación

## El Proceso KDD

```
┌─────────────────────────────────────────────────────────────────┐
│                      PROCESO KDD                               │
└─────────────────────────────────────────────────────────────────┘
                                                                 
  1. SELECCIÓN      ←── ¿Qué datos necesito?                    
         │                                                     
  2. PREPROCESO    ←── Limpiar valores faltantes, outliers       
         │                                                     
  3. TRANSFORM    ←── Normalizar,Encoding,Agregar           
         │                                                     
  4. MINING       ←── Aplicar algoritmos (cluster, rules)      
         │                                                     
  5. EVALUACIÓN   ←── Validar patrones encontrados           
         │                                                     
  6. KNOWLEDGE    ←── Output: insights accionables           
```

````python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs
````

## Paso 1: Selección de Datos

````python
X, y = make_blobs(n_samples=500, centers=3, n_features=2, 
                 random_state=42, cluster_std=1.5)

df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df['target'] = y

print(f"Dataset: {df.shape[0]} muestras, {df.shape[1]} columnas")
df.head()
````

## Paso 2: Preprocesamiento

````python
df_clean = df.copy()

df_clean['feature_1'] = df_clean['feature_1'].fillna(df_clean['feature_1'].mean())
df_clean['feature_2'] = df_clean['feature_2'].fillna(df_clean['feature_2'].mean())

print("Datos preprocesados: valores faltantes imputed")
````

## Paso 3: Transformación

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean[['feature_1', 'feature_2']])

print(f"Datos transformados: media=0, desviación=1")
print(f"Media: {X_scaled.mean(axis=0)}")
print(f"Std: {X_scaled.std(axis=0)}")
````

## Paso 4: Mining - Clustering

````python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

df_clean['cluster'] = clusters
print(f"Clustering completado: {len(np.unique(clusters))} clusters")
````

## Paso 5: Evaluación

````python
from sklearn.metrics import adjusted_rand_score

ari = adjusted_rand_score(y, clusters)
print(f"Adjusted Rand Index: {ari:.3f}")
print("(1.0 = perfecta alineación, 0.0 = aleatorio)")
````

## Resumen: Proceso KDD

| Etapa | Objetivo | Técnicas |
|------|---------|----------|
| Selección | Elegir datos relevantes | SQL, filtros |
| Preprocesamiento | Limpiar datos | Imputación, outlier removal |
| Transformación | Cambiar formato | Normalización, encoding |
| Mining | Encontrar patrones | Clustering, rules |
| Evaluación | Validar resultados | Métricas, visualización |
