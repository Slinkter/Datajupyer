````python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
````

## Ejemplo 1: Clustering con K-Means

### El Algoritmo K-Means

```
1. Elegir K centroides iniciales (aleatorios)
2. Asignar cada punto al centroide más cercano
3. Recalcular centroides como promedio de puntos asignados
4. Repetir hasta que no cambien las asignaciones
```

````python
np.random.seed(42)

X, _ = make_blobs(n_samples=300, centers=4, n_features=2,
                 random_state=42, cluster_std=1.2)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

print(f"Clustering completado!")
print(f"Número de clusters: {len(np.unique(labels))}")
print(f"Puntos por cluster: {np.bincount(labels)}")
````

## Ejemplo 2: DBSCAN para detección de outliers

### ¿Por qué DBSCAN?

DBSCAN (Density-Based Spatial Clustering) es diferente:
- **No requiere especificar K** - encuentra automáticamente
- **Encuentra outliers** - puntos en regiones de baja densidad
- **Forma arbitraria** - no asume clusters esféricos

````python
X_with_outliers = np.vstack([X, np.array([[15, 15], [16, 16]])])

dbscan = DBSCAN(eps=1.5, min_samples=5)
db_labels = dbscan.fit_predict(X_with_outliers)

outliers = db_labels == -1
print(f"Outliers detectados: {outliers.sum()}")
print(f"Clusters encontrados: {len(set(db_labels)) - (1 if -1 in db_labels else 0)}")
````

## Ejemplo 3: Reglas de Asociación

### Mercado Basket Analysis

Las **reglas de asociación** encuentran relaciones del tipo:
"Si el cliente compra **pan**, también compra **mantequilla**"

Esto se llama **market basket analysis**.

````python
transacciones = [
    ['pan', 'leche', 'huevos'],
    ['pan', 'mantequilla'],
    ['leche', 'huevos'],
    ['pan', 'leche', 'mantequilla'],
    ['huevos', 'mantequilla', 'jugo'],
    ['pan', 'jugo'],
    ['pan', 'leche', 'huevos', 'mantequilla'],
    ['leche', 'mantequilla'],
]

items = list(set([item for t in transacciones for item in t]))
print(f"Items únicos: {items}")
````

````python
def support(itemset, transacciones):
    count = sum(1 for t in transacciones if all(i in t for i in itemset))
    return count / len(transacciones)

def confidence(antecedent, consequent, transacciones):
    ant_count = sum(1 for t in transacciones if all(i in t for i in antecedent))
    if ant_count == 0:
        return 0
    both_count = sum(1 for t in transacciones if all(i in t for i in antecedent + consequent))
    return both_count / ant_count

print(f"Support(pan): {support(['pan'], transacciones):.1%}")
print(f"Confidence(pan → mantequilla): {confidence(['pan'], ['mantequilla'], transacciones):.1%}")
````
