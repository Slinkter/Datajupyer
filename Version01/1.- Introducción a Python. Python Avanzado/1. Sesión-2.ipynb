# Módulo 1.2: Funciones y Programación Orientada a Objetos en Python

Esta sesión profundiza en dos de los pilares más importantes de Python: las **funciones** como bloques de código reutilizables y el **modelo de programación orientada a objetos (POO)**, que nos permite estructurar el código de manera modular y escalable.

## 1. Funciones: Bloques de Lógica Reutilizables

Las funciones son la principal herramienta para dividir el código en bloques de lógica manejables y reutilizables. Permiten organizar el código, hacerlo más legible y evitar la repetición. Una función bien diseñada toma unas entradas (argumentos), realiza una tarea específica y, opcionalmente, devuelve un resultado.

### 1.1. Definición Básica

La sintaxis básica para definir una función utiliza la palabra clave `def`, seguida del nombre de la función, paréntesis para los argumentos y dos puntos. El cuerpo de la función debe estar indentado. Para la interpolación de strings, se recomienda usar **f-strings** (disponibles desde Python 3.6), ya que son más legibles y eficientes que el antiguo formato con `%`.

```python
def saludar(nombre_usuario, saludo="Buenos días"):
    """Saluda a un usuario con un saludo personalizable."""
    mensaje = f"Hola, {nombre_usuario}. ¡Te deseo {saludo}!"
    print(mensaje)

# Llamamos a la función con ambos argumentos
saludar("Ricardo", "una excelente semana")

# Llamamos a la función usando el valor por defecto para 'saludo'
saludar("Ana")
```

### 1.2. Argumentos Avanzados: `*args` y `**kwargs`

Python ofrece una sintaxis muy flexible para manejar un número variable de argumentos:

- **`*args`**: Permite pasar un número variable de **argumentos posicionales**, que se empaquetan en una **tupla** dentro de la función.
- **`**kwargs`**: Permite pasar un número variable de **argumentos de palabra clave (keyword)**, que se empaquetan en un **diccionario** dentro de la función.

```python
def procesar_datos(*args, **kwargs):
    """Una función genérica que recibe cualquier tipo de argumento."""
    print(f"Argumentos posicionales (tupla): {args}")
    print(f"Argumentos de palabra clave (diccionario): {kwargs}")

print("Llamada 1:")
procesar_datos(101, 'cliente_A', 200.50, status='activo', region='Norte')

print("\nLlamada 2 (desempaquetando una lista y un diccionario):")
items = [102, 'cliente_B', 350.0]
detalles = {'status': 'pendiente', 'prioridad': 'alta'}
procesar_datos(*items, **detalles)
```

### 1.3. Decoradores: Modificando Funciones

Un decorador es una función que toma otra función como argumento, le añade alguna funcionalidad y devuelve otra función, sin alterar el código de la función original. Es una forma poderosa de aplicar la metaprogramación. Se usan con la sintaxis `@decorador`.

```python
import time

def temporizador(func):
    """Un decorador que mide y muestra el tiempo de ejecución de una función."""
    def wrapper(*args, **kwargs):
        inicio = time.time()
        resultado = func(*args, **kwargs)
        fin = time.time()
        print(f"La función '{func.__name__}' tardó {fin - inicio:.4f} segundos en ejecutarse.")
        return resultado
    return wrapper

@temporizador
def tarea_larga(n):
    """Simula una tarea que consume tiempo."""
    sum(i*i for i in range(n))

tarea_larga(1000000)
```

## 2. Programación Orientada a Objetos (POO)

La POO es un paradigma de programación que utiliza **objetos** para representar entidades del mundo real (o abstracto). Cada objeto es una **instancia** de una **clase**.

- **Clase:** Una plantilla o molde para crear objetos. Define un conjunto de atributos (datos) y métodos (funciones) que los objetos de esa clase tendrán.
- **Objeto:** Una instancia de una clase. Contiene datos concretos en sus atributos y puede ejecutar los métodos definidos en su clase.

### 2.1. Definición de una Clase

En Python 3, una clase se define con la palabra clave `class`. No es necesario heredar de `object` explícitamente, ya que es el comportamiento por defecto.

```python
class Vehiculo:
    # Atributo de clase: es compartido por todas las instancias de la clase
    garantia_meses = 12

    def __init__(self, marca, modelo, anio):
        """El constructor de la clase. Se llama al crear una nueva instancia."""
        # Atributos de instancia: son únicos para cada objeto
        self.marca = marca
        self.modelo = modelo
        self.anio = anio
        self._kilometraje = 0 # Un atributo 'privado' por convención

    def describir(self):
        """Un método de instancia que opera sobre los datos del objeto."""
        return f"{self.marca} {self.modelo} ({self.anio}) con {self._kilometraje} km."

    def avanzar(self, km):
        """Modifica el estado interno del objeto."""
        if km > 0:
            self._kilometraje += km

# Creamos dos objetos (instancias) de la clase Vehiculo
coche1 = Vehiculo("Toyota", "Corolla", 2022)
coche2 = Vehiculo("Ford", "Mustang", 2023)

coche1.avanzar(150)
print(coche1.describir())
print(coche2.describir())

# El atributo de clase es el mismo para ambos
print(f"Garantía Coche 1: {coche1.garantia_meses} meses")
print(f"Garantía Coche 2: {coche2.garantia_meses} meses")
```

### 2.2. Propiedades (`@property`)

El decorador `@property` permite tratar un método de una clase como si fuera un atributo de solo lectura. Esto es útil para calcular valores derivados o para proteger el acceso directo a un atributo.

```python
import datetime

class Vehiculo:
    def __init__(self, marca, modelo, anio):
        self.marca = marca
        self.modelo = modelo
        self.anio = anio

    @property
    def antiguedad(self):
        """Calcula la antigüedad del vehículo como un atributo de solo lectura."""
        anio_actual = datetime.date.today().year
        return anio_actual - self.anio

coche = Vehiculo("Honda", "Civic", 2018)

# Accedemos a 'antiguedad' como si fuera un atributo, no un método
print(f"El coche es un {coche.marca} {coche.modelo}.")
print(f"Tiene {coche.antiguedad} años de antigüedad.")
```
