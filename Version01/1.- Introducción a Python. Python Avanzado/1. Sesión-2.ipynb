## 1. ¿Por qué usar funciones?

### El problema sin funciones:
```python
# ❌ Código repetitivo y difícil de mantener
print("Hola Juan, bienvenido!")
print("Hola María, bienvenido!")
print("Hola Pedro, bienvenido!")
```

### La solución con funciones:
```python
# ✅ Código reutilizable
def saludar(nombre):
    print(f"Hola {nombre}, bienvenido!")

saludar("Juan")
saludar("María")
saludar("Pedro")
```

> **Analogía**: Una función es como una receta de cocina. La escribes UNA vez y la puedes usar MIL veces.

## 2. Anatomía de una Función

```
     ┌──────────┐←────────── Nombre de la función
     │ def     │←────────── Palabras reservadas
     │ saludar │
     │ (nombre) │←────────── Parámetro (entrada)
     │    :    │
     │ print() │←────────── Cuerpo (acción)
     │ return │←────────── Retorno (salida)
     └──────────┘
```

````python
# Estructura básica de una función
def saludar(nombre):
    """Documentación: Saluda a un usuario por nombre"""
    mensaje = f"Hola {nombre}, bienvenido!"
    return mensaje  # Retorna el valor

# Llamar la función
resultado = saludar("Juan")
print(resultado)
````

## 3. Tipos de Parámetros

### 3.1 Parámetros Posicionales
```python
def sumar(a, b, c):  # a, b, c son posicionales
    return a + b + c

sumar(1, 2, 3)  # 1→a, 2→b, 3→c
```

````python
# Parámetros posicionales
def describir_persona(nombre, edad, ciudad):
    return f"{nombre} tiene {edad} años y vive en {ciudad}"

# Se asignan en orden
describir_persona("Ana", 28, "CDMX")
````

### 3.2 Parámetros con Valor por Defecto

```python
def saludar(nombre, mensaje="Hola"):
    # mensaje tiene valor por defecto
    return f"{mensaje}, {nombre}!"

saludar("Ana")        # usa "Hola"
saludar("Ana", "Hi")  # usa "Hi"
```

````python
# Parámetros con valores por defecto
def crear_usuario(nombre, rol="usuario", activo=True):
    return {
        "nombre": nombre,
        "rol": rol,
        "activo": activo
    }

# Usar valores por defecto
crear_usuario("Carlos")

# Sobrescribir algunos
crear_usuario("Carlos", rol="admin")
````

### 3.3 *args (Argumentos Variables)

Permite recibir cualquier cantidad de argumentos:

```python
def sumar_todos(*numeros):
    # numeros es una TUPLA con todos los argumentos
    total = 0
    for n in numeros:
        total += n
    return total

sumar_todos(1, 2)           # → 3
sumar_todos(1, 2, 3, 4, 5)  # → 15
```

````python
# *args - argumentos variables
def promediar(*notas):
    """Calcula el promedio de cualquier cantidad de notas"""
    if not notas:
        return 0
    return sum(notas) / len(notas)

promediar(8, 9, 7)  # 3 args
promediar(8, 9, 7, 10, 9, 8)  # 6 args
````

### 3.4 **kwargs (Argumentos con Nombre)

Permite recibir argumentos con nombre (diccionario):

```python
def configurar(**config):
    # config es un DICCIONARIO
    for clave, valor in config.items():
        print(f"{clave}: {valor}")

configurar(color="azul", tamaño="L")
```

````python
# **kwargs - argumentos con nombre
def crear_perfil(**datos):
    """Recibe cualquier cantidad de argumentos con nombre"""
    return datos

crear_perfil(nombre="Ana", edad=28, ciudad="CDMX")
````

### 📊 Resumen de Parámetros

| Orden | Tipo | Sintaxis | Resultado |
|:----:|------|---------|----------|
 | 1 | Posicional | `def f(a, b)` | Tupla |
 | 2 | Defecto | `def f(a=1)` | Valor default |
 | 3 | *args | `def f(*args)` | Tupla |
 | 4 | **kwargs | `def f(**kwargs)` | Diccionario |

## 4. Scope (Alcance de Variables)

¿De dónde puede ver una variable una función?

```
┌─────────────────────────────────┐
│  ÁMBITO GLOBAL (módulo)         │
│  ─────────────────────────────  │
│  variable_global = 10            │
│  ┌─────────────────────┐        │
│  │ FUNCIÓN           │        │
│  │ ────────────────   │        │
│  │ global 可访问    │ No    │
│  │ variable_global │ ✅ SÍ  │
│  │                │        │
│  │ local = 5        │ ✅ SÍ  │
│  │ # Solo existe     │        │
│  │ # dentro de función          │
│  └─────────────────────┘        │
��─────────────────────────────────┘
```

**Regla de oro**: Las funciones pueden LEER variables globales, pero no pueden MODIFICARLAS directamente.

````python
# Ejemplo de Scope
contador_global = 0  # Variable global

def incrementar():
    """La función puede LEER la global"""
    global contador_global  # Necesitas declararla
    contador_global += 1
    return contador_global

incrementar()
incrementar()
contador_global
````

> ** mejor práctica**: Evita usar variables globales. Pasa los valores como parámetros.

## 5. Funciones Anidadas (Clousures)

Una función que retorna otra función:

```python
def crear_multiplicador(factor):
    """Crea una función que multiplica""""
    def multiplicar(x):
        return x * factor
    return multiplicar  # Retorna la función

duplicar = crear_multiplicador(2)
triplicar = crear_multiplicador(3)

duplicar(5)   # → 10
triplicar(5)  # → 15
```

````python
# Función anidada (Closure)
def crear_saludo(prefix):
    """Crea una función de saludo personalizada"""
    def saludar(nombre):
        return f"{prefix}, {nombre}!"
    return saludar

hola = crear_saludo("Hola")
adios = crear_saludo("Adiós")

hola("Ana")   # → "Hola, Ana!"
adios("Ana")  # → "Adiós, Ana!"
````

## 6. Decoradores

Un decorador es una función que MODIFICA otra función:

```
┌───────────────────────────────────┐
│  @mi_decorador   (antes de función)  │
│  def mi_funcion():                  │
│      pass                          │
└───────────────────────────────────┘
              ↓
┌───────────────────────────────────┐
│  mi_funcion = mi_decorador(mi_funcion)│
└───────────────────────────────────┘
```

````python
# Decorador simple
def cronometrar(func):
    """Mide el tiempo de ejecución"""
    def wrapper(*args, **kwargs):
        import time
        inicio = time.time()
        resultado = func(*args, **kwargs)
        print(f"Tiempo: {time.time() - inicio:.4f}s")
        return resultado
    return wrapper

@cronometrar
def proceso_lento():
    import time
    time.sleep(0.1)
    return "Completado!"

proceso_lento()
````

## 7. Introducción a Clases (POO)

### ¿Qué es una clase?

Una **clase** es un plano/plantilla para crear **objetos**.

```
┌─────────────────────────────────────────────┐
│  CLASE: Persona                      │
│  ────────────────────────────────     │
│  ATRIBUTOS (datos):                │
│    - nombre                       │
│    - edad                         │
│    - correo                       │
│  MÉTODOS (funciones):              │
│    - saludar()                    │
│    - cumplir_años()               │
└─────────────────────────────────────────────┘
              ↓
┌─────────────────────────────────────────────┐
│  OBJETO (instancia)                 │
│  ────────────────────────────────     │
│  ana = Persona("Ana", 28)        │
│  juan = Persona("Juan", 32)      │
└─────────────────────────────────────────────┘
```

````python
# Definir una clase
class Persona:
    """Clase que representa una persona"""
    
    # Atributo de clase (común a todas las instancias)
    especie = "Humano"
    
    # Constructor - se ejecuta al crear un objeto
    def __init__(self, nombre, edad):
        self.nombre = nombre    # Atributo de instancia
        self.edad = edad      # Atributo de instancia
    
    # Método - función dentro de clase
    def saludar(self):
        return f"Hola, soy {self.nombre}"
    
    def cumplenos(self):
        self.edad += 1
    
    # Representación legible del objeto
    def __str__(self):
        return f"Persona({self.nombre}, {self.edad} años)"
````

````python
# Crear objetos (instancias de la clase)
ana = Persona("Ana", 28)
juan = Persona("Juan", 32)

# Usar métodos y atributos
ana.saludar()
ana.especie  # Acceder atributo de clase

print(ana)
ana.cumplenos()
print(ana)
````

### Métodos Especiales (Dunder Methods)

Python tiene métodos con doble guión bajo (`__`):

| Método | Para qué sirve |
|-------|-------------|
| `__init__` | Constructor (crear objeto) |
| `__str__` | Representación legible (str()) |
| `__repr__` | Representación técnica (debug) |
| `__eq__` | Comparar con == |
| `__len__` | Obtener longitud (len()) |
| `__add__` | Sumar con + |

````python
# Métodos especiales
class Carrito:
    def __init__(self):
        self.items = []
    
    def __len__(self):
        return len(self.items)
    
    def __add__(self, item):
        self.items.append(item)
        return self

carrito = Carrito()
carrito + "Manzana"
carrito + "Pan"

len(carrito)  # → 2
````

### Ejercicio 1: Función Calculadora
Crea una función calculadora simple:
```python
def calculadora(a, b, operacion):
    # operacion puede ser: "suma", "resta", "multiplica", "divide"
    pass
```

````python
# Solución:
def calculadora(a, b, operacion):
    operaciones = {
        "suma": lambda a, b: a + b,
        "resta": lambda a, b: a - b,
        "multiplica": lambda a, b: a * b,
        "divide": lambda a, b: a / b if b != 0 else "Error"
    }
    return operaciones.get(operacion, lambda *_: "Invalid")(a, b)

calculadora(10, 5, "suma")
````

### Ejercicio 2: Clase Biblioteca
Crea una clase que gestione libros:

````python
# Solución:
class Biblioteca:
    def __init__(self):
        self.libros = []
    
    def agregar(self, titulo, autor):
        self.libros.append({"titulo": titulo, "autor": autor})
    
    def buscar(self, autor):
        return [l for l in self.libros if l["autor"] == autor]
    
    def __len__(self):
        return len(self.libros)

biblioteca = Biblioteca()
biblioteca.agregar("Cien Años", "García Márquez")
biblioteca.agregar("1984", "Orwell")
len(biblioteca)
````

### Clases
- Los objetos son una encapsulación de variables y funciones en una sola entidad.
- Los objetos toman sus variables y funciones de las clases. 
- Las clases son esencialmente una plantilla para crear sus objetos.

````python
class MiClase(object):
    """
    Mi clase doc
    """
    class_prop = "Propiedad de clase"

    def __init__(self, nombre=None, verbose=False) :
        """
        Documentacion de constructor
        """
        self.data = []
        self._private_data = []
        self._nombre = nombre
    
    def __del__(self):
        pass
    
    def __add__(self, a):
        pass
    
    def __getitem__(self, key):
        pass
    
    def function(self):
        """
        Documentacion de methodo
        """
        self._privada()
        print("Este mensaje se imprime desde la clase")
    
    def _privada(self):
        """
        Documentacion de methodo
        """
        print("Este mensaje se imprime desde la clase")
        
    @staticmethod
    def metodo_estatico():
        print("Statico, no tiene acceso a 'self'")
    
    @classmethod
    def verboso(cls):
        print("Metodo de clase")
        a = cls(verbose=False)
        del a.data
    
    @property
    def nombre(self):
        print("Calling nombre")
        return self._nombre
    
    @nombre.setter
    def nombre(self, a):
        self._nombre = a
        print("Calling setter")
    
    def algo(self):
        """
        doc de lo que hago
        """
        raise NotImplementedError()
````

````python
mi_objeto = MiClase()
mi_objeto.data = 12
mi_objeto.cls_list.append(1)
mi_objeto.function()
mi_objeto = MiClase()
mi_objeto.cls_list.append(1)
````
