```python
import time 

def temporazidor(func):
    def wrapper(*args,**kwargs):
        inicio = time.time()
        resultado= func(*args,**kwargs)
        fin = time.time()
        print(f"la funcion {func.__name__} tardo {fin - inicio:.4f} segundos en ejecutarse")
        return resultado
    return wrapper

@temporazidor
def tareas(n):
    sum(i*i for i in range(n))

tareas(100000000)
    

```

```python
class Vehiculo:
    # Atributo de clase: es compartido por todas las instancias de la clase
    garantia_meses = 12

    def __init__(self, marca, modelo, anio):
        """El constructor de la clase. Se llama al crear una nueva instancia."""
        # Atributos de instancia: son únicos para cada objeto
        self.marca = marca
        self.modelo = modelo
        self.anio = anio
        self._kilometraje = 0 # Un atributo 'privado' por convención

    def describir(self):
        """Un método de instancia que opera sobre los datos del objeto."""
        return f"{self.marca} {self.modelo} ({self.anio}) con {self._kilometraje} km."

    def avanzar(self, km):
        """Modifica el estado interno del objeto."""
        if km > 0:
            self._kilometraje += km

# Creamos dos objetos (instancias) de la clase Vehiculo
coche1 = Vehiculo("Toyota", "Corolla", 2022)
coche2 = Vehiculo("Ford", "Mustang", 2023)

coche1.avanzar(120)
print(coche1.describir())
print(coche2.describir())


# El atributo de clase es el mismo para ambos
print(f"Garantía Coche 1: {coche1.garantia_meses} meses")
print(f"Garantía Coche 2: {coche2.garantia_meses} meses")
```

```python
import datetime
```

```python
class VehiculoProperty:
    def __init__(self,marca,modelo,anio):
        self.marcar = marca
        self.modelo= modelo
        self.anio=anio

    @property
    def antiguedad(seft):
        current_year = datetime.date.today().year
        rpta = current_year - seft.anio
        return rpta
    
coche3 = VehiculoProperty("Honda","civic",2018)
print(coche3.antiguedad)

```
