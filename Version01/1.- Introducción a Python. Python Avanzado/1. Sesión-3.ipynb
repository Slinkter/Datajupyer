## 1. Manejo de Excepciones

Python usa excepciones para gestionar errores que ocurren durante la ejecución. En lugar de que el programa se detenga abruptamente, podemos 'capturar' estos errores y manejarlos de forma controlada.

### Filosofías: LBYL vs. EAFP

- **LBYL (Look Before You Leap - Mira antes de saltar):** Consiste en verificar explícitamente las precondiciones. Por ejemplo, comprobar si un archivo existe antes de intentar abrirlo.
- **EAFP (Easier to Ask for Forgiveness than Permission - Es más fácil pedir perdón que permiso):** Consiste en ejecutar el código directamente (`try`) y capturar las excepciones (`except`) que puedan surgir. **Este es el estilo preferido y más idiomático en Python**, ya que es más limpio y evita problemas de concurrencia (condiciones de carrera).

````python
# Estilo EAFP (Pythonic)
mi_dict = {"clave": "valor"}

try:
    print(mi_dict["clave_inexistente"])
except KeyError:
    print("Error: La clave no fue encontrada. Se usará un valor por defecto.")
````

La estructura `try...except` puede tener bloques adicionales:
- `else`: Se ejecuta si el bloque `try` **no** lanza ninguna excepción.
- `finally`: Se ejecuta **siempre**, haya o no haya excepción. Es ideal para tareas de limpieza, como cerrar archivos o conexiones.

````python
try:
    resultado = 10 / 2
except ZeroDivisionError:
    print("No se puede dividir por cero.")
else:
    # Esto se ejecutará porque no hubo error
    print(f"El resultado es: {resultado}")
finally:
    # Esto se ejecutará siempre
    print("Fin del bloque de manejo de excepciones.")
````

## 2. Comprehensions

Son una forma concisa y legible de crear colecciones (listas, diccionarios, conjuntos) a partir de otras secuencias. A menudo son más rápidas que usar bucles `for` explícitos.

````python
# List Comprehension: crear una lista de cuadrados
cuadrados = [x**2 for x in range(10)]
print(f"List comprehension: {cuadrados}")

# Set Comprehension: crear un conjunto de letras únicas
palabra = "programacion"
letras_unicas = {letra for letra in palabra}
print(f"Set comprehension: {letras_unicas}")

# Dict Comprehension: crear un diccionario de números y sus cuadrados
dict_cuadrados = {x: x**2 for x in range(5)}
print(f"Dict comprehension: {dict_cuadrados}")
````

## 3. Generadores

Los generadores permiten crear iteradores de una manera sencilla. A diferencia de las listas, los generadores producen valores 'perezosamente' (lazy evaluation), uno a la vez y solo cuando se solicitan. Esto los hace extremadamente eficientes en el uso de memoria, especialmente con secuencias muy grandes.

### Expresiones Generadoras
Tienen una sintaxis similar a las list comprehensions, pero usando paréntesis.

````python
import sys

# Una lista ocupa memoria para todos sus elementos a la vez
lista_grande = [i for i in range(10000)]
print(f"Tamaño en memoria de la lista: {sys.getsizeof(lista_grande)} bytes")

# Un generador solo ocupa una pequeña cantidad de memoria constante
generador_grande = (i for i in range(10000))
print(f"Tamaño en memoria del generador: {sys.getsizeof(generador_grande)} bytes")

# Iteramos sobre el generador para obtener sus valores
suma_total = sum(generador_grande)
print(f"La suma de los valores del generador es: {suma_total}")
````

### Funciones Generadoras
Usan la palabra clave `yield` para devolver un valor y pausar su estado hasta la próxima llamada.

````python
def contador_infinito():
    num = 0
    while True:
        yield num
        num += 1

gen = contador_infinito()

print(next(gen)) # 0
print(next(gen)) # 1
print(next(gen)) # 2
````

## 4. Funciones Lambda

Una función lambda es una pequeña función anónima (sin nombre). Se define con la palabra clave `lambda` y está restringida a una sola expresión.

Son útiles cuando necesitas una función simple por un corto período de tiempo, comúnmente como argumento de una función de orden superior (ej. `sorted`, `map`, `filter`).

````python
# Lista de tuplas (nombre, edad)
personas = [("Ana", 28), ("Juan", 22), ("Beatriz", 35)]

# Usamos una lambda como 'key' para ordenar por el segundo elemento (edad)
personas_ordenadas = sorted(personas, key=lambda persona: persona[1])

print(f"Personas ordenadas por edad: {personas_ordenadas}")
````

## 5. Estructura de Proyectos: Módulos y Paquetes

- **Módulo:** Un archivo `.py` es un módulo. Su nombre es el nombre del archivo.
- **Paquete:** Una carpeta que contiene módulos. Para que Python la trate como un paquete, debe contener un archivo `__init__.py` (que puede estar vacío). Este archivo se ejecuta cuando se importa el paquete.

Una estructura de proyecto típica se ve así:

```
mi_proyecto/
├── mi_paquete/
│   ├── __init__.py
│   ├── modulo_calculos.py
│   └── modulo_utilidades.py
├── tests/
│   ├── test_calculos.py
│   └── test_utilidades.py
└── main.py
```

Desde `main.py`, podrías importar y usar el código así:

````python
# En main.py
# from mi_paquete.modulo_calculos import sumar
# from mi_paquete.modulo_utilidades import formatear_fecha

# print(sumar(5, 10))
# print(formatear_fecha(...))
````

## 6. Distribución de Paquetes (Estándar 2025)

Para que otros puedan instalar y usar tu código con `pip install`, necesitas empaquetarlo. El enfoque moderno utiliza un archivo `pyproject.toml` para definir la configuración del proyecto y las dependencias de construcción.

Este archivo reemplaza al antiguo `setup.py` para la configuración de metadatos del proyecto.

Un `pyproject.toml` mínimo se ve así:

```toml
# pyproject.toml

[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

[project]
name = "mi-paquete-genial"
version = "0.1.0"
authors = [
  { name="Tu Nombre", email="tu@email.com" },
]
description = "Un paquete de ejemplo para demostrar la distribución moderna."
requires-python = ">=3.9"
classifiers = [
    "Programming Language :: Python :: 3",
    "License :: OSI Approved :: MIT License",
    "Operating System :: OS Independent",
]
```
