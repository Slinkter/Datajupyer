## 1. Manejo de Excepciones

Python usa excepciones para gestionar errores durante la ejecución.

### Filosofías: LBYL vs. EAFP

- **LBYL (Look Before You Leap):** Verificar condiciones antes de ejecutar.
- **EAFP (Easier to Ask for Forgiveness than Permission):** Ejecutar y capturar errores. **Estilo preferido en Python.**

````python
# Estilo EAFP (Pythonic)
mi_dict = {"clave": "valor"}

try:
    print(mi_dict["clave_inexistente"])
except KeyError:
    print("Error: La clave no fue encontrada. Se usará un valor por defecto.")
````

La estructura `try...except` puede tener bloques adicionales:
- `else`: Se ejecuta si el bloque `try` **no** lanza ninguna excepción.
- `finally`: Se ejecuta **siempre**, ideal para cerrar archivos o conexiones.

````python
try:
    resultado = 10 / 2
except ZeroDivisionError:
    print("No se puede dividir por cero.")
else:
    print(f"El resultado es: {resultado}")
finally:
    print("Fin del bloque de manejo de excepciones.")
````

## 2. Comprehensions

Forma concisa de crear colecciones a partir de otras secuencias. Más rápidas que bucles `for` explícitos.

````python
# List Comprehension
cuadrados = [x**2 for x in range(10)]
print(f"List comprehension: {cuadrados}")

# Set Comprehension
palabra = "programacion"
letras_unicas = {letra for letra in palabra}
print(f"Set comprehension: {letras_unicas}")

# Dict Comprehension
dict_cuadrados = {x: x**2 for x in range(5)}
print(f"Dict comprehension: {dict_cuadrados}")
````

## 3. Generadores

Producen valores uno a la vez (lazy evaluation). Muy eficientes en memoria con secuencias grandes.

### Expresiones Generadoras
Sintaxis similar a list comprehensions pero con paréntesis.

````python
import sys

# Lista: ocupa memoria para todos los elementos a la vez
lista_grande = [i for i in range(10000)]
print(f"Tamaño en memoria de la lista: {sys.getsizeof(lista_grande)} bytes")

# Generador: ocupa una cantidad constante de memoria
generador_grande = (i for i in range(10000))
print(f"Tamaño en memoria del generador: {sys.getsizeof(generador_grande)} bytes")

suma_total = sum(generador_grande)
print(f"La suma de los valores del generador es: {suma_total}")
````

### Funciones Generadoras
Usan `yield` para devolver un valor y pausar su estado hasta la próxima llamada.

````python
def contador_infinito():
    num = 0
    while True:
        yield num
        num += 1

gen = contador_infinito()

print(next(gen))  # 0
print(next(gen))  # 1
print(next(gen))  # 2
````

## 4. Funciones Lambda

Función anónima de una sola expresión. Útil como argumento de `sorted`, `map`, `filter`.

````python
# Lista de tuplas (nombre, edad)
personas = [("Ana", 28), ("Juan", 22), ("Beatriz", 35)]

# Lambda como key para ordenar por edad
personas_ordenadas = sorted(personas, key=lambda persona: persona[1])

print(f"Personas ordenadas por edad: {personas_ordenadas}")
````
